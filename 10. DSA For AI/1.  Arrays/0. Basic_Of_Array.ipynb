# Array in Python

## Introduction

An **Array** is a data structure used to store multiple values of the **same data type** in a single variable. Arrays store elements in **contiguous memory locations**, and each element can be accessed using its **index**.

In Python, the most commonly used collection is a **List**, but Python also provides an **array module** for creating arrays.

---

# Why Use Arrays?

Without arrays:

```python
a = 10
b = 20
c = 30
d = 40
e = 50
```

Using an array:

```python
from array import *

arr = array('i', [10, 20, 30, 40, 50])


print(type(arr))
print(arr[3])
```


Here all values are stored in one variable.

---

# Features of Arrays

1. Store multiple elements in one variable.
2. Elements are stored sequentially.
3. Indexing starts from 0.
4. Fast access to elements.
5. All elements must be of the same data type.

---



# Creating an Array

First import the array module:



```python
## First way to import the array module..
import array

arr = array.array('i', [10,20,30,40,50,60])   # array.array represents we are using array function from array module...

print(type(arr))
print(arr)

print(arr[4])
```

```python
## Second way to import the array module..
from array import *  # this line includes * (everything) from array module..

arr = array('i',[10,20,30,40,50,60,70])

print(type(arr))
print(arr[6])
```

### Integer Array

```python

arr = array('i',[10,20,30,40,50,60,70])  ## 'i' represents the type code: 'int' / data type: 'integer'

print(type(arr))
print(arr[6])
```

# Accessing Array Elements

Array elements are accessed using indexes.

```python

arr = array('i', [10, 20, 30, 40, 50])

print(arr[0])
print(arr[2])
```


---

# Traversing an Array

```python
arr = array('i',[10,20,30,40,50,60,70])

for i in arr:
    print(i)
```

---

# Array Operations

## 1. Insertion

### Append Element

```python
lst = [60,70,80,90]
print(type(lst))

arr = array('i', [10,20,30,40,50])
print(f"Before appending: {arr}")

for num in lst:
    arr.append(num)

print(f"After appending: {arr}")
```

---

### Insert at Specific Position

```python
arr = array('i', [10,20,40,50,60])

arr.insert(2,30)

print(arr)
```

---

## 2. Deletion

### Remove Element

```python
arr = array('i', [10,20,40,50,60])

arr.remove(60)
print(arr)
```


---

### Delete Using Pop

```python
arr = array('i', [10,20,30,40,50])

arr.pop()
print(arr)
```

## 3. Updating Elements

```python
arr = array('i', [10,2,30,40,50,60])
print(f"Array before updating: {arr}")
arr[1] = 20
print(f"Array after updating: {arr}")
```


---

## 4. Searching Elements

```python
arr = array('i', [10, 20, 30, 40, 50, 60])

print(arr.index(50))
```


---

# Array Slicing


```python
arr = array('i', [10,20,30,40,50,60,70,80,90])

print(arr[0::])
print(arr[0:4])
print(arr[0::2])
print(arr[::-1])
```

---

# Length of Array

```python
print(len(arr))
```


---

# Reverse an Array

```python
arr = array('i', [10, 20, 30, 40, 50, 60, 70, 80, 90])

arr.reverse()

print(arr)
```



---

# Advantages of Arrays

1. Fast access using indexes.
2. Efficient memory usage.
3. Easy traversal of elements.
4. Suitable for storing large amounts of similar data.

---

# Disadvantages of Arrays

1. Fixed data type.
2. Insertion and deletion can be costly.
3. Size management is less flexible than Python lists.
4. Cannot store different data types together.

---

# Time Complexity of Array Operations

| Operation       | Complexity |
| --------------- | ---------- |
| Access          | O(1)       |
| Update          | O(1)       |
| Search          | O(n)       |
| Insert (end)    | O(1)       |
| Insert (middle) | O(n)       |
| Delete          | O(n)       |
| Traverse        | O(n)       |
