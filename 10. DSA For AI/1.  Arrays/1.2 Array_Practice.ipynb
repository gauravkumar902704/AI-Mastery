Ek array diya gaya hai. Tumhe uske elements ka order ulta (reverse) karna hai.


Array ka first element last ban jayega.

Second element second-last ban jayega.

Aur aise hi last element first ban jayega.

Visualize karo:

```text
Index

0   1   2   3   4

10  20  30  40  50

↓

50  40  30  20  10
```

---

##### Real-life Example

Jaise ek line me log khade hain.

```text
A B C D E
```

Sab log opposite direction me khade ho gaye.

```text
E D C B A
```

---

##### Main Idea

Array ke dono ends se elements ko swap karte jao.

Jab middle tak pahunch jao, reverse complete ho jayega.

````python
def Reverse_array(arr):
    size = len(arr)
    left = 0
    right = size - 1

    while right > left:
        arr[left],arr[right] = arr[right], arr[left]
        left += 1
        right -= 1



arr = array('i', [10, 20, 30, 40, 50, 60])
print("Array before reversing: ",arr)
Reverse_array(arr)
print("Array after reversing: ",arr)
````

### **Ques.4. Identify elements with Highest and Lowest Frequency**




---

##### Problem Statement

---

##### Real-life Example

Class attendance.

Roll numbers

```text

```

Sabse jyada attendance

Roll 5

Sabse kam

Roll 2

---

##### Main Idea

Har unique element ki occurrence count karo.

Fir maximum aur minimum frequency compare karo.

````python
# Identify elements with Highest and Lowest Frequency
def Highest_and_Lowest_Frequency(arr):

    if len(arr) == 0:
        print("Given array is empty.")
        return

    # Count the frequency of elements
    frequency = {}

    for value in arr:
        if value in frequency:
            frequency[value] += 1
        else:
            frequency[value] = 1

    # Find maximum and minimum frequency
    max_frequency = 0
    low_frequency = float('inf')

    for value, freq in frequency.items():

        if freq > max_frequency:
            max_frequency = freq

        if freq < low_frequency:
            low_frequency = freq

    # Collect all elements having max/min frequency
    max_elements = []
    low_elements = []

    for value, freq in frequency.items():
        if freq == max_frequency:
            max_elements.append(value)
        if freq == low_frequency:
            low_elements.append(value)


    
    # Step 4: Print Result
    print("Frequency Dictionary :", frequency)
    print()

    print(f"Highest Frequency Elements : {max_elements}")
    print(f"Highest Frequency          : {max_frequency}")

    print()

    print(f"Lowest Frequency Elements  : {low_elements}")
    print(f"Lowest Frequency           : {low_frequency}")

````

````python
# Test Cases
arr = [2, 5, 1, 3, 2, 4, 5, 6, 7, 8, 9]
arr1 = [1, 4, 5, 7, 5, 2, 4, 6, 2, 4, 8, 9, 5, 4]
arr2 = [1, 1, 2, 2, 3, 4]
arr3 = []
arr4 = [5, 5, 5, 5]
arr5 = [1, 2, 3, 4]


Highest_and_Lowest_Frequency(arr5)
````

````python
# Find the mode of the array...
def find_Mode(arr):
    if len(arr) == 0:
        print("Given Array is Empty.")
        return
    
    # Calculating the frequencies of elements...
    frequencies = {}
    for value in arr:
        if value in frequencies:
            frequencies[value] += 1
        else:
            frequencies[value] = 1

    # Find out the Maximum Frequency element...
    max_frequency = 0
    for value, freq in frequencies.items():
        if freq > max_frequency:
            max_frequency = freq

    # Find out the Mode elements...
    mode_element = []
    for value, freq in frequencies.items():
        if freq  == max_frequency:
            mode_element.append(value)

    # Print Result
    print("Frequency Dictionary :", frequencies)
    print()

    if len(mode_element) == len(frequencies) and len(frequencies) > 1:
        print("No Mode")
    else:
        print("Mode :", mode_element)
````

## 1. Missing Number

**Problem:** Array me `0` se `n` tak ke numbers hone chahiye, ek number missing hai.

**Example:**

```text
[0,1,3,4]
Missing = 2
```

**Hint:** Socho agar koi number missing na hota to total sum kitna hota? Actual sum se compare karo. Dusra powerful idea: XOR.

````python
def missing_number(arr):
    # Sum of n number's..
    sum_of_numbers = 0
    for i in range(len(arr) + 1):
        sum_of_numbers += i

    # Sum of array numbers...
    sum_of_arr_numbers = 0
    for num in arr:
        sum_of_arr_numbers += num

    number = sum_of_numbers - sum_of_arr_numbers

    return number


arr = [0,1,2,3,5,6]
result = missing_number(arr)
print(result)
    
    
````

````python
def missing_number(arr):
    xor = len(arr)
    for i in range(len(arr)):
        xor ^= i
        xor ^= arr[i]

    return xor


arr = [0, 1, 2, 3, 5, 6]
result = missing_number(arr)
print(result)
````

````python
def missing_number(arr):
    expected_xor = 0
    array_xor = 0

    for i in range(len(arr) + 1):
        expected_xor ^= i

    for num in arr:
        array_xor ^= num

    return expected_xor ^ array_xor


arr = [0, 1, 2, 3, 5, 6]
print(missing_number(arr))
````

## 2. Find the Unique Element

**Problem:** Sab elements 2 baar aaye hain, sirf ek element 1 baar aaya hai.

**Example:**

```text
[2,3,2,4,4]
Answer = 3
```

**Hint:** Frequency map bana sakte ho. Agar condition hai ki sab do-do baar aaye hain, to XOR ka concept bhi bahut useful hai.

````python
def Unique_Element(arr):
    frequencies = {}
    for value in arr:
        if value in frequencies:
            frequencies[value] += 1
        else:
            frequencies[value] = 1

    low_Frequency = float('inf')
    unique_number = None
    for value, freq in frequencies.items():
        if freq < low_Frequency:
            low_Frequency = freq
            unique_number = value

    return unique_number
````

````python
arr = [1,2,1,4,5,2,4,6,6]
result = Unique_Element(arr)
print(result)
````

## 3. Two Sum

**Problem:** Do numbers find karo jinka sum target ke barabar ho.

**Example:**

```text
[2,7,11,15]
Target = 9
Answer = 2 + 7
```

**Hint:** Har element ke liye socho, "Target banane ke liye mujhe aur kya chahiye?" (Target − Current). Us value ko fast search karna hi main trick hai.


````python
def Two_sum(arr, target):
    i = [0]
    size = len(arr) 
    j = size - 1

    while(i < j):
        if (arr[i] + arr[j]) == target:
            return [arr[i],arr[j]]

````
