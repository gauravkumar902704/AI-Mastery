````python
def linearSearch(arr, target):

    print(arr)
    for i,value in enumerate(arr):
        if value == target:
            return f"{target} is found at index: {i}"
        
    return f"{target} is not present Entier array."
        
arr3 = array('i', map(int, input("Enter the elements of array by separating them 'space'. ").split()))
target = int(input("Enter the element to be search."))

result = linearSearch(arr3, target)
print(result)
````

### **Ques.1. Reverse an Array.**


---

##### Problem Statement

Ek array diya gaya hai. Tumhe uske elements ka order ulta (reverse) karna hai.

##### Example

```text
Original:
[10, 20, 30, 40, 50]

Reverse:
[50, 40, 30, 20, 10]
```

---

##### Samajhne ki baat

Array ka first element last ban jayega.

Second element second-last ban jayega.

Aur aise hi last element first ban jayega.

Visualize karo:

```text
Index

0   1   2   3   4

10  20  30  40  50

↓

50  40  30  20  10
```

---

##### Real-life Example

Jaise ek line me log khade hain.

```text
A B C D E
```

Sab log opposite direction me khade ho gaye.

```text
E D C B A
```

---

##### Main Idea

Array ke dono ends se elements ko swap karte jao.

Jab middle tak pahunch jao, reverse complete ho jayega.

````python
def Reverse_array(arr):
    size = len(arr)
    left = 0
    right = size - 1

    while right > left:
        arr[left],arr[right] = arr[right], arr[left]
        left += 1
        right -= 1



arr = array('i', [10, 20, 30, 40, 50, 60])
print("Array before reversing: ",arr)
Reverse_array(arr)
print("Array after reversing: ",arr)
````

### **Ques.4. Identify elements with Highest and Lowest Frequency**




---

##### Problem Statement

Array me har element kitni baar aaya hai ye count karo.

Fir batao

* Sabse jyada baar kaunsa element aaya
* Sabse kam baar kaunsa element aaya

---

##### Example

```text
Array

[2,3,5,2,3,2,4]
```

Frequency

```text
2 → 3 times

3 → 2 times

5 → 1 time

4 → 1 time
```

Highest Frequency

```text
2
```

Lowest Frequency

```text
5 and 4
```

Dono sirf ek baar aaye.

---

Another Example


Frequency

```text
1 → 3

2 → 2

3 → 1
```

Highest

```text
1
```

```text
3
```

---

##### Real-life Example

Class attendance.

Roll numbers

```text
5

7

5

2

7

5
```

Sabse jyada attendance

Roll 5

Sabse kam

Roll 2

---

##### Main Idea

Har unique element ki occurrence count karo.

Fir maximum aur minimum frequency compare karo.

````python
# Identify elements with Highest and Lowest Frequency
def Highest_and_Lowest_Frequency(arr):

    if len(arr) == 0:
        print("Given array is empty.")
        return

    # Count the frequency of elements
    frequency = {}

    for value in arr:
        if value in frequency:
            frequency[value] += 1
        else:
            frequency[value] = 1

    # Find maximum and minimum frequency
    max_frequency = 0
    low_frequency = float('inf')

    for value, freq in frequency.items():

        if freq > max_frequency:
            max_frequency = freq

        if freq < low_frequency:
            low_frequency = freq

    # Collect all elements having max/min frequency
    max_elements = []
    low_elements = []

    for value, freq in frequency.items():
        if freq == max_frequency:
            max_elements.append(value)
        if freq == low_frequency:
            low_elements.append(value)


    
    # Step 4: Print Result
    print("Frequency Dictionary :", frequency)
    print()

    print(f"Highest Frequency Elements : {max_elements}")
    print(f"Highest Frequency          : {max_frequency}")

    print()

    print(f"Lowest Frequency Elements  : {low_elements}")
    print(f"Lowest Frequency           : {low_frequency}")

````

````python
# Test Cases
arr = [2, 5, 1, 3, 2, 4, 5, 6, 7, 8, 9]
arr1 = [1, 4, 5, 7, 5, 2, 4, 6, 2, 4, 8, 9, 5, 4]
arr2 = [1, 1, 2, 2, 3, 4]
arr3 = []
arr4 = [5, 5, 5, 5]
arr5 = [1, 2, 3, 4]


Highest_and_Lowest_Frequency(arr5)
````

---

##### Real-life Example

Class me favourite color survey hua.

```text

Red

Green

Black
```

Blue sabse jyada students ne choose kiya.

Isliye

Mode = Blue

````python
# Find the mode of the array...
def find_Mode(arr):
    if len(arr) == 0:
        print("Given Array is Empty.")
        return
    
    # Calculating the frequencies of elements...
    frequencies = {}
    for value in arr:
        if value in frequencies:
            frequencies[value] += 1
        else:
            frequencies[value] = 1

    # Find out the Maximum Frequency element...
    max_frequency = 0
    for value, freq in frequencies.items():
        if freq > max_frequency:
            max_frequency = freq

    # Find out the Mode elements...
    mode_element = []
    for value, freq in frequencies.items():
        if freq  == max_frequency:
            mode_element.append(value)

    # Print Result
    print("Frequency Dictionary :", frequencies)
    print()

    if len(mode_element) == len(frequencies) and len(frequencies) > 1:
        print("No Mode")
    else:
        print("Mode :", mode_element)
````

## Difference Between Frequency Question and Mode

| Question                   | Kya Find Karna Hai?                                                                                           |
| -------------------------- | ------------------------------------------------------------------------------------------------------------- |
| Highest & Lowest Frequency | Dono highest aur lowest frequency wale elements identify karne hain. Ek ya multiple answers ho sakte hain.    |
| Mode                       | Sirf highest frequency wala element (ya highest frequency share karne wale multiple elements) find karna hai. |
