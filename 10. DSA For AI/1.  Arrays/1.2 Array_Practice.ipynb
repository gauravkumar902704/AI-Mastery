### **Ques.1. Find the average of array elements.**

```python
"""        Sum of all numbers 
Average = ---------------------
        n(Total number of numbers)
"""
arr1 = array('i', map(int , input("Enter the Elements of array by separating 'Space'. ").split()))

sum = 0
for value in arr1:
    sum += value

average = sum / (len(arr1))

print("Average of Array Elements: ",average)
```

### **Ques.2. Multiply each elements of Array by 10.**

```python
arr2 = array('i', map(int, input("Enter the Elements of array by separating 'Space'. ").split()))

print("Array before multiplying of 10. ",arr2)

for i, value in enumerate(arr2):
    arr2[i] = value * 10

print("Array after multiplying of 10. ",arr2)
```

### **Ques.3. Search for an element in an Array {Linear Search}.**

```python
def linearSearch(arr, target):

    print(arr)
    for i,value in enumerate(arr):
        if value == target:
            return f"{target} is found at index: {i}"
        
    return f"{target} is not present Entier array."
        
arr3 = array('i', map(int, input("Enter the elements of array by separating them 'space'. ").split()))
target = int(input("Enter the element to be search."))

result = linearSearch(arr3, target)
print(result)
```

### **Ques.4. Find the maximum element in an array.**

```python
def Max(arr4):
    large = 0
    for i in range(len(arr4)-1):
        if arr4[i] > large:
            large = arr4[i]
    return large
        
arr4 = array('i', map(int, input("Enter the array elements by separating them space. ").split()))
result = Max(arr4)
print(f"Max: {result}")
```

### **Ques.5. Return Sum of +ve and -ve numbers.**

```python
def sum_positive_or_negative_nums(arr):
    Sum_of_Positive = 0
    Sum_of_negative = 0
    print(arr)
    for value in arr:
        if value > 0:
            Sum_of_Positive += value
        elif value < 0:
            Sum_of_negative += value
        else:
            continue
    return f"Sum of Positive Numbers: {Sum_of_Positive}\nSum of Negative Numbers: {Sum_of_negative}"

arr5 = array('i', map(int, input("Enter the array elements by separating them space. ").split()))
result = sum_positive_or_negative_nums(arr5)
print(result)
    
```

### **Ques.6. Count the number of Zeroes and Onces.**

```python
def Number_of_Zeroes_and_Onces(arr):
    Zeroes = 0
    Onces = 0
    print(arr)
    for value in arr:
        if value == 0:
            Zeroes += 1
        elif value == 1:
            Onces += 1
    return f"Zeroes: {Zeroes}\nOnces: {Onces}"

arr5 = array('i', map(int, input("Enter the array elements by separating them space. ").split()))
result = Number_of_Zeroes_and_Onces(arr5)
print(result)
```
