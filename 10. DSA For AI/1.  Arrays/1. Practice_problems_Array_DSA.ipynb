## 1. Find the Sum of All Elements

### Question

Given an array of integers, find the sum of all elements.

### Example

```python
Input:
[10, 20, 30, 40]

Output:
100
```

### Understanding

Imagine you have marks of students:

```python
[10, 20, 30, 40]
```

You need total marks.

Process:

```python
sum = 0

sum += 10
sum += 20
sum += 30
sum += 40
```

Final answer:

```python
100
```

### Concept Used

* Traversal
* Accumulator Pattern

### Complexity

```python
O(n)
```



---

## 2. Find Maximum Element

### Question

Find the largest element in an array.

### Example

```python
Input:
[4, 2, 9, 1, 7]

Output:
9
```

### Understanding

Assume first element is maximum.

```python
max = 4
```

Now compare with every element.

```python
4 vs 2
4 vs 9 → update max
9 vs 1
9 vs 7
```

Answer:

```python
9
```

### Concept

Finding optimal value while traversing.

## 11. Find Missing Number

### Question

Numbers from 1 to n exist.

One number is missing.

### Example

```python
Input:
[1,2,3,5]

Output:
4
```

### Understanding

Expected Sum:

```python
n*(n+1)/2
```

Actual Sum:

```python
1+2+3+5
```

Difference = Missing Number

### Complexity

```python
O(n)
```
