### Data Structures and Algorithms (DSA) – 

**Data Structure**: A way of organizing and storing data so it can be accessed and modified efficiently.

Examples:

* **Array** – stores elements in a sequence.
* **Linked List** – stores elements connected by pointers.
* **Stack** – follows LIFO (Last In, First Out).
* **Queue** – follows FIFO (First In, First Out).
* **Tree** – hierarchical structure.
* **Graph** – represents networks and relationships.
* **Hash Table** – stores key-value pairs for fast lookup.

**Algorithm**: A step-by-step procedure or set of instructions used to solve a problem or perform a task.

Examples:

* Searching (Linear Search, Binary Search)
* Sorting (Bubble Sort, Merge Sort, Quick Sort)
* Finding shortest paths in graphs
* Solving optimization problems

### Simple Definition

**DSA is the study of how data is organized (Data Structures) and how problems are solved efficiently (Algorithms).**

### Example

Suppose you have a phone contacts list:

* **Data Structure**: How contacts are stored (array, hash table, etc.).
* **Algorithm**: How you find a contact (linear search, binary search, etc.).

### Why Learn DSA?

* Improves problem-solving skills.
* Helps write efficient code.
* Important for coding interviews and placements.
* Used in software development, AI/ML, databases, operating systems, and web applications.




### What is Efficiency in Programming?

**Efficiency** means how well a program uses **time** and **memory** to complete a task.

A program is considered efficient if it:

* Runs **faster** (uses less execution time).
* Uses **less memory** (RAM).

### Types of Efficiency

#### 1. Time Efficiency

Measures how long a program takes to execute.

Example:

````python
import time
# O(n)
start = time.time()
n = 10
for i in range(n):
    print(i)
print(time.time()-start)
````

This loop runs `n` times.

````python
# O(n²)
n = 5
for i in range(n):
    for j in range(n):
        print(i, j, end=" | ")
    print('')
````

This runs `n × n` times, so it is slower for large `n`.

#### 2. Space Efficiency

Measures how much memory a program uses.

Example:

````python
numbers = [1, 2, 3, 4, 5]
````

This list requires memory to store all elements.

````python
for i in range(1, 6):
    print(i)
````


This uses less memory because no list is stored.



### Why is Efficiency Important?

Imagine searching for a name in:

* A list of **10 contacts** → almost instant.
* A list of **10 million contacts** → efficiency becomes crucial.

Efficient programs:

* Run faster.
* Handle large amounts of data.
* Save system resources.
* Provide better user experience.

### How Do We Measure Efficiency?

Using **Big O Notation**:

| Complexity | Name         | Performance         |
| ---------- | ------------ | ------------------- |
| O(1)       | Constant     | Excellent           |
| O(log n)   | Logarithmic  | Very Good           |
| O(n)       | Linear       | Good                |
| O(n log n) | Linearithmic | Good                |
| O(n²)      | Quadratic    | Slow for large data |
| O(2ⁿ)      | Exponential  | Very Slow           |

### Simple Definition

👉 **Efficiency in programming is the ability of a program to solve a problem using the least possible time and memory.**

# Asymptotic Notations

**Asymptotic Notations** are mathematical tools used to describe the **time complexity** and **space complexity** of an algorithm as the input size (**n**) becomes very large.

They help us compare algorithms without depending on the computer's speed, programming language, or hardware.

---

## Why Do We Need Asymptotic Notations?

Suppose two algorithms perform the same task:

### Algorithm A

```python
for i in range(n):
    print(i)
```

Time complexity: **O(n)**

### Algorithm B

```python
for i in range(n):
    for j in range(n):
        print(i, j)
```

Time complexity: **O(n²)**

For small values of `n`, both may seem fast. But for large values of `n`, Algorithm A is much more efficient.

Asymptotic notation helps us analyze this growth.

---

# Types of Asymptotic Notations

There are three main types:

1. **Big O Notation (O)**
2. **Omega Notation (Ω)**
3. **Theta Notation (Θ)**

---

# 1. Big O Notation (O)

### Definition

Big O represents the **upper bound** or **worst-case complexity** of an algorithm.

It tells us the maximum time an algorithm can take.

### Example

```python
for i in range(n):
    print(i)
```

Number of operations ≈ n

**Complexity = O(n)**

### Real-Life Example

Suppose you search for a student in a class register.

In the worst case, the student is at the last position.

You may need to check all students.

**Worst Case = O(n)**

### Graphical Meaning

```
Actual Time
     /
    /
   /
  /____ Big O
```

Big O gives an upper limit.

### Common Big O Complexities

| Complexity | Name         |
| ---------- | ------------ |
| O(1)       | Constant     |
| O(log n)   | Logarithmic  |
| O(n)       | Linear       |
| O(n log n) | Linearithmic |
| O(n²)      | Quadratic    |
| O(n³)      | Cubic        |
| O(2ⁿ)      | Exponential  |
| O(n!)      | Factorial    |

---

# 2. Omega Notation (Ω)

### Definition

Omega represents the **lower bound** or **best-case complexity**.

It tells us the minimum time required by an algorithm.

### Example

Linear Search:

```python
arr = [10,20,30,40,50]
```

Searching for `10`.

Since it is at the first position:

Only one comparison is needed.

**Best Case = Ω(1)**

### Real-Life Example

Finding a book in a library.

If the book is right in front of you:

Time required is minimum.

**Best Case = Ω(1)**

### Graphical Meaning

```
Omega
____
    \
     \
      \
Actual Time
```

Omega gives a lower limit.

---

# 3. Theta Notation (Θ)

### Definition

Theta represents the **tight bound**.

It gives both:

* Upper Bound
* Lower Bound

When an algorithm's growth rate is exactly known, we use Theta.

### Example

```python
for i in range(n):
    print(i)
```

The loop always runs `n` times.

Best Case = n

Worst Case = n

Therefore:

**Θ(n)**

### Real-Life Example

Counting all students in a classroom.

Whether students are seated differently or not, you must count every student.

Time always depends on `n`.

Hence:

**Θ(n)**

### Graphical Meaning

```
Upper Bound
   /
  /
 / Actual Function
/
\ 
 \ Lower Bound
```

The actual function lies between both bounds.

---

# Comparison of O, Ω and Θ

| Notation         | Meaning                | Case                 |
| ---------------- | ---------------------- | -------------------- |
| O(f(n))          | Upper Bound            | Worst Case           |
| Ω(f(n))          | Lower Bound            | Best Case            |
| Θ(f(n))          | Tight Bound            | Average/Exact Growth |
| ---------------- | ------------------     |                      |
| O(n²)            | At most n²             |                      |
| Ω(n)             | At least n             |                      |
| Θ(n log n)       | Exactly n log n growth |                      |

---

# Example: Linear Search

```python
def linear_search(arr, key):
    for i in range(len(arr)):
        if arr[i] == key:
            return i
    return -1
```

### Best Case

Element found at first position.

```
Ω(1)
```

### Worst Case

Element found at last position or not found.

```
O(n)
```

### Average Case

```
Θ(n)
```

---

# Example: Binary Search

```python
def binary_search(arr, key):
    low = 0
    high = len(arr)-1

    while low <= high:
        mid = (low+high)//2

        if arr[mid] == key:
            return mid
        elif arr[mid] < key:
            low = mid+1
        else:
            high = mid-1
```

### Complexity

Best Case:

```
Ω(1)
```

Worst Case:

```
O(log n)
```

Average Case:

```
Θ(log n)
```

---

# Key Points for Exams

1. Asymptotic notations analyze algorithm performance for large input sizes.
2. They ignore constants and lower-order terms.
3. Big O represents the worst-case complexity.
4. Omega represents the best-case complexity.
5. Theta represents the exact or tight-bound complexity.
6. They help compare algorithms independently of hardware and programming language.

### One-Line Definition

👉 **Asymptotic Notations are mathematical representations used to describe the growth rate and efficiency of algorithms as input size increases.**


# Different Types of Time Complexity

**Time Complexity** measures how the running time of an algorithm grows as the input size (**n**) increases.

---

## 1. Constant Time – O(1)

The execution time remains the same regardless of input size.

### Example

````python
arr = [10, 20, 30, 40, 50]
print(arr[2])
````


Accessing an element by index always takes one step.

### Complexity

```text
O(1)
```

### Real-Life Example

Looking up a word in a dictionary using a page number.

---

## 2. Logarithmic Time – O(log n)

The input size is reduced by a fixed factor in each step.

### Example: Binary Search

````python
while low <= high:
    mid = (low + high) // 2
````


Each step eliminates half of the remaining elements.

### Complexity

```text
O(log n)
```

### Real-Life Example

Finding a name in a sorted phone book.

---

---

## 3. Linear Time – O(n)

The running time grows directly with the input size.

### Example


````python
n = 5
for i in range(n):
    print(i)
````


If n doubles, the work doubles.

### Complexity

```text
O(n)
```

### Real-Life Example

Checking every student in a class attendance list.

---


## 4. Linearithmic Time – O(n log n)

Combination of linear and logarithmic growth.

### Example

Merge Sort and Heap Sort.

```python
# Merge Sort
T(n) = n log n
```

### Complexity

```text
O(n log n)
```

### Real-Life Example

Efficiently sorting a large list of records.

---

---

## 5. Quadratic Time – O(n²)

Two nested loops over the same input.

### Example

````python
n = 5
for i in range(n):
    for j in range(n):
        print(i, j)
````


Total operations = n × n.

### Complexity

```text
O(n²)
```

### Real-Life Example

Comparing every student with every other student.

---


## 6. Cubic Time – O(n³)

Three nested loops.

### Example

````python
n = 5
for i in range(n):
    for j in range(n):
        for k in range(n):
            print(i, j, k)
````

### Complexity

```text
O(n³)
```

### Real-Life Example

Some matrix multiplication algorithms.

---

## 7. Exponential Time – O(2ⁿ)

The running time doubles with every additional input.

### Example

Recursive Fibonacci:

````python
def fib(n):
    if n <= 1:
        return n
    return fib(n-1) + fib(n-2)

print(fib(12))
````


### Complexity

```text
O(2ⁿ)
```

### Real-Life Example

Trying every possible combination of choices.

---

## 8. Factorial Time – O(n!)

The most expensive common complexity.

### Example

Generating all permutations.

````python
from itertools import permutations
````


### Complexity

```text
O(n!)
```

### Real-Life Example

Finding all possible arrangements of people in a queue.
