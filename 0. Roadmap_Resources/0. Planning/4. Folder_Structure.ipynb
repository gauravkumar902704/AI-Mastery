1. Theory पढ़ो
2. Notes बनाओ
3. Code लिखो
4. Mini project बनाओ
5. GitHub पर upload करो
6. Revision folder update करो

यह structure तुम्हें:

* AI mastery
* Semester + placement prep
* Interview prep
* Projects portfolio
* Research learning

सब एक ही जगह manage करने में मदद करेगा।
