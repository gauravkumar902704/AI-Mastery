# Importing Modules in Python: Modules and Packages

Modules and Packages are very important concepts in Python because they help us organize code and reuse functionality.

---

# 1. What is a Module?

A **module** is simply a Python file (`.py`) that contains:

* Functions
* Variables
* Classes
* Statements

Example:

Create a file named `calculator.py`


````python
def add(a, b):
    return a + b

def subtract(a, b):
    return a - b
````



This file itself is a **module**.

---

# Why Use Modules?

Without modules:

```python
# All code in one file
```

Problems:

* Large code
* Difficult to manage
* Difficult to reuse

With modules:

```python
calculator.py
main.py
```

Benefits:

✅ Code Reusability

✅ Better Organization

✅ Easy Maintenance

✅ Reduces Duplication

---

# 2. Importing a Module

Suppose we have:

### calculator.py

```python
def add(a, b):
    return a + b

def multiply(a, b):
    return a * b
```

### main.py

```python
import calculator

print(calculator.add(10, 20))
print(calculator.multiply(5, 6))
```

Output:

```python
30
30
```

---

# How import Works

```python
import module_name
```

Example:

```python
import math
```

Now all functions of `math` can be accessed using:

```python
math.function_name()

#### Example

````python
import math

print(math.sqrt(25))
````

---

# 3. Import Specific Functions

Instead of importing the whole module:


```python
import math

print(math.sqrt(25))
```

You can import only required functions.

```python
from math import sqrt

print(sqrt(25))
```

Output:

```python
5.0
```

---

# Import Multiple Functions

```python
from math import sqrt, factorial
```

Example:

````python
from math import sqrt, factorial

print(sqrt(49))
print(factorial(5))
````


---

# 4. Import Everything

```python
from math import *
```


````python
from math import *

print(sqrt(64))
print(factorial(5))
````



### Why Not Recommended?

Suppose two modules have the same function name.

Then confusion occurs.

Therefore avoid:

```python
from module import *
```

in large projects.

---

# 5. Using Alias

Alias means giving another name.

Syntax:

```python
import module_name as alias
```

````python
import math as m

print(m.sqrt(81))
````


---

Another Example:

```python
import numpy as np
```

```python
import pandas as pd
```

Very common in Data Science.

---

# 6. Import Specific Function with Alias


````python
from math import factorial as fact

print(fact(5))
````

# 9. The **name** Variable

Every Python module has a special variable:

```python
__name__
```

Example:

```python
print(__name__)
```

If the file runs directly:

```python
__main__
```

If imported:

```python
module_name
```



---

# Example

### calculator.py

```python
def add(a, b):
    return a + b

if __name__ == "__main__":
    print(add(5, 5))
```

When run directly:

```python
10
```

When imported:

```python
No output
```

This prevents test code from running during import.

# 10. What is a Package?

A **package** is a collection of modules organized in directories.

Think:

```text
Module  = Single Python File

Package = Folder containing modules
```

Example:

```text
mypackage/
│
├── calculator.py
├── greetings.py
└── __init__.py
```

---



# Why Packages?

Suppose a company project has:

```text
100 Python files
```

Keeping all in one folder is messy.

Packages help organize them.

---

# Package Structure Example

```text
college/
│
├── students.py
├── teachers.py
├── library.py
└── __init__.py
```

# **init**.py File

Traditionally:

```python
__init__.py
```

marks a folder as a package.

Example:

```python
print("Package Loaded")
```

Modern Python can work without it, but it is still commonly used.

# Importing from Package

Structure:

```text
mypackage/
│
├── calculator.py
└── __init__.py
```


calculator.py


````python
def add(a, b):
    return a+b
````

main.py

````python
from MyPackage import calculator

print(calculator.add(10,20))
````

---

# Import Specific Function from Package


````python
from MyPackage.calculator import add

print(add(10,20))
````

# Common Import Errors

## 1. Module Not Found

```python
import xyz
```

Output:

```python
ModuleNotFoundError
```

Reason:

Module doesn't exist.

---

## 2. Wrong Function Name

```python
from math import square
```

Output:

```python
ImportError
```

Because `square()` doesn't exist in `math`.

# Best Practices

✅ Import only required functions

```python
from math import sqrt
```

✅ Use aliases for long module names

```python
import numpy as np
```

✅ Organize large projects using packages

✅ Use `if __name__ == "__main__":`

❌ Avoid:

```python
from module import *
```

# Frequently Asked Viva Questions

### Q1. What is a module?

A Python file containing functions, classes, variables, and statements.

---

### Q2. What is a package?

A collection of related modules organized in a directory.

---

### Q3. Difference Between Module and Package?

| Module            | Package             |
| ----------------- | ------------------- |
| Single `.py` file | Folder of modules   |
| Small unit        | Larger organization |
| Contains code     | Contains modules    |

---

### Q4. What is `__init__.py`?

A special file used to indicate a package and initialize package-level code.

---

### Q5. What is the use of `__name__ == "__main__"`?

It ensures that code runs only when the file is executed directly, not when imported.
