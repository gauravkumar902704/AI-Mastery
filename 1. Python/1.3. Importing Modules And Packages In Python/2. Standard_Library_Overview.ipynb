* Operating system interaction
* Data structures
* Mathematical operations
* Date and time manipulation

* Testing and debugging
* Multithreading and multiprocessing

# CSV Module in Python (Comma-Separated Values)

The **`csv`** module is a **Python Standard Library** module used to **read from** and **write to CSV (Comma-Separated Values) files**.

A **CSV file** is a plain text file where data is stored in rows and columns, with each value separated by a comma (or another delimiter such as `;` or `\t`).

Since the `csv` module is part of the **Python Standard Library**, **no installation is required**.

```python
import csv
```

---

# What is a CSV File?

Example: **students.csv**

```csv
Name,Age,Course
Rahul,20,CSE
Amit,21,AI
Priya,19,IT
```

This represents the following table:

| Name  | Age | Course |
| ----- | --- | ------ |
| Rahul | 20  | CSE    |
| Amit  | 21  | AI     |
| Priya | 19  | IT     |

---

# Why Use CSV?

CSV files are commonly used because they are:

* Easy to read
* Human-readable
* Supported by Excel and Google Sheets
* Small in size
* Used for data exchange between applications

# Reading CSV Files

Suppose **students.csv** contains:

```csv
Name,Age,Course
Rahul,20,CSE
Amit,21,AI
Priya,19,IT
```

---

## Method 1: Using `csv.reader()`

```python
import csv

with open("students.csv", "r") as file:
    reader = csv.reader(file)

    for row in reader:
        print(row)
```

### Output

```python
['Name', 'Age', 'Course']
['Rahul', '20', 'CSE']
['Amit', '21', 'AI']
['Priya', '19', 'IT']
```

### Explanation

* `csv.reader()` reads one row at a time.
* Each row is returned as a **list of strings**.

---

## Skipping the Header

```python
import csv

with open("students.csv", "r") as file:
    reader = csv.reader(file)

    next(reader)      # Skip header

    for row in reader:
        print(row)
```

Output

```python
['Rahul', '20', 'CSE']
['Amit', '21', 'AI']
['Priya', '19', 'IT']
```

---

# Accessing Individual Columns

```python
import csv

with open("students.csv") as file:
    reader = csv.reader(file)

    next(reader)

    for row in reader:
        print("Name:", row[0])
        print("Age :", row[1])
        print("Course:", row[2])
        print()
```

Output

```
Name: Rahul
Age : 20
Course: CSE

Name: Amit
Age : 21
Course: AI

Name: Priya
Age : 19
Course: IT
```

# Writing CSV Files

## `csv.writer()`

```python
import csv

with open("students.csv", "w", newline="") as file:
    writer = csv.writer(file)

    writer.writerow(["Name", "Age", "Course"])
    writer.writerow(["Rahul", 20, "CSE"])
    writer.writerow(["Amit", 21, "AI"])
```

Generated file

```csv
Name,Age,Course
Rahul,20,CSE
Amit,21,AI
```

### Why `newline=""`?

When writing CSV files on Windows, using:

```python
newline=""
```

prevents **blank lines** from appearing between rows.

Always use:

```python
with open("file.csv", "w", newline="")
```

when writing CSV files.

---

# Writing Multiple Rows

```python
import csv

data = [
    ["Name", "Age", "Course"],
    ["Rahul", 20, "CSE"],
    ["Amit", 21, "AI"],
    ["Priya", 19, "IT"]
]

with open("students.csv", "w", newline="") as file:
    writer = csv.writer(file)

    writer.writerows(data)
```

`writerows()` writes multiple rows at once.

---

# Reading CSV as Dictionaries

Instead of lists, you can read rows as dictionaries.

```python
import csv

with open("students.csv") as file:
    reader = csv.DictReader(file)

    for row in reader:
        print(row)
```

Output

```python
{'Name': 'Rahul', 'Age': '20', 'Course': 'CSE'}
{'Name': 'Amit', 'Age': '21', 'Course': 'AI'}
{'Name': 'Priya', 'Age': '19', 'Course': 'IT'}
```

Now access values using column names:

```python
import csv

with open("students.csv") as file:
    reader = csv.DictReader(file)

    for row in reader:
        print(row["Name"], row["Course"])
```

Output

```
Rahul CSE
Amit AI
Priya IT
```

---

# Writing Dictionaries

```python
import csv

with open("students.csv", "w", newline="") as file:

    fields = ["Name", "Age", "Course"]

    writer = csv.DictWriter(file, fieldnames=fields)

    writer.writeheader()

    writer.writerow({
        "Name": "Rahul",
        "Age": 20,
        "Course": "CSE"
    })

    writer.writerow({
        "Name": "Amit",
        "Age": 21,
        "Course": "AI"
    })
```

Output

```csv
Name,Age,Course
Rahul,20,CSE
Amit,21,AI
```

### Example Workflow

```python
import csv

# Write data
students = [
    ["Name", "Age", "Course"],
    ["Rahul", 20, "CSE"],
    ["Amit", 21, "AI"],
]

with open("students.csv", "w", newline="") as file:
    writer = csv.writer(file)
    writer.writerows(students)

# Read data
with open("students.csv", "r") as file:
    reader = csv.reader(file)
    for row in reader:
        print(row)
```

This example demonstrates the complete cycle of creating a CSV file, writing data into it, and reading the data back using Python's built-in `csv` module.
