# 1. What is Matplotlib?

**Definition**

Matplotlib is an open-source Python library used to create static, animated, and interactive visualizations.

It converts numerical data into graphical representations.

Example:

Instead of looking at this

| Month | Sales |
| ----- | ----- |
| Jan   | 20    |
| Feb   | 30    |
| Mar   | 50    |

You can create

📈 Line Chart

which is much easier to understand.

# 38. Complete Example


````python


months=["Jan","Feb","Mar","Apr"]

sales=[20,35,30,45]

plt.figure(figsize=(8,5))

plt.plot(
months,
sales,
color="blue",
marker="o",
linewidth=3,
label="Sales"
)

plt.title("Monthly Sales")

plt.xlabel("Months")

plt.ylabel("Sales")

plt.grid(True)

plt.legend()

plt.show()
````

# 39. Object-Oriented (OO) Syntax (Recommended)

Besides `pyplot`, Matplotlib also supports an object-oriented approach, which is preferred for larger projects.

````python

fig, ax = plt.subplots(figsize=(8,5))

ax.plot(x, y, color='blue', marker='o')
ax.set_title("Monthly Sales")
ax.set_xlabel("Month")
ax.set_ylabel("Sales")
ax.grid(True)

plt.show()
````



Advantages:

* Better control over multiple plots.
* Easier to customize complex figures.
* Standard approach in professional data science projects.

# 42. Matplotlib + NumPy Example


````python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 2*np.pi, 100)
y = np.sin(x)

plt.plot(x, y, color='green')
plt.title("Sine Wave")
plt.xlabel("Angle (radians)")
plt.ylabel("sin(x)")
plt.grid(True)
plt.show()
````

# 44. Common Errors

### 1. `ModuleNotFoundError`

```python

```

**Solution:**

```bash

```

---

### 2. Nothing is displayed

**Cause:** Missing `plt.show()`.

```python

```

---

### 3. `ValueError`

```python
x = [1, 2, 3]
y = [10, 20]
```

**Reason:** `x` and `y` must have the same length.
