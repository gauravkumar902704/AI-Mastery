# 1. What is Pandas?

**Definition:**

Pandas is an open-source Python library built on top of **NumPy** that is used for:

* Data Analysis
* Data Cleaning
* Data Manipulation
* Data Visualization Preparation
* Machine Learning Data Preprocessing

It works mainly with **tabular data** (rows and columns), similar to Excel or SQL tables.

# 19. GroupBy





`groupby()` is one of the most powerful functions in Pandas. It lets you **split your data into groups, perform calculations on each group, and then combine the results**.

Think of it like asking questions such as:

* "What is the average salary in each department?"
* "How many employees are in each city?"
* "Which department has the highest total salary?"

This is known as the **Split → Apply → Combine** process.


---

# Understanding `groupby()` with an Example

Suppose you have this data:

| Name    | Department | Salary |
| ------- | ---------- | ------ |
| Alice   | IT         | 60000  |
| Bob     | HR         | 50000  |
| Charlie | IT         | 70000  |
| David   | HR         | 55000  |
| Emma    | Sales      | 65000  |


### Without `groupby()`

Finding the average salary for IT manually would require filtering first.

### With `groupby()`

```python
df.groupby("Department")["Salary"].mean()
```

Output:

| Department | Average Salary |
| ---------- | -------------: |
| HR         |          52500 |
| IT         |          65000 |
| Sales      |          65000 |

Pandas automatically:

1. Splits the data by department.
2. Calculates the average salary for each department.
3. Returns the combined result.




---

# Syntax

```python
df.groupby("Column")
```

or

```python
df.groupby("Column")["Another_Column"].mean()
```

General pattern:

```python
df.groupby("group_column")["value_column"].function()
```

---

`groupby("Color")` separates them into three groups.

The same idea applies to data: group by a category, then calculate statistics for each group.

# Interview Tip

A simple way to explain `groupby()` is:

> "`groupby()` is used to divide data into groups based on one or more columns, perform calculations (such as `sum`, `mean`, `count`, `max`, or `min`) on each group, and return the aggregated results. It is commonly used in reporting, business analytics, sales analysis, HR dashboards, finance, healthcare, and many other real-world data analysis tasks."

Once you're comfortable with `groupby()`, you'll be able to create summaries that are the foundation of dashboards and business reports.
