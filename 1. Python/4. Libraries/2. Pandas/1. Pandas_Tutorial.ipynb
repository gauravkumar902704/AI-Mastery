# Complete Pandas Tutorial 

Pandas is one of the most powerful Python libraries for **data analysis, data manipulation, and data cleaning**. It provides easy-to-use data structures and functions to work with structured data.

---


# 1. What is Pandas?

**Definition:**

Pandas is an open-source Python library built on top of **NumPy** that is used for:

* Data Analysis
* Data Cleaning
* Data Manipulation
* Data Visualization Preparation
* Machine Learning Data Preprocessing

It works mainly with **tabular data** (rows and columns), similar to Excel or SQL tables.


---

# 2. Why Use Pandas?

Without Pandas:

* Reading files is difficult.
* Filtering data is complicated.
* Statistical calculations require many loops.

With Pandas:

* Few lines of code.
* Faster processing.
* Easy data cleaning.
* Powerful data analysis.

# 4. Pandas Data Structures

Pandas has two major data structures.



## A. Series (1-D)

A Series is a one-dimensional labeled array.

Syntax

```python
import pandas as pd

s = pd.Series(data)
```


````python

# Example



s = pd.Series([10,20,30,40])

print(s)

````


---

### Series with Custom Index

````python
s = pd.Series([100,200,300], index=['A','B','C'])

print(s)
````

````python
labels = ["Gaurav Kumar", "Tarun Kumar", "Ankit Kumar", "Dev Karan", "Saim"]
s = pd.Series([101,102,103,104,105], index=labels)

print(s)
````

````python
d = {"Gaurav Kumar": 101, "Tarun Kumar": 102, "Ankit Kumar": 103, "Dev Karan": 104, "Saim": 105}
s = pd.Series(d)

print(s)
````

````python
data = {
    "Name":["Ram","Shyam","Mohan"],
    "Age":[20,21,19]
}

s = pd.Series(data)
print(s)
````


---

### Access Series


````python
print(s["Gaurav Kumar"])
````

````python
print(s['A'])
````


---

# B. DataFrame (2-D)

A DataFrame is a two-dimensional table.

It contains

* Rows
* Columns

Syntax

````python
data = {
    "Name":["Ram","Shyam","Mohan"],
    "Age":[20,21,19]
}

df = pd.DataFrame(data)

print(df)
````
