# 2. Titanic Survival Prediction

### Dataset

[https://www.kaggle.com/c/titanic/data](https://www.kaggle.com/c/titanic/data)

### Questions

1. Overall survival rate.
2. Survival by gender.
3. Survival by passenger class.
4. Average age of survivors.
5. Missing values analysis.
6. Fill missing ages.
7. Family size calculation.
8. Youngest survivor.
9. Richest passenger.
10. Correlation between Fare and Survival.

```python
path1 = (r"..\0. Datasets\1. Titanic_Dataset\gender_submission.csv")


path2 = (r"..\0. Datasets\1. Titanic_Dataset\train.csv")

path3 = (r"..\0. Datasets\1. Titanic_Dataset\test.csv")
```

```python
import pandas as pd
```

```python

Titanic_gender_df = pd.read_csv(path1)
Titanic_Train_df = pd.read_csv(path2)
Titanic_Test_df = pd.read_csv(path3)

Titanic_gender_df.head()
```

```python
print(Titanic_gender_df.shape)
print(Titanic_gender_df.isnull().sum())
```

```python
Titanic_gender_df.describe()
```

### **Question 1: Overall Survival Rate**

```python
Titanic_Train_df["Survived"].mean()*100
```

### **Question 2: Survival by Gender**

```python
Titanic_Train_df.groupby("Sex")["Survived"].mean()*100
```

### **Question 3: Survival by Passenger Class**

```python
Titanic_Train_df.groupby('Pclass')["Survived"].mean()*100
```

### **Question 4: Average Age Of Survivors**

```python
Titanic_Train_df[Titanic_Train_df["Survived"] == 1]["Age"].mean()
```

### **Question 5: Missing Values Analysis**

```python
(Titanic_Train_df.isnull().sum()/len(Titanic_Train_df))*100
```

### **Question 6: Fill Missing Ages**

```python
Titanic_Train_df["Age"] = Titanic_Train_df["Age"].fillna(Titanic_Train_df['Age'].median())
```

```python
Titanic_Train_df["Age"].isnull().sum()
```

```python
(Titanic_Train_df.isnull().sum()/len(Titanic_Train_df))*100
```

### **Question 7: Family Size Calculation**

Family Size = sibSp + Parch + 1

```python
Titanic_Train_df["FamilySize"] = Titanic_Train_df["SibSp"] + Titanic_Train_df["Parch"] + 1
```

```python
Titanic_Train_df.head()
```

### **Question 8: Youngest Survivor**

```python
survivors = Titanic_Train_df[Titanic_Train_df["Survived"]==1]
survivors.sort_values("Age").head(1)
```

### **Question 9: Richest Passenger**
Highest Fare

```python
Titanic_Train_df.loc[Titanic_Train_df["Fare"].idxmax()]
```

### **Question 10: Correlation between Fare and Survival**

```python
Titanic_Train_df[["Fare", "Survived"]].corr()
```
