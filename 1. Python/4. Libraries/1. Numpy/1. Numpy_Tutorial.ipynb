# What is NumPy?


**Definition:**

NumPy is an open-source Python library that provides:

* Fast multidimensional arrays
* Mathematical functions
* Linear algebra
* Statistics
* Random number generation
* Matrix operations
* Fourier Transform
* Scientific computations

Officially,

> NumPy is the fundamental package for numerical computing in Python.

# Why was NumPy created?


Python lists are flexible but **slow** for numerical computations.

Example:

Imagine you have

```
10 million numbers
```

Adding them using Python lists takes much longer than using NumPy because NumPy is written mainly in **C**, making it much faster.

# Broadcasting

One of NumPy's most powerful features.

````python
a = np.array([1,2,3,4])
a = a+10

print(a)
````


Instead of looping

```python
for i in range(len(a)):
    a[i]+=10
```

NumPy automatically broadcasts the value.

# Advantages of NumPy

| Advantage                        | Explanation                                                                       |
| -------------------------------- | --------------------------------------------------------------------------------- |
| Very Fast                        | Implemented mainly in C for high performance                                      |
| Memory Efficient                 | Uses contiguous memory blocks                                                     |
| Easy Mathematical Operations     | Vectorized operations avoid explicit loops                                        |
| Supports Multidimensional Arrays | Work with 1D, 2D, 3D, and higher dimensions                                       |
| Rich Library                     | Includes statistics, linear algebra, FFT, random numbers, and more                |
| Highly Compatible                | Integrates well with Pandas, Scikit-learn, TensorFlow, PyTorch, OpenCV, and SciPy |
| Open Source                      | Free and widely supported by the community                                        |

---

# Disadvantages of NumPy

| Disadvantage             | Explanation                                                                                |
| ------------------------ | ------------------------------------------------------------------------------------------ |
| Homogeneous Data         | All elements in an array should ideally have the same data type.                           |
| Fixed Size               | Arrays cannot be resized efficiently like Python lists.                                    |
| Less Flexible            | Not suitable for storing mixed data types (e.g., strings, numbers, objects together).      |
| Learning Curve           | Concepts like broadcasting and advanced indexing can be confusing for beginners.           |
| Limited for Tabular Data | For spreadsheets or tables with different column types, Pandas is usually a better choice. |
