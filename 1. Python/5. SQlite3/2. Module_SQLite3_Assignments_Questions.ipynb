2. Write a Python function to create a table named `employees` with columns `id` (integer), `name` (text), `age` (integer), and `department` (text) in the `test.db` database.


```python
def Create_Table():
    connection = sqlite3.connect(path)
    cursor = connection.cursor()
    cursor.execute('''
        CREATE TABLE IF NOT EXISTS students(
            id INTEGER PRIMARY KEY,
            name TEXT NOT NULL,
            age INTEGER,
            department TEXT
            )
    ''')
    connection.commit()
    connection.close()
    print("Table 'students' created successfully.")

# Test
Create_Table()
```

```python
# Use this when your table is creates wrong.
connection = sqlite3.connect(path)
cursor = connection.cursor()
cursor.execute("DROP TABLE IF EXISTS students")
connection.commit()
connection.close()
```

```python
def Create_Table2():
    connection = sqlite3.connect(path)
    cursor = connection.cursor()
    cursor.execute("""
        CREATE TABLE IF NOT EXISTS employees(
            E_id INTEGER PRIMARY KEY,
            Name TEXT NOT NULL,
            Age INTEGER,
            Dept TEXT
        )
""")
    connection.commit()
    connection.close()
    print("Table 'employees' created successfully.")

# Test
Create_Table2()
```

### Assignment 2: Inserting Data

1. Write a Python function to insert a new employee into the `employees` table.


```python
def Insert_Records(table, id , name, age, dept):
    try:
        if table == 'students':
            connection = sqlite3.connect(path)
            cursor = connection.cursor()
            cursor.execute("""
                insert or ignore into students(id, name, age, department)
                values (?, ?, ?, ?)
            """, (id, name, age, dept))
            connection.commit()
            connection.close()
            print('Record added to Students table...')

        elif table == 'employees':
            connection = sqlite3.connect(path)
            cursor = connection.cursor()
            cursor.execute('''
                insert or ignore into employees(E_id, Name, Age, Dept)
                values (?, ?, ?, ?)
            ''', (id, name, age, dept))
            connection.commit()
            connection.close()
            print('Record added to employees table...')
    except Exception as e:
        print(e)

# Test
Insert_Records('students',101,'Gaurav Kumar', 21, 'Information Technology')
Insert_Records('employees', 6, 'Evel', 22, 'HR')
```

1. Write a Python function to fetch and display all records from the `employees` table.


```python
def Fetch_All_Records(table):
    try:
        if table == 'students':
            connection = sqlite3.connect(path)
            cursor = connection.cursor()
            cursor.execute("SELECT * FROM students")
            records = cursor.fetchall()
            connection.close()

            for record in records:
                print(record)

        elif table == 'employees':
            connection = sqlite3.connect(path)
            cursor = connection.cursor()
            cursor.execute("SELECT * FROM employees")
            records = cursor.fetchall()
            connection.close()

            for record in records:
                print(record)


    except Exception as e:
        print(e)    
```
