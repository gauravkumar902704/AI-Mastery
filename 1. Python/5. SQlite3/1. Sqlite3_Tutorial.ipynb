# 36. Parameterized Queries (Recommended)

Never concatenate user input into SQL strings. Use placeholders to prevent SQL injection.

```python
import sqlite3

conn = sqlite3.connect("student.db")
cur = conn.cursor()

cur.execute(
    "INSERT INTO Student (Name, Marks) VALUES (?, ?)",
    ("Aman", 92.5)
)

conn.commit()
conn.close()
```

For multiple inserts:

```python
students = [
    ("Riya", 88),
    ("Karan", 76),
    ("Neha", 95)
]

cur.executemany(
    "INSERT INTO Student (Name, Marks) VALUES (?, ?)",
    students
)
```

# 38. ACID Properties

* **Atomicity:** Either all operations succeed or none do.
* **Consistency:** Database remains valid before and after transactions.
* **Isolation:** Concurrent transactions do not interfere.
* **Durability:** Committed changes survive crashes.

---

# 39. Best Practices

* Use `INTEGER PRIMARY KEY` for unique IDs.
* Enable foreign key support using `PRAGMA foreign_keys = ON`.
* Use transactions for multiple related operations.
* Create indexes on frequently searched columns.
* Always use parameterized queries (`?`) in Python.
* Avoid `SELECT *` in production; select only needed columns.
* Close database connections after use.
* Back up databases regularly.

# 40. Common Errors

| Error                      | Cause                                 | Solution                               |
| -------------------------- | ------------------------------------- | -------------------------------------- |
| no such table              | Table does not exist                  | Check table name or create it          |
| UNIQUE constraint failed   | Duplicate value                       | Insert a unique value                  |
| NOT NULL constraint failed | Missing required value                | Provide a value                        |
| datatype mismatch          | Wrong data type                       | Use compatible values                  |
| database is locked         | Another process is using the database | Close other connections or retry later |
