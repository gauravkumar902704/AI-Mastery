# File Handling in Python

File handling is used to **create, read, write, append, and manage files** in Python.

## Opening a File

Syntax: 
```python
file = open("filename.txt", "mode")
```

### Common Modes

| Mode   | Description                                                     |
| ------ | --------------------------------------------------------------- |
| `"r"`  | Read (default)                                                  |
| `"w"`  | Write (creates file if not exists, overwrites existing content) |
| `"a"`  | Append (adds data to end of file)                               |
| `"x"`  | Create a new file                                               |
| `"r+"` | Read and write                                                  |
| `"b"`  | Binary mode (e.g., images, PDFs)                                |

---

## Reading a File

### Read Entire File

````python
file = open("sample.txt", "r")
content = file.read()
print(content)
file.close()
````

### Read One Line

````python
file = open("sample.txt", "r")
print(file.readline())
file.close()
````

### Read All Lines

````python
with open('sample.txt', mode='r') as file:
    All_lines = file.readlines()
    print(All_lines)
    
````

````python
lines = ['\nFirst line\n','Second line\n','Third line\n','Fourth line\n']
try:
    with open('newfile.txt', mode='a') as file:
        file.writelines(lines)

except FileNotFoundError:
    print("Such File does not exist")


with open('newfile.txt', mode='r') as file:
    content = file.read()
    print(content)
````


## Best Practices

* Use `with open()` instead of `open()` and `close()`.
* Handle exceptions using `try-except`.
* Use appropriate file modes.
* Close files properly to avoid memory leaks.

Example with exception handling:


````python
try:
    with open("sample.txt", "r") as file:
        print(file.read())
except FileNotFoundError:
    print("File does not exist")
````

## 1. `tell()`

The `tell()` method returns the current position of the file pointer as the number of bytes from the beginning of the file.

### Syntax

```python
file.tell()
```

### Example

````python
with open('big.txt','r') as file:
    print(file.read(10))
    print(file.tell())
````

````python
f = open("sample.txt", "r")

print(f.tell())      # Output: 0 (pointer is at the beginning)

f.read(5)
print(f.tell())      # Output: 5

f.close()
````

**Explanation:**

* Initially, the pointer is at position `0`.
* After reading 5 characters (in a simple text file), the pointer moves to position `5`.
* `tell()` shows the current location.

---

### **2. `seek()`**

The `seek()` method moves the file pointer to a specified position.

### Syntax

```python
file.seek(offset, whence)
```

### Parameters

* **offset**: Number of bytes to move.
* **whence** (optional):

  * `0` → Beginning of the file (default)
  * `1` → Current position (mainly supported in binary mode)
  * `2` → End of the file (mainly supported in binary mode)

---

### Example 1: Move to the beginning

````python
f = open("sample.txt", "r")

f.read(5)
print(f.tell())      # 5

f.seek(0)
print(f.tell())      # 0

print(f.read())
f.close()
````


---

### Example 2: Jump to a specific position


````python
f = open("sample.txt", "r")

f.seek(6)
print(f.read())

f.close()
````

---

## Example Using Both `tell()` and `seek()`


````python
f = open("sample.txt", "r")

print("Initial position:", f.tell())

print(f.read(4))
print("Position after reading:", f.tell())

f.seek(0)
print("After seek:", f.tell())

print(f.read())

f.close()
````



---

## Summary

| Method                 | Purpose                                  | Returns                     |
| ---------------------- | ---------------------------------------- | --------------------------- |
| `tell()`               | Shows the current file pointer position  | Current position (in bytes) |
| `seek(offset, whence)` | Moves the file pointer to a new position | `None`                      |

### Quick Analogy

Imagine a **bookmark** in a book:

* **`tell()`** answers: *"Which page is my bookmark currently on?"*
* **`seek()`** lets you move the bookmark to another page before continuing to read.

These methods are especially useful when you need to reread part of a file, skip sections, or randomly access data without reading the entire file from the beginning.
