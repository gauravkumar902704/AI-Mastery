# Working with File Paths in Python

A **file path** is the location of a file or folder in a computer's file system. Python provides the **`os`** and **`pathlib`** modules to work with file paths.

## 1. Types of File Paths

### (a) Absolute Path

An **absolute path** specifies the complete location of a file from the root directory.

**Example (Windows):**

```python
C:\Users\Rahul\Documents\sample.txt
```

**Example (Linux/macOS):**

```python
/home/rahul/Documents/sample.txt
```

---

### (b) Relative Path

A **relative path** specifies the file location relative to the current working directory.

**Example:**

```python
sample.txt
```

or

```python
data/sample.txt
```

---


# 2. Using the `os` Module

First, import the module:

````python
import os
````

### (a) Get the Current Working Directory


````python
print(os.getcwd())
````


---

### (b) Change the Current Directory


````python
os.chdir(r"C:\Users\Lenovo\Desktop\AI Mastery\1. Python\1.2. Functions In Python")
print(os.getcwd())
````

````python
os.chdir(r"C:\Users\Lenovo\Desktop\AI Mastery\1. Python\3. File_Handling")
print(os.getcwd())
````

### (c) Join File Paths

Instead of manually writing `\` or `/`, use `os.path.join()`.

````python
path = os.path.join("Documents", "Python", "sample.txt")
print(path)
````

---

### (d) Check if a File Exists

````python
print(os.path.exists("sample.txt"))
````


---

### (e) Check Whether It Is a File or Directory


````python
print(os.path.isfile("sample.txt"))
print(os.path.isdir("Documents"))
````

---

# 3. Using the `pathlib` Module (Recommended)

`pathlib` provides a modern, object-oriented way to work with file paths.


````python
from pathlib import Path
````

### (a) Create a Path


````python
path = Path("sample.txt")
print(path)
````

---

### (b) Get the Current Directory


````python
print(Path.cwd())
````


---

### (c) Check if a File Exists


````python
path = Path("sample.txt")

print(path.exists())
````


---

### (d) Check if It Is a File or Folder


````python
path = Path("sample.txt")

print(path.is_file())
print(path.is_dir())
````

---

### (e) Join Paths


````python
path = Path("Documents") / "Python" / "sample.txt"

print(path)
````


(`pathlib` automatically uses the correct separator for your operating system.)

---

# 4. Extract File Information


````python
path = Path("Documents/sample.txt")

print(path.name)       # sample.txt
print(path.stem)       # sample
print(path.suffix)     # .txt
print(path.parent)     # Documents
````

---

# 5. Example: Open a File Using a Path


````python
path = Path("sample.txt")

with open(path, "r") as file:
    print(file.read())
````

# Summary

| Function/Method    | Purpose                                       |
| ------------------ | --------------------------------------------- |
| `os.getcwd()`      | Get the current working directory             |
| `os.chdir(path)`   | Change the current working directory          |
| `os.path.join()`   | Join file paths safely                        |
| `os.path.exists()` | Check if a file/folder exists                 |
| `os.path.isfile()` | Check if the path is a file                   |
| `os.path.isdir()`  | Check if the path is a directory              |
| `Path.cwd()`       | Get the current working directory (`pathlib`) |
| `Path.exists()`    | Check whether the path exists                 |
| `Path.is_file()`   | Check if the path is a file                   |
| `Path.is_dir()`    | Check if the path is a directory              |
| `Path.name`        | Get the file name                             |
| `Path.stem`        | Get the file name without the extension       |
| `Path.suffix`      | Get the file extension                        |
| `Path.parent`      | Get the parent directory                      |

### Best Practice

For new Python code, prefer the **`pathlib`** module because it is easier to read, platform-independent, and more powerful than the older `os.path` functions.
