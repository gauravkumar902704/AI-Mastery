# Classes and Objects in Python (OOP) – 

Object-Oriented Programming (OOP) is a programming paradigm that organizes code into **objects** instead of just functions and variables.

Python is an **object-oriented programming language**, meaning almost everything in Python is an object (numbers, strings, lists, functions, etc.).

---

# What is a Class?

A **class** is a blueprint or template for creating objects.

It defines:

* Data (attributes/variables)
* Functions (methods)

Think of a class as a design of a house.

Example:

A house blueprint contains:

* Number of rooms
* Color
* Kitchen
* Bathroom

But the blueprint itself is not a real house.

Similarly,

A class only defines what an object will contain.

---

```python
    pass
```


````python
# Example
class Student:
    pass
````

# Another Real Example – Bank Account

````python
class BankAccount:

    def __init__(self, owner, balance):

        self.owner = owner

        self.balance = balance

    def deposit(self, amount):

        self.balance += amount

        print(amount, "deposited")

    def withdraw(self, amount):

        if amount <= self.balance:

            self.balance -= amount

            print(amount, "withdrawn")

        else:

            print("Insufficient Balance")

    def show_balance(self):

        print("Current Balance:", self.balance)
````


Create Object


````python
acc1 = BankAccount("Rahul", 5000)

acc1.show_balance()

acc1.deposit(2000)

acc1.withdraw(1500)

acc1.show_balance()
````

# Important Points

* A **class** is a blueprint or template.
* An **object** is an instance of a class.
* Attributes store data for each object.
* Methods define the behavior of objects.
* `self` refers to the current object and is passed automatically when calling instance methods.
* The `__init__()` constructor initializes object attributes automatically during object creation.
* A single class can create multiple objects, each with its own independent state.

````python
class ComplexNumber2:
    def __init__(self, real=0, imag=0):
        self.real = real
        self.imag = imag

    # String Representation
    def __str__(self):
        if self.real == 0 and self.imag == 0:
            return "0"
        elif self.real == 0:
            return f"{self.imag}i"
        elif self.imag == 0:
            return f"{self.real}"
        elif self.imag > 0:
            return f"{self.real}+{self.imag}i"
        else:
            return f"{self.real}{self.imag}i"

    # Official Representation
    def __repr__(self):
        return f"ComplexNumber({self.real}, {self.imag})"

    # Addition
    def __add__(self, other):
        if not isinstance(other, ComplexNumber2):
            return NotImplemented

        return ComplexNumber2(
            self.real + other.real,
            self.imag + other.imag
        )

    # Subtraction
    def __sub__(self, other):
        if not isinstance(other, ComplexNumber2):
            return NotImplemented

        return ComplexNumber2(
            self.real - other.real,
            self.imag - other.imag
        )

    # Multiplication
    def __mul__(self, other):
        if not isinstance(other, ComplexNumber2):
            return NotImplemented

        real = self.real * other.real - self.imag * other.imag
        imag = self.real * other.imag + self.imag * other.real

        return ComplexNumber2(real, imag)

    # Division
    def __truediv__(self, other):
        if not isinstance(other, ComplexNumber2):
            return NotImplemented

        denominator = other.real**2 + other.imag**2

        if denominator == 0:
            raise ZeroDivisionError("Cannot divide by zero complex number.")

        real = (
            self.real * other.real +
            self.imag * other.imag
        ) / denominator

        imag = (
            self.imag * other.real -
            self.real * other.imag
        ) / denominator

        return ComplexNumber2(real, imag)

    # Equality
    def __eq__(self, other):
        if not isinstance(other, ComplexNumber2):
            return False

        return self.real == other.real and self.imag == other.imag

    # Inequality
    def __ne__(self, other):
        return not self == other

    # Conjugate
    def conjugate(self):
        return ComplexNumber2(self.real, -self.imag)

    # Modulus (Magnitude)
    def modulus(self):
        return (self.real**2 + self.imag**2) ** 0.5

    # Argument (Angle in radians)
    def argument(self):
        import math
        return math.atan2(self.imag, self.real)
````
