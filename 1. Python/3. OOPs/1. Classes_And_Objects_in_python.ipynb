# Classes and Objects in Python (OOP) – 

Object-Oriented Programming (OOP) is a programming paradigm that organizes code into **objects** instead of just functions and variables.

Python is an **object-oriented programming language**, meaning almost everything in Python is an object (numbers, strings, lists, functions, etc.).

---

# What is a Class?

A **class** is a blueprint or template for creating objects.

It defines:

* Data (attributes/variables)
* Functions (methods)

Think of a class as a design of a house.

Example:

A house blueprint contains:

* Number of rooms
* Color
* Kitchen
* Bathroom

But the blueprint itself is not a real house.

Similarly,

A class only defines what an object will contain.

---

## Syntax

```python
class ClassName:
    pass
```


````python
# Example
class Student:
    pass
````

Here,

* `Student` is a class.
* It currently contains nothing.

# What is an Object?

An object is an **instance of a class**.

If class = blueprint

then

object = actual building.

Example:

Blueprint

```
Car
```

Actual Cars

```
Car 1
Car 2
Car 3
```

Each car is an object of the Car class.

## Creating an Object

Syntax

```python
object_name = ClassName()
```

Example

````python
class Student:
    pass

s1 = Student()
s2 = Student()

print(s1)
print(s2)
````

Output

```

<__main__.Student object at 0x...>
```

Notice:

`s1` and `s2` are different objects.



---

# What is `self`?

This is the most important concept in Python OOP.

Whenever we call

```python
s1.display()
```

Python automatically converts it into

```python
Student.display(s1)
```

That means

`self` refers to the **current object**.

---

Example

````python
class Student:

    def display(self):
        print(self)
````

````python
s1 = Student()

s1.display()

print(s1)
````


Output

```
<__main__.Student object at 0x...>

<__main__.Student object at 0x...>
```

Notice

Both addresses are identical.

That means

```
self == s1
```

# Constructor (`__init__`)

A constructor is a special method that automatically runs when an object is created.

Syntax:

````python
class Student:

    def __init__(self):
        print("Constructor Called")
````


Object

```python
s1 = Student()
```

Output

```
Constructor Called
```

No need to call it manually.

---

# Why Constructor?

Instead of writing

```python
s1 = Student()

s1.name = "Rahul"

s1.age = 20
```

We write

````python
class Student:

    def __init__(self, name, age):

        self.name = name

        self.age = age
````



Create object

```python
s1 = Student("Rahul",20)
```

Now

````python
s1.name = "Gaurav Kumar"
s1.age = 22
````

````python

print(s1.name)
print(s1.age)
````



Cleaner and easier.

# Complete Example

````python
class Student:

    def __init__(self, name, age, marks):

        self.name = name

        self.age = age

        self.marks = marks

    def display(self):

        print("Name :", self.name)

        print("Age :", self.age)

        print("Marks :", self.marks)
````


Objects


````python
s1 = Student("Rahul",20,90)

s2 = Student("Amit",21,95)

s1.display()

print()

s2.display()
````

# Another Real Example – Bank Account

````python
class BankAccount:

    def __init__(self, owner, balance):

        self.owner = owner

        self.balance = balance

    def deposit(self, amount):

        self.balance += amount

        print(amount, "deposited")

    def withdraw(self, amount):

        if amount <= self.balance:

            self.balance -= amount

            print(amount, "withdrawn")

        else:

            print("Insufficient Balance")

    def show_balance(self):

        print("Current Balance:", self.balance)
````


Create Object


````python
acc1 = BankAccount("Rahul", 5000)

acc1.show_balance()

acc1.deposit(2000)

acc1.withdraw(1500)

acc1.show_balance()
````

# Another Example – Car

````python
class Car:

    def __init__(self, brand, model):

        self.brand = brand

        self.model = model

    def start(self):

        print(self.brand, self.model, "started.")
````

Create objects

````python
c1 = Car("Toyota","Fortuner")

c2 = Car("BMW","X5")

c1.start()

c2.start()
````

# Class vs Object

| Feature | Class                                | Object                |
| ------- | ------------------------------------ | --------------------- |
| Meaning | Blueprint                            | Instance of class     |
| Memory  | No separate memory for instance data | Has its own memory    |
| Purpose | Defines attributes and methods       | Stores actual data    |
| Number  | One class can create many objects    | Each object is unique |
| Example | `Student`                            | `s1`, `s2`            |

---

# Important Points

* A **class** is a blueprint or template.
* An **object** is an instance of a class.
* Attributes store data for each object.
* Methods define the behavior of objects.
* `self` refers to the current object and is passed automatically when calling instance methods.
* The `__init__()` constructor initializes object attributes automatically during object creation.
* A single class can create multiple objects, each with its own independent state.

# Interview Questions

### 1. What is a class?

A class is a user-defined blueprint that defines the properties (attributes) and behaviors (methods) of objects.

### 2. What is an object?

An object is an instance of a class that contains actual data and can invoke the class's methods.

### 3. What is the difference between a class and an object?

A class is a template; an object is a concrete instance created from that template.

### 4. What is `self` in Python?

`self` is a reference to the current object. It allows methods to access and modify the object's attributes.

### 5. What is the purpose of `__init__()`?

`__init__()` is the constructor method that is called automatically when an object is created. It is used to initialize the object's attributes.
