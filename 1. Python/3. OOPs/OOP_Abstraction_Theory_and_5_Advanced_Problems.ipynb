# OOP: Abstraction

## 1. Definition

**Abstraction** is an OOP principle that exposes essential operations while hiding implementation details.

It focuses on **what an object does**, rather than **how it does it**.

**Real-world example:** When using an ATM, a customer selects withdraw, enters an amount, and receives cash. The customer does not need to understand the bank's internal transaction processing.

In Python, abstraction is commonly implemented using:
- Abstract classes
- Abstract methods
- The `abc` module

## 2. Abstract Classes and Abstract Methods

An **abstract class** defines a common structure for related classes. It may contain both abstract methods and fully implemented methods.

An **abstract method** declares an operation that concrete subclasses must implement.

Python provides `ABC` and `abstractmethod` through the `abc` module.

Important points:
- A class inherits from `ABC` to act as an abstract base class.
- Mark a method with `@abstractmethod` to require an implementation in concrete subclasses.
- An abstract class cannot be instantiated while it has unimplemented abstract methods.
- A subclass becomes instantiable after implementing all inherited abstract methods.

```python
from abc import ABC, abstractmethod

class Payment(ABC):

    @abstractmethod
    def pay(self, amount):
        pass

class CardPayment(Payment):
    def pay(self, amount):
        print(f"Processing card payment of ₹{amount}")

class UPIPayment(Payment):
    def pay(self, amount):
        print(f"Processing UPI payment of ₹{amount}")

card = CardPayment()
upi = UPIPayment()

card.pay(1500)
upi.pay(800)

# Payment()  # TypeError: abstract class cannot be instantiated
```

## 3. Abstract Classes Can Contain Concrete Methods

An abstract class does not need to contain only abstract methods. It can also provide shared implementation that subclasses inherit.

Use abstract methods for operations that vary by subclass, and concrete methods for behavior that can be shared.

```python
from abc import ABC, abstractmethod

class Report(ABC):

    @abstractmethod
    def generate_content(self):
        pass

    def print_header(self):
        print("Company Monthly Report")
        print("----------------------")

class SalesReport(Report):
    def generate_content(self):
        print("Total sales: ₹5,00,000")

report = SalesReport()
report.print_header()       # Inherited concrete method
report.generate_content()   # Subclass implementation
```

## 4. Constructor in an Abstract Class

An abstract base class can have an `__init__()` method. Subclasses can call it with `super()` to initialize common attributes.

This avoids repeating common initialization logic.

```python
from abc import ABC, abstractmethod

class Employee(ABC):
    def __init__(self, employee_id, name):
        self.employee_id = employee_id
        self.name = name

    @abstractmethod
    def calculate_salary(self):
        pass

class FullTimeEmployee(Employee):
    def __init__(self, employee_id, name, monthly_salary):
        super().__init__(employee_id, name)
        self.monthly_salary = monthly_salary

    def calculate_salary(self):
        return self.monthly_salary

employee = FullTimeEmployee(101, "Aman", 45000)
print(employee.name)
print(employee.calculate_salary())
```

## 5. Multiple Abstract Methods

An abstract class can define multiple required operations. A concrete subclass must implement every inherited abstract method before it can be instantiated.

```python
from abc import ABC, abstractmethod

class Database(ABC):

    @abstractmethod
    def connect(self):
        pass

    @abstractmethod
    def execute_query(self, query):
        pass

    @abstractmethod
    def disconnect(self):
        pass

class MySQLDatabase(Database):
    def connect(self):
        print("Connected to MySQL.")

    def execute_query(self, query):
        print(f"Executing MySQL query: {query}")

    def disconnect(self):
        print("MySQL connection closed.")

db = MySQLDatabase()
db.connect()
db.execute_query("SELECT * FROM employees")
db.disconnect()
```

## 6. Interface-Like Design with Abstract Base Classes

Python does not have a separate `interface` keyword like some other languages. An abstract base class can provide an interface-like contract by defining abstract methods that subclasses must implement.

For example, different storage systems can expose the same `save()` and `load()` operations while using different internal implementations.

An abstract base class may also provide shared behavior and attributes, so it is more flexible than a pure interface in some languages.

```python
from abc import ABC, abstractmethod

class Storage(ABC):

    @abstractmethod
    def save(self, key, value):
        pass

    @abstractmethod
    def load(self, key):
        pass

class MemoryStorage(Storage):
    def __init__(self):
        self._data = {}

    def save(self, key, value):
        self._data[key] = value

    def load(self, key):
        return self._data.get(key)

storage = MemoryStorage()
storage.save("theme", "dark")
print(storage.load("theme"))
```

## 7. Abstraction vs Encapsulation

| Abstraction | Encapsulation |
|---|---|
| Hides implementation complexity. | Controls access to internal state. |
| Focuses on what an object does. | Focuses on how data and behavior are packaged and protected. |
| Commonly uses abstract classes and methods. | Commonly uses private-by-convention attributes, properties and methods. |
| Example: a common `pay()` contract. | Example: preventing invalid changes to an account balance. |

## 8. Advantages of Abstraction

- Reduces complexity for code that uses an object.
- Establishes a clear contract for subclasses.
- Supports interchangeable implementations.
- Encourages modular and maintainable design.
- Makes large systems easier to extend.

## 9. Key Points

- Import `ABC` and `abstractmethod` from `abc`.
- An abstract class cannot be instantiated while abstract methods remain unimplemented.
- A subclass must implement all inherited abstract methods to become concrete.
- Abstract classes can contain constructors, attributes, and concrete methods.
- Use abstraction when multiple implementations should follow a common contract.
