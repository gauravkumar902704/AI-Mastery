# ⭐ Question 1: Student Class (Very Easy)

### Objective

Create a class named `Student`.

### Attributes

* Name
* Roll Number
* Marks

### Methods

* Display Details
* Update Marks
* Check Pass/Fail (Passing Marks = 40)

### Example

```python
s1 = Student("Gaurav", 101, 85)

s1.display()

s1.update_marks(92)

s1.is_pass()
```

### Concepts

* Constructor
* Instance Variables
* Simple Methods

````python
# Create a class named `Student`.

class Student:

    def __init__(self, name, roll_number, marks):
        '''
    Attributes:

        * Name
        * Roll Number
        * Marks
    '''
        self.name = name
        self.roll_number = roll_number
        if 0 <= marks <= 100:
            self.marks = marks
        else:
            raise ValueError("Marks should be between 0 and 100")
        
        

    '''
        Methods:

            * Display Details
            * Update Marks
            * Check Pass/Fail (Passing Marks = 40)
    '''

    def update_marks(self, marks):
        if 0 <= marks <= 100:
            self.marks = marks
        else:
            raise ValueError("Marks should be between 0 and 100")
        

    def pass_or_fail(self):

        if self.marks >= 40:
            return "PASS"
        return "FAIL"
        
    def display_details(self):
        return(
            f"Name : {self.name}\n"
            f"Roll Number : {self.roll_number}\n"
            f"Marks : {self.marks}\n"
            f"Result : {self.pass_or_fail()}\n"
        )


    
````

````python
Student1 = Student("Gaurav Kumar", 2303410130012, 71)
print(Student1.display_details())
Student1.update_marks(15)
print(Student1.display_details())
````

# ⭐⭐ Question 2: Rectangle Class

### Objective

Create a class named `Rectangle`.

### Attributes

* Length
* Width

### Methods

* Area
* Perimeter
* Check Square

### Example

```python
r = Rectangle(10, 5)

print(r.area())

print(r.perimeter())

print(r.is_square())
```

### Concepts

* Mathematical Methods
* Returning Values

````python
class Rectangle:
    """
    Rectangle class

    Description
    -----------
    A Rectangle is a four-sided polygon where opposite sides are equal
    and all interior angles are 90 degrees.
    """
    def __init__(self, length, width):
        """
        Attributes
        ----------
        length : float/int
            Length of the rectangle.

        width : float/int
            Width of the rectangle.

        """
        if length <= 0 or width <= 0:
            raise ValueError("Length and Width must be greater than zero.")
        self.length = length
        self.width  = width

    

    # Methods
    #-------
    def area(self):
        """    
            area()
                Calculates and returns the area of the rectangle.

                Formula:
                    Area = Length × Width
        """
        return self.length * self.width

    def perimeter(self):
        """
        perimeter()
            Calculates and returns the perimeter of the rectangle.

            Formula:
                Perimeter = 2 × (Length + Width)
        """
        return 2* (self.length + self.width)

    def check_square(self):
        """
            check_square()
                Checks whether the rectangle is a square.

                Condition:
                    If Length == Width
                        → Square
                    Else
                        → Not a Square
        """
        if self.length == self.width:
            return "Is Square"
        return "Not a Square" 
    
    def __str__(self):
        return(
        f"Length : {self.length}\n"
        f"Width : {self.width}\n"
        f"Area : {self.area()}\n"
        f"Perimeter : {self.perimeter()}\n"
        f"Is Square : {self.check_square()}\n"
        )
````

````python
rect = Rectangle(10,10)
print(rect)
````

# ⭐⭐⭐ Question 3: BankAccount (Completed)

### Methods

* Deposit
* Withdraw
* Transfer
* Check Balance

### Concepts

* Validation
* Method Calling Method
* Code Reuse


````python
class BankAccount:
    '''
    BankAccount class

    Description
    -----------
    A BankAccount represents a customer's bank account.
    It allows users to deposit money, withdraw money,
    transfer money to another account, and check the
    available account balance securely.
    '''
    unique_account_number = 1000000000001
    initial_balance  = 0.0

    def __init__(self, name):
        '''
            Attributes
            ----------
            account_holder : str
                Name of the account holder.

            account_number : int/str
                Unique account number.

            balance : float
                Current available balance in the account.
        '''
        self.account_holder = name
        self.account_number = BankAccount.unique_account_number
        BankAccount.unique_account_number += 1
        self.account_balance = BankAccount.initial_balance

    #     Methods
    #     -------
    def deposit(self,deposit_amount):
        '''
        deposit(amount)
        Adds money to the account balance.

        Validation:
            • Amount must be greater than 0.

        Formula:
            New Balance = Current Balance + Deposit Amount
        '''
        if deposit_amount <= 0:
            raise ValueError("Amount must be greater than 0")
        self.account_balance += deposit_amount

    def withdraw(self,withdraw_amount):
        '''
        withdraw(amount)
        Removes money from the account balance.

        Validation:
            • Amount must be greater than 0.
            • Amount must not exceed the available balance.

        Formula:
            New Balance = Current Balance - Withdraw Amount
        '''
        if withdraw_amount <= 0:
            raise ValueError("Amount must be greater than 0.")
        elif withdraw_amount > self.account_balance:
            raise ValueError("Amount must not exceed the available balance.")
        self.account_balance -= withdraw_amount

    def transfer(self, receiver_account, amount):
        '''
        transfer(receiver_account, amount)
        Transfers money from one account to another.

        Validation:
            • Transfer amount must be greater than 0.
            • Sender must have sufficient balance.

        Method Calling Method:
            • Calls withdraw(amount)
            • Calls receiver_account.deposit(amount)

        Code Reuse:
            • Reuses the existing deposit() and withdraw()
              methods instead of writing the same logic again.
        '''
        self.withdraw(amount)
        receiver_account.deposit(amount)
        print(
                f"✅ ₹{amount} transferred from "
                f"{self.account_holder} to {receiver_account.account_holder}"
            )

    def check_balance(self):
        """Display current balance."""

        return (f"Current Balance : ₹{self.account_balance}")


    def __str__(self):
        return (
            f"\nAccount Holder : {self.account_holder}\n"
            f"Account Number : {self.account_number}\n"
            f"Balance        : ₹{self.account_balance}"
        )
````

````python
acc1 = BankAccount("Gaurav Kumar")
print(acc1)
````

````python
acc2 = BankAccount("Tarun Kumar")
print(acc2)
````

````python
acc1.deposit(10000)
print(acc1)
acc2.deposit(50000)
print(acc2)
````

````python
acc2.withdraw(500)
````

````python
print(acc2)
````

````python
acc2.transfer(acc1, 1000)

````

````python
acc2.check_balance()
````

# ⭐⭐⭐⭐ Question 4: Employee Class

### Attributes

* Name
* Employee ID
* Salary

### Methods

* Increase Salary
* Annual Salary
* Display Details

### Example

```python
emp = Employee("Rahul",1001,50000)

emp.increase_salary(10)

print(emp.annual_salary())
```

### Concepts

* Percentage Calculation
* Updating Object State

````python
class Employee:
    '''
    Description
    -----------
    An Employee represents a person working in an organization.
    The class stores employee details and provides methods to
    calculate salary, apply bonuses, and display employee
    information.
    '''

    def __init__(self, name, E_id, dept, salary):
        
        '''
        Attributes
        ----------
            employee_name : str
                Name of the employee.

            employee_id : int/str
                Unique employee ID.

            department : str
                Department in which the employee works.

            basic_salary : float
                Monthly basic salary of the employee.
        '''
        self.name = name
        self.employee_id = E_id
        self.department = dept
        if salary < 0:
            raise ValueError("Salary cannot be negative.")
        self.salary = salary

    #   Methods
    #   -------
    def calculate_salary(self):
        '''
        Calculates and returns the employee's salary.

        Formula:
            Salary = Basic Salary
        '''
        return self.salary
    
    def apply_bonus(self, bonus = 0):
        '''
        Adds bonus to the employee's salary.

        Validation:
            • Bonus must be greater than or equal to 0.

        Formula:
            New Salary = Current Salary + Bonus

        '''
        if bonus < 0:
            raise ValueError("Bonus must be greater than or equal to 0.")
        
        self.salary += bonus
        return self.salary 
        
    
    def update_department(self,new_department):
        '''
        Updates the employee's department.

        Validation:
            • Department name must not be empty.
        '''
        if not new_department.strip() :
            raise ValueError("Department name must not be empty.")
        self.department = new_department

    def display_details(self):
        '''
            Displays all employee information.
        '''
        #print(f"Employee Name: {self.name}")
        #print(f"Employee Id: {self.employee__id}")
        #print(f"Department Name: {self.department}")
        #print(f"Salary: {self.salary}")
        print(self)
        

    def __str__(self):
        return (
            f"\nEmployee Name: {self.name}\n"
            f"Employee Id: {self.employee_id}\n"
            f"Department Name: {self.department}\n"
            f"Salary : ₹{self.salary}"
        )

````

````python
e1 = Employee("Gaurav Kumar",10141,'CSE',100000)
print(e1)
````

````python
e1.display_details()
````

````python
e1.apply_bonus(5452)
````

````python
e1.update_department("A.I.")
print(e1)
````
