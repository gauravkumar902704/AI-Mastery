# Object-Oriented Programming: Polymorphism

## 1. What is Polymorphism?

**Polymorphism** means *many forms*. In OOP, it allows objects of different classes to respond to the same method, function, or operator in their own way.

For example, a `CreditCardPayment` and a `UPIPayment` can both provide a `pay()` method. The calling code uses `pay()` without needing to know the internal payment method.

**Main idea:** one interface, different implementations.

## 2. Polymorphism Through Method Overriding

Method overriding occurs when a child class defines a method with the same name as a method in its parent class. When that method is called on a child object, Python uses the child's implementation.

This is a common way to achieve runtime polymorphism.

````python
class Notification:
    def send(self, message):
        print(f"Sending notification: {message}")

class EmailNotification(Notification):
    def send(self, message):
        print(f"Email sent: {message}")

class SMSNotification(Notification):
    def send(self, message):
        print(f"SMS sent: {message}")

notifications = [EmailNotification(), SMSNotification()]

for notification in notifications:
    notification.send("Your order has been shipped.")
````

## 3. Polymorphism Through a Common Interface

Unrelated classes can provide methods with the same name. A function can work with any object that provides the required method.

This makes code extensible: a new class can be introduced without changing the function that uses the interface.

````python
class PDFReport:
    def export(self):
        print("Exporting report as PDF.")

class ExcelReport:
    def export(self):
        print("Exporting report as Excel.")

def export_report(report):
    report.export()

export_report(PDFReport())
export_report(ExcelReport())
````

## 4. Duck Typing

Python follows **duck typing**: the behavior of an object matters more than its declared class.

The idea is: if an object provides the methods an operation needs, it can be used for that operation. Python generally checks this when the code runs, not by requiring a particular parent class.

````python
class Laptop:
    def connect(self):
        print("Laptop connected to the network.")

class SmartTV:
    def connect(self):
        print("Smart TV connected to the network.")

def connect_device(device):
    device.connect()

connect_device(Laptop())
connect_device(SmartTV())
````

## 5. Operator Overloading

Operator overloading gives built-in operators a defined meaning for objects of a user-defined class.

Python uses special methods, also called *dunder methods*, for this purpose:

| Operator | Special method |
|---|---|
| `+` | `__add__()` |
| `-` | `__sub__()` |
| `*` | `__mul__()` |
| `==` | `__eq__()` |
| `<` | `__lt__()` |
| `str()` | `__str__()` |

For example, defining `__add__()` lets two objects of a class work with `+`.

````python
class Money:
    def __init__(self, amount):
        self.amount = amount

    def __add__(self, other):
        if not isinstance(other, Money):
            return NotImplemented
        return Money(self.amount + other.amount)

    def __str__(self):
        return f"₹{self.amount:.2f}"

total = Money(250) + Money(175)
print(total)
````

## 6. Method Overloading in Python

Traditional method overloading—defining several methods with the same name but different parameter lists—is not directly supported in Python. A later definition with the same name replaces the earlier one.

Similar behavior can be implemented using:
- Default parameter values.
- `*args` and `**kwargs`.
- Type checks or `functools.singledispatch` when appropriate.

This is different from method overriding.

````python
class Calculator:
    def add(self, *numbers):
        if not numbers:
            return 0
        return sum(numbers)

calculator = Calculator()
print(calculator.add(10, 20))
print(calculator.add(10, 20, 30, 40))
````

## 7. Polymorphism: Overloading vs. Overriding

| Method Overriding | Method Overloading |
|---|---|
| Child class redefines a parent method. | Same method name supports different parameter patterns. |
| Supported directly in Python. | Not supported directly through repeated definitions. |
| Commonly used for runtime polymorphism. | Can be approximated with defaults, `*args`, or dispatch tools. |

## 8. Why Polymorphism Is Useful

- **Extensibility:** new implementations can be added with little or no change to calling code.
- **Loose coupling:** calling code depends on a method contract rather than concrete classes.
- **Maintainability:** implementation details remain inside their respective classes.
- **Consistency:** different objects can be used through a shared interface.
