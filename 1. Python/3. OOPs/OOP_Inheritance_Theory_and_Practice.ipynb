# OOP: Inheritance

## 1. Definition
**Inheritance** is an Object-Oriented Programming (OOP) concept in which a class (child/derived class) acquires the attributes and methods of another class (parent/base class). It supports code reuse and helps organize related classes.

## 2. Basic Syntax (Python)
```python
class Parent:
    # attributes and methods
    pass

class Child(Parent):
    # additional attributes and methods
    pass
```

## 3. Example
A `Dog` is an `Animal`, so `Dog` can reuse methods defined in `Animal`.

````python
class Animal:
    def eat(self):
        print("This animal eats food.")

class Dog(Animal):
    def bark(self):
        print("The dog barks.")

dog = Dog()
dog.eat()   # inherited method
dog.bark()  # child class method
````

## 4. Types of Inheritance

1. **Single inheritance:** One child class inherits from one parent class.
2. **Multiple inheritance:** One child class inherits from two or more parent classes.
3. **Multilevel inheritance:** A class inherits from a child class, forming a chain.
4. **Hierarchical inheritance:** Multiple child classes inherit from the same parent class.
5. **Hybrid inheritance:** A combination of two or more inheritance types.

## 5. Method Overriding
A child class can provide its own implementation of a method already defined in the parent class. This is called **method overriding**.

````python
class Animal:
    def sound(self):
        print("Animal makes a sound.")

class Cat(Animal):
    def sound(self):  # overrides the parent method
        print("Meow")
        print("That means it's a Billi..")

object = Cat()
object.sound()
````

## 6. The `super()` Function
`super()` lets a child class call methods—including the initializer—of its parent class.

````python
class Person:
    def __init__(self, name):
        self.name = name

class Student(Person):
    def __init__(self, name, roll_no):
        super().__init__(name)
        self.roll_no = roll_no

student = Student("Aman", 101)
print(student.name, student.roll_no)
````

## 7. Advantages
- Reuses existing code.
- Reduces duplication.
- Makes related classes easier to maintain.
- Supports method overriding and specialization.

2. **Using `__init__`, Method overriding:** Create a `Person` class that stores `name` and `age`. Create a `Student` class that inherits from `Person` and stores `roll_no`. Use `super()` to initialize the parent attributes.


````python
class Person:
    def __init__(self, name , age):
        self.name = name
        self.age = age
        
    def details(self):
        print("Name: ",self.name,"\nAge: ",self.age)

class Student(Person):
    def __init__(self, name, age, roll_no):
        super().__init__(name, age)
        self.roll_no = roll_no

    def details(self):
        print("Roll_no: ",self.roll_no)
        super().details()

obj = Student("Gaurav Kumar", 23,2303410130012)
obj.details()

````


4. **Hierarchical inheritance:** Create a parent class `Employee` with a method `show_company()`. Create `Developer` and `Manager` classes that inherit from it, each with one additional method of its own.


````python

class Employee:
    def __init__(self, E_id, Employee_name, company_name):
        self.E_id = E_id
        self.Employee_name = Employee_name
        self.company_name = company_name

    def show_company(self):
        print(f"Company Name: {self.company_name}")


class Developer(Employee):
    def developer_details(self):
        print(f"Employee ID: {self.E_id}")
        print(f"Employee Name: {self.Employee_name}")
        print("Position: Developer")
        self.show_company()


class Manager(Employee):
    def manager_details(self):
        print(f"Employee ID: {self.E_id}")
        print(f"Employee Name: {self.Employee_name}")
        print("Position: Manager")
        self.show_company()


# Developer object
obj1 = Developer(101, "Sohan Tiwari", "Salesforce")
obj1.developer_details()

print("--------------------")

# Manager object
obj2 = Manager(102, "Raju", "HCLTech")
obj2.manager_details()
````

5. **Multiple inheritance:** Create classes `Camera` and `Phone`, each with one method. Create a `Smartphone` class that inherits from both and demonstrate all three methods.

````python

class Camera:
    def click_photo(self):
        print("Photo captured successfully!")


class Phone:
    def make_call(self):
        print("Making a phone call...")


class Smartphone(Camera, Phone):
    def use_apps(self):
        print("Running smartphone applications...")


# Creating Smartphone object
mobile = Smartphone()

# Calling all three methods
mobile.click_photo()
mobile.make_call()
mobile.use_apps()
````

## 1. Single Inheritance – Banking System

Level 1

Single Inheritance

Problem: Create a banking system using single inheritance.

Scenario:

Ek bank mein customers ka account hota hai. Har account ke paas account number, account holder name aur balance hota hai. Lekin Savings Account mein additional interest rate hota hai.

Your task:

A. Create a parent class `BankAccount`

Attributes:

- `account_number`
- `account_holder`
- `balance`

Methods:

- `display_account()` – Account ki complete details print kare.
- `deposit(amount)` – Account mein money add kare.
- `withdraw(amount)` – Account se money deduct kare. Agar balance insufficient hai toh appropriate message show kare.

B. Create a child class `SavingsAccount`

- `BankAccount` se inherit kare.
- Additional attribute `interest_rate` add kare.
- `calculate_interest()` method banaye jo balance ke according annual interest calculate kare.

C. Create objects and test

- Ek `SavingsAccount` object banao.
- Initial account details print karo.
- ₹5,000 deposit karo.
- ₹2,000 withdraw karo.
- Annual interest calculate karo.

Expected output (format):

```
Account Holder: Rahul
Account Number: 101
Balance: 20000
Deposit: 5000
Withdrawal: 2000
Annual Interest: 2300
```

Assume initial balance ₹20,000 and interest rate 10%. Output values ko apne calculations ke according verify karna.

````python
class BankAccount:
    def __init__(self, account_number, account_holder, balance):
        self.account_number = account_number
        self.account_holder = account_holder
        self.balance = balance

    def display_account(self):
        print("Account Number: ",self.account_number)
        print("Account Holder: ",self.account_holder)
        print("Balance: ",self.balance)

    def current_balance(self):
        print("Current Balance is: ", self.balance)

    def deposit(self, amount):
        if amount <= 0:
            raise ValueError("Amount must be greater than zero.")

        self.balance += amount
        print("Deposited successfully.")
        self.current_balance()


    def withdraw(self, amount):
        if amount <= 0:
            raise ValueError("Amount must be greater than zero.")

        if amount > self.balance:
            raise ValueError("Insufficient balance.")

        self.balance -= amount
        print("Withdrawal successful.")
        self.current_balance()

class SavingsAccount(BankAccount):
    def __init__(self, account_number, account_holder, balance, interest_rate):
        super().__init__(account_number, account_holder, balance)
        self.interest_rate = interest_rate

    def calculate_interest(self):
        interest = (self.balance * self.interest_rate) / 100
        print("Annual Interest:", interest)


# Creating the object of SavingsAccount Class..

acc1 = SavingsAccount(201240510479, "Gaurav Kumar", 20000, 10)
print("-------------------------------")
acc1.display_account()
print("-------------------------------")
acc1.deposit(5642)
print("--------------------------------")
acc1.withdraw(42)
print("--------------------------------")
acc1.calculate_interest()

````
