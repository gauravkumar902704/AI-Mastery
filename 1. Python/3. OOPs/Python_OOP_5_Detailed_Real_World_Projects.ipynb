# Python OOP & Fundamentals — 5 Real-World Projects

## Practical project workbook

This notebook contains five progressively challenging projects designed to strengthen Python fundamentals and the four pillars of Object-Oriented Programming:

- Encapsulation
- Inheritance
- Polymorphism
- Abstraction

The projects are intentionally written as **implementation briefs**, not solved tutorials. Build them yourself, test edge cases, and refactor your code as the project grows.

## How to use this workbook

For every project:

1. Read the problem statement and business rules carefully.
2. Write down the classes, attributes, methods and relationships before coding.
3. Draw a simple class diagram on paper.
4. Implement the minimum working version first.
5. Add validation, exceptions and file persistence.
6. Test normal cases, invalid cases and boundary cases.
7. Refactor into modules only after the core logic works.

**Important:** Do not put the entire application in one class or one large function. Keep responsibilities separate. Do not store money as floating-point values in a real financial application; use `decimal.Decimal` for accurate currency calculations.

## Project progression

| Project | Level | Main learning focus |
|---|---|---|
| 1. Banking Management System | Intermediate | Classes, encapsulation, inheritance, exceptions |
| 2. E-commerce Management System | Intermediate | Composition, polymorphism, collections, files |

| 4. Employee Payroll & HR System | Advanced | All four OOP pillars, business rules |

# Project 1 — Banking Management System

<badge>Intermediate</badge>

## 1. Problem statement

A small financial institution wants a command-line banking application to manage customers, bank accounts and transactions. The system must support different account types with different rules. Customers should be able to deposit money, withdraw money, transfer funds and inspect their transaction history.

Your task is to design the application using OOP rather than writing a collection of unrelated functions.

## 2. Core requirements

### Customer management
- Create a customer with a unique customer ID, full name, email and phone number.
- Validate required fields and reject duplicate customer IDs.
- A customer may own more than one bank account.
- Display customer details without exposing internal account data unnecessarily.

### Account management
- Create at least two account types: `SavingsAccount` and `CurrentAccount`.
- Every account must have a unique account number, an owner and a balance.
- Support deposit, withdrawal, balance enquiry and account statement.
- Do not allow direct, uncontrolled modification of an account's balance.

### Transaction management
- Record deposits, withdrawals and transfers.
- Each transaction should contain an ID, type, amount, timestamp, account reference and description.
- A transfer must debit one account and credit another.
- A failed transfer must not leave the system in a partially updated state.
- Reject zero or negative transaction amounts.

## 3. Suggested class design

| Class | Responsibility |
|---|---|
| `Customer` | Stores customer profile and account references |
| `BankAccount` | Common account behaviour and protected balance operations |
| `SavingsAccount` | Savings-specific rules, such as minimum balance |
| `CurrentAccount` | Current-account rules, such as an overdraft limit |
| `Transaction` | Represents one financial transaction |
| `Bank` | Coordinates customers, accounts and transfers |

You may introduce additional classes if they improve the design.

## 4. OOP concepts you must demonstrate

- **Encapsulation:** protect account balance and validate every balance-changing operation.
- **Inheritance:** derive savings and current accounts from a common account class.
- **Polymorphism:** each account type should apply its own withdrawal rules through a common method.
- **Composition:** a bank manages customers and accounts; an account has transaction records.
- **Abstraction:** consider which common account operations should be defined as an abstract contract.

## 5. Business rules

1. Account numbers and customer IDs must be unique.
2. Deposit amount must be greater than zero.
3. Savings accounts must maintain their configured minimum balance after withdrawal.
4. Current accounts may go below zero only up to their configured overdraft limit.
5. An account cannot transfer money to itself.
6. A transfer is allowed only if the source account has sufficient available funds under its account rules.
7. If either side of a transfer fails, both account balances must remain unchanged.
8. Store timestamps consistently using `datetime`.
9. Never print or log sensitive information such as passwords or full payment credentials.

## 6. Required Python features

- Dictionaries for account/customer lookup.
- Lists for transaction history.
- `Decimal` for currency.
- Custom exceptions such as `InsufficientFundsError`, `InvalidAmountError` and `AccountNotFoundError`.
- JSON persistence for customers, accounts and transactions.
- Functions or methods for searching and filtering transaction history.

## 7. Acceptance tests

Your application should correctly handle these scenarios:

- Create one customer and two different account types.
- Deposit 5,000 into a savings account.
- Attempt a withdrawal that would violate the minimum balance; it must be rejected.
- Withdraw an allowed amount and verify the updated balance.
- Transfer money between two accounts and verify both balances.
- Attempt a transfer with insufficient funds; neither balance should change.
- Reject a negative deposit and a zero withdrawal.
- Restart the application and confirm that saved records can be loaded.

## 8. Suggested development milestones

1. Implement customer and account classes.
2. Add account-specific deposit and withdrawal rules.
3. Add transaction records and transfer handling.
4. Add input validation and custom exceptions.
5. Add JSON save/load functionality.
6. Write automated tests and refactor the application.

## 9. Expected deliverables

- Working command-line application.
- Class diagram.
- Separate modules for models, services, exceptions and storage.
- JSON data file generated by the application.
- Test file covering successful and failed transactions.
- README explaining setup, features and design decisions.

## 10. Challenge extensions

- Add fixed deposits and recurring deposits.
- Add daily withdrawal limits.
- Add transaction search by date and type.
- Add a monthly account statement in CSV format.
- Add role-based staff operations.

**Your task:** Design and implement the complete system. Do not skip the failed-transaction tests.

# Project 2 — E-commerce Management System

<badge>Intermediate</badge>

## 1. Problem statement

Build the backend logic for a small online shopping platform. Customers should be able to browse products, add items to a cart, apply discounts, place orders and request cancellations or refunds.

The goal is to model a realistic workflow while keeping product, cart, order, payment and inventory responsibilities separate.

## 2. Core requirements

### Product catalogue
- Create products with SKU, name, description, price and available quantity.
- Support different product categories.
- Search products by name or category.
- Reject duplicate SKUs and invalid prices.
- Prevent stock from becoming negative.

### Shopping cart
- A cart belongs to one customer.
- Add, remove and update product quantities.
- Reject quantities that are zero, negative or greater than available stock.
- Calculate subtotal, discount, tax and final total separately.
- A cart should not directly modify product prices.

### Discount engine
Implement at least three discount strategies:
- Percentage discount.
- Fixed-amount discount.
- Category-specific discount.

Each discount should have validation rules, such as minimum order value, maximum discount and expiry date.

### Orders and payments
- Convert a valid cart into an order.
- Store a snapshot of product names, prices and quantities at purchase time.
- Support at least two payment method implementations, such as card simulation and cash on delivery.
- Track order states: pending, confirmed, shipped, delivered and cancelled.
- Prevent invalid state transitions.
- Support cancellation and refund rules.

## 3. Suggested class design

| Class | Responsibility |
|---|---|
| `Product` | Product information and stock |
| `Customer` | Customer profile |
| `ShoppingCart` | Cart items and cart calculations |
| `CartItem` | Product and selected quantity |
| `DiscountStrategy` | Common discount contract |
| `PercentageDiscount` | Percentage-based discount |
| `FixedDiscount` | Fixed-amount discount |
| `PaymentMethod` | Common payment contract |
| `CardPayment` | Simulated card payment |
| `CashOnDelivery` | Cash-on-delivery behaviour |
| `Order` | Order details and state transitions |
| `InventoryService` | Stock reservation and release |

Treat this as a starting point, not a mandatory class list.

## 4. OOP concepts you must demonstrate

- **Composition:** a cart contains cart items; an order contains order-line snapshots.
- **Encapsulation:** stock and order state can only change through validated methods.
- **Polymorphism:** discount strategies and payment methods share common interfaces.
- **Abstraction:** define abstract contracts for discount and payment behaviour.
- **Inheritance:** use it only where subclasses genuinely share an is-a relationship.

## 5. Business rules

1. Product SKU must be unique.
2. Product price cannot be negative.
3. Cart quantity must be a positive integer.
4. The order total must be calculated from the price snapshot, not a later catalogue price.
5. Discounts cannot make the payable amount negative.
6. Expired discounts must be rejected.
7. Stock must be checked again when placing an order, not only when adding to the cart.
8. If payment fails, the order must not be marked as paid.
9. Cancelling an eligible order must release reserved stock.
10. A delivered order cannot be cancelled through the normal cancellation flow.

## 6. Required Python features

- Dictionaries for products and orders.
- Lists for cart items and order history.
- Sets for unique category/tag collections.
- `Decimal` for money.
- `Enum` for order and payment states.
- Custom exceptions for stock, discount and payment failures.
- JSON persistence for products, customers and orders.
- CSV export for order reports.

## 7. Acceptance tests

- Add two products to a cart and verify the subtotal.
- Apply a valid percentage discount and verify the final amount.
- Try to apply an expired discount; it must be rejected.
- Try to add more units than available stock.
- Place an order and verify stock reduction/reservation.
- Change a product's catalogue price after purchase; the existing order total must not change.
- Simulate a payment failure and verify that the order is not marked paid.
- Cancel an eligible order and verify stock release.
- Reject cancellation of a delivered order.

## 8. Suggested development milestones

1. Build product and customer models.
2. Implement cart and cart-item composition.
3. Implement discount strategies using polymorphism.
4. Implement payment methods and order states.
5. Add inventory reservation and cancellation.
6. Add persistence, reporting and tests.

## 9. Expected deliverables

- Working command-line shopping workflow.
- Class diagram showing composition and interfaces.
- Separate catalogue, cart, discount, payment, order and storage modules.
- Sample product/order JSON data.
- Automated tests for pricing, stock and order transitions.
- README with sample usage.

## 10. Challenge extensions

- Add product reviews and ratings.
- Add coupons with usage limits.
- Add shipping fee strategies.
- Add a wishlist.
- Add pagination and sorting for product search.
- Add a mock notification service for order updates.

**Your task:** Implement the end-to-end purchase flow. Pay special attention to stock consistency and order price snapshots.

# Project 3 — Hospital Management System

<badge>Intermediate–Advanced</badge>

## 1. Problem statement

Build a hospital administration application that manages patients, doctors, appointments, prescriptions and billing. The system must prevent scheduling conflicts and keep patient records organised.

This project is about modelling relationships between objects and enforcing rules across multiple classes.

## 2. Core requirements

### People and roles
- Maintain patient and staff profiles.
- Support doctors with a specialty and consultation duration.
- Support administrative staff who can schedule or cancel appointments.
- Use role-specific behaviour without duplicating common person details.

### Appointment scheduling
- Book an appointment for a patient with a doctor.
- Store appointment ID, date/time, duration, status and reason.
- Check doctor availability before booking.
- Prevent overlapping appointments for the same doctor.
- Prevent a patient from booking overlapping appointments.
- Support rescheduling and cancellation.
- Maintain appointment history.

### Medical records
- Maintain patient visit records.
- Record diagnosis, notes, prescriptions and visit date.
- Restrict record modification to authorised workflows.
- Do not expose one patient's records through another patient's account.

### Billing
- Generate an invoice for a completed consultation.
- Support consultation fees and additional service charges.
- Track invoice status: unpaid, paid or partially paid.
- Record payments and outstanding balances.

## 3. Suggested class design

| Class | Responsibility |
|---|---|
| `Person` | Shared identity and contact information |
| `Patient` | Patient-specific information and visit history |
| `HospitalStaff` | Common staff information |
| `Doctor` | Specialty, schedule and consultation rules |
| `Receptionist` | Appointment administration |
| `Appointment` | Appointment details and state |
| `MedicalRecord` | Visit notes and diagnosis |
| `Prescription` | Prescribed medicines and instructions |
| `Invoice` | Consultation charges and balance |
| `Payment` | Payment record |
| `AppointmentService` | Booking and conflict checks |

You may use an abstract base class for shared staff behaviour if it is useful.

## 4. OOP concepts you must demonstrate

- **Inheritance:** model shared person/staff attributes without repeating code.
- **Encapsulation:** protect appointment state and sensitive record updates.
- **Abstraction:** define contracts for services or staff operations where appropriate.
- **Polymorphism:** support different staff roles through common operations.
- **Composition:** an invoice contains line items; a medical record may contain prescriptions.
- **Association:** appointments connect patient and doctor objects.

## 5. Business rules

1. Patient and staff IDs must be unique.
2. A doctor cannot have two active appointments that overlap.
3. A patient cannot have overlapping appointments.
4. Cancelled appointments do not block future time slots.
5. Rescheduling must validate the new slot before changing the existing appointment.
6. Only completed consultations can generate a final consultation invoice.
7. A payment cannot exceed the outstanding invoice balance unless the system explicitly supports credit.
8. Medical record access must be checked before displaying or modifying records.
9. Invalid dates, missing doctors and unknown patients must produce clear errors.
10. Store date/time values consistently and document the chosen timezone policy.

## 6. Required Python features

- `datetime` and `timedelta` for appointment scheduling.
- Dictionaries for patient, doctor and appointment lookup.
- Lists for medical history and appointment history.
- Enums for appointment and invoice states.
- Custom exceptions for scheduling conflicts and access errors.
- JSON persistence for application records.
- Unit tests for overlapping time intervals.

## 7. Acceptance tests

- Register a patient and two doctors.
- Book an appointment in an available slot.
- Attempt to book another appointment for the same doctor at an overlapping time; reject it.
- Attempt to book overlapping appointments for the same patient with different doctors; reject it.
- Cancel an appointment and confirm the slot becomes available.
- Reschedule an appointment into an occupied slot; the original appointment must remain unchanged.
- Complete a consultation and generate an invoice.
- Record a partial payment and verify the outstanding balance.
- Attempt unauthorised access to a medical record; deny it.

## 8. Suggested development milestones

2. Implement doctor availability and appointment models.

# Project 4 — Employee Payroll and HR Management System

<badge>Advanced</badge>

## 1. Problem statement

A company needs a system to manage employee profiles, attendance, leave, salary calculations, bonuses and monthly payroll. Different employee categories follow different compensation rules.

Build a flexible payroll engine where adding a new employee type does not require rewriting the main payroll-processing workflow.

## 2. Core requirements

### Employee management
- Assign each employee a unique employee ID.
- Store name, department, joining date and employment status.
- Support permanent, contract and freelance employees.
- Maintain employee records and update them through validated methods.

### Attendance and leave
- Record attendance by date.
- Track working days, absence, approved leave and overtime.
- Prevent duplicate attendance records for the same employee and date.
- Validate leave requests against available leave balance.
- Keep approved and rejected leave requests in history.

### Compensation
- Permanent employees: monthly base salary plus eligible allowances and deductions.
- Contract employees: agreed rate and payable working days/hours.
- Freelancers: project/task-based payment.
- Calculate overtime, bonus, tax/deductions and net pay.
- Keep each employee type's calculation policy separate.

### Payroll processing
- Generate payroll for a selected month.
- Prevent duplicate payroll generation for the same employee and period.
- Produce payslips with salary components.
- Track payroll status: draft, approved and paid.
- Export payroll summaries to CSV.

## 3. Suggested class design

| Class | Responsibility |
|---|---|
| `Employee` | Common employee identity and shared behaviour |
| `PermanentEmployee` | Permanent compensation policy |
| `ContractEmployee` | Contract-based compensation |
| `Freelancer` | Task/project-based compensation |
| `AttendanceRecord` | One employee's attendance for a date |
| `LeaveRequest` | Leave dates, reason and approval status |
| `CompensationPolicy` | Contract for salary calculation rules |
| `PayrollItem` | One employee's payroll calculation |
| `Payslip` | Salary breakdown for one period |
| `PayrollService` | Coordinates payroll generation |
| `PayrollRepository` | Loads and saves payroll records |

You may choose a different design if you can explain why it is better.

## 4. OOP concepts you must demonstrate

- **Abstraction:** define an abstract compensation calculation contract.
- **Inheritance:** share common employee data through a base class.
- **Polymorphism:** calculate pay for different employee types through one common method.
- **Encapsulation:** validate salary, leave balance and payroll status changes.
- **Composition:** a payslip contains payroll components; payroll processing uses attendance and compensation policies.
- **Dependency separation:** payroll logic should not be tightly coupled to console input or file storage.

## 5. Business rules

1. Employee IDs must be unique.
2. Salary, hourly rate and project payment cannot be negative.
3. Attendance cannot be recorded twice for the same employee and date.
4. Leave balance cannot become negative.
5. Overtime must follow a clearly defined policy.
6. Each employee type must calculate compensation according to its own rules.
7. Net pay = gross earnings − total deductions.
8. A payroll item cannot be approved if required attendance or compensation data is missing.
9. A paid payroll period cannot be generated again for the same employee.
10. Changes to an employee's current salary must not silently rewrite previously generated payslips.

For learning purposes, implement a clearly documented simplified tax/deduction policy. Do not present it as legally compliant payroll or tax advice.

## 6. Required Python features

- Abstract Base Classes from `abc`.
- `Decimal` for monetary values.
- `datetime` for payroll periods and attendance.
- Dictionaries for employee lookup.
- Lists for attendance, leave and payslip history.
- Enums for employment and payroll statuses.
- Custom exceptions for duplicate payroll and invalid compensation.
- CSV export and JSON persistence.
- Unit tests with expected salary calculations.

## 7. Acceptance tests

- Verify deductions and net pay.
- Reject a negative salary or rate.
- Reject duplicate attendance for one employee on one date.

# Project 5 — ML Experiment and Model Management System

<badge>Advanced · AI/ML oriented</badge>

## 1. Problem statement

Build a small Python framework that manages machine-learning experiments. A user should be able to register datasets, configure model runs, train different model implementations, evaluate them using metrics, save model artefacts and compare experiment results.

This project connects OOP design with practical ML engineering. Start with simulated or simple scikit-learn models; do not begin with a complex deep-learning pipeline.

## 2. Core requirements

### Dataset management
- Register a dataset with a unique ID, name, file path and target column.
- Validate that the file exists before loading it.
- Load CSV data using pandas.
- Record basic metadata such as row count, column names and missing-value counts.
- Keep dataset loading separate from model training.

### Model interface
- Define a common abstract model contract with methods such as `train`, `predict` and `save`.
- Implement at least two model classes, for example:
  - `LinearRegressionModel`
  - `DecisionTreeModel`
- Each model should accept configuration parameters.
- A model implementation should be replaceable without changing the experiment manager.

### Experiment tracking
- Create an experiment with an ID, dataset reference, model type, parameters and timestamp.
- Track status: created, running, completed or failed.
- Record training duration and evaluation metrics.
- Store errors for failed runs without marking them as successful.
- Allow users to compare completed experiments.

### Evaluation
- Split data into training and testing sets.
- Support suitable metrics, such as MAE, MSE and R² for regression.
- Do not apply regression metrics blindly to classification tasks.
- Keep evaluation logic separate from model-specific training logic.
- Record the metric names and values with each run.

### Model artefacts and reports
- Save trained models using an appropriate serialization mechanism.
- Save experiment metadata as JSON.
- Export experiment summaries to CSV.
- Load a saved model and make predictions on new input data.
- Keep file paths configurable rather than hard-coded.

## 3. Suggested class design

| Class | Responsibility |
|---|---|
| `Dataset` | Dataset metadata and loading |
| `BaseModel` | Abstract model contract |
| `LinearRegressionModel` | Regression model implementation |
| `DecisionTreeModel` | Tree-based model implementation |
| `DataSplitter` | Training/testing split |
| `Metric` | Common metric contract or metric representation |
| `RegressionEvaluator` | Calculates regression metrics |
| `Experiment` | One configured model run and its result |
| `ExperimentManager` | Creates, runs and compares experiments |
| `ModelRepository` | Saves and loads model artefacts |
| `ReportExporter` | Exports experiment summaries |

These are suggested boundaries. Avoid creating classes that have no clear responsibility.

## 4. OOP concepts you must demonstrate

- **Abstraction:** define a `BaseModel` contract for training, prediction and saving.
- **Inheritance:** implement different model classes from the common model contract.
- **Polymorphism:** let the experiment manager call the same methods on different model implementations.
- **Encapsulation:** keep experiment state and transitions controlled.
- **Composition:** an experiment uses a dataset, model, evaluator and result object.
- **Open/closed design:** adding a new model should not require rewriting experiment orchestration.

## 5. Business and engineering rules

1. Experiment IDs must be unique.
2. A dataset must be validated before training starts.
3. Training and test data must be separated before model evaluation.
4. Evaluation must use the test set, not the training set.
5. A failed training run must be marked as failed and retain a useful error message.
6. Only completed experiments should appear in successful-model comparisons.
7. Store model parameters and metric values with the experiment metadata.
8. Do not overwrite an existing model artefact accidentally.
9. Use reproducible random seeds where the underlying library supports them.
10. Keep data loading, model training, evaluation, persistence and reporting as separate responsibilities.
11. Validate input feature columns before prediction.
12. Do not commit private datasets, credentials or large model artefacts to a public repository.

## 6. Required Python features

- Classes, inheritance and abstract base classes.
- Type hints for public methods.
- Dictionaries for model configurations and experiment metadata.
- Lists for experiment history.
- `pathlib` for filesystem paths.
- `datetime` for run timestamps.
- `json` and `csv` for metadata and reports.
- `logging` for run information.
- Exception handling for missing files and failed training.
- `pandas` and `scikit-learn` for a basic ML workflow.
- `joblib` or an appropriate library mechanism for model persistence.

## 7. Acceptance tests

- Register a valid CSV dataset and inspect its metadata.
- Attempt to register a missing file; produce a clear error.
- Create experiments using two different model implementations.
- Run both experiments through the same manager interface.
- Verify that both runs record their parameters and metrics.
- Force an invalid configuration and verify that the run is marked failed.
- Compare only completed experiments.
- Save a trained model and load it again.
- Verify that loaded-model predictions work with the expected feature columns.
- Export experiment summaries to CSV and metadata to JSON.
- Add a third model implementation without changing the experiment manager's core workflow.

## 8. Suggested development milestones

1. Design dataset and abstract model interfaces.
2. Implement two concrete model classes.
3. Build dataset validation and train/test splitting.
4. Implement evaluator and metric recording.
5. Build experiment lifecycle and manager.
6. Add model persistence and report export.
7. Add logging, tests and a clean package structure.

## 9. Expected project structure

A possible structure is:

```text
ml_experiment_manager/
│
├── app.py
├── config.py
├── models/
│   ├── base_model.py
│   ├── linear_regression.py
│   └── decision_tree.py
├── data/
│   ├── dataset.py
│   └── data_splitter.py
├── evaluation/
│   └── regression_evaluator.py
├── experiments/
│   ├── experiment.py
│   └── experiment_manager.py
├── storage/
│   └── model_repository.py
├── reports/
│   └── report_exporter.py
├── tests/
│   ├── test_models.py
│   ├── test_experiments.py
│   └── test_storage.py
└── README.md
```

You may change this structure when you have a reason. Keep the first version small and understandable.

## 10. Challenge extensions

- Add classification models and classification metrics.
- Add experiment tags and search filters.
- Add cross-validation.
- Add model versioning.
- Add a simple Streamlit interface.
- Add experiment comparison charts.
- Add a plugin-style model registry.
