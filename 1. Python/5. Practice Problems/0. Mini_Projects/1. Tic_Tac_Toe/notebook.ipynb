# Cell 9 — Winner Checking

Ye game ka **main logic function** hai.

### Function

`check_winner(board, player)`

Tic-Tac-Toe me total **8 winning combinations** hoti hain.

### Rows

```text
X | X | X
---------
  |   |
---------
  |   |
```

3 possible rows.

### Columns

```text
X |   |
---------
X |   |
---------
X |   |
```

3 possible columns.

### Diagonals

```text
X |   |
---------
  | X |
---------
  |   | X
```

and

```text
  |   | X
---------
  | X |
---------
X |   |
```

2 diagonals.

Therefore:

```text
3 rows
+ 3 columns
+ 2 diagonals
--------------
= 8 winning combinations
```

Function should return:

```text
True
```

if current player wins, otherwise:

```text
False
```

# Cell 10 — Draw Checking

### Function

`check_draw(board)`

Game draw kab hoga?


```text

```
