# map() Function in Python (Complete Detailed Notes)

The `map()` function is one of Python's most useful built-in functions.

It is used to **apply a function to every item of an iterable** (list, tuple, set, etc.) and return the transformed values.

---

# What is map()?

Suppose you have a list:

```python
numbers = [1, 2, 3, 4, 5]
```

And you want to square every element.

Without `map()`:

````python
numbers = [1,2,3,4,5]
result = []

for num in numbers:
    result.append(num * num)

print(result)
````

Using `map()`:

````python
numbers = [1,2,3,4,5]

square = list(map(lambda x: x*x, numbers))
print(square)
````

---

# Syntax

```python
map(function, iterable)
```

### Parameters

| Parameter | Description            |
| --------- | ---------------------- |
| function  | Function to apply      |
| iterable  | List, tuple, set, etc. |

### Return Value

Returns a **map object**.

To see results:

````python
list(map(...))
````

---

# How map() Works

Example:
```python
numbers = [1, 2, 3]
```

Code:

````python
result = map(lambda x: x*2, numbers)
````


Internally:

```python
1 → 2
2 → 4
3 → 6
```

Result:

```python
[2, 4, 6]
```

# Example 1: Using Normal Function


````python
def square(x):
    return x*x

numbers = [1,2,3,4,5]

result = map(square, numbers)

print(list(result))
````


---

# Example 2: Using Lambda Function

````python
numbers = [1,2,3,4,5]

result = list(map(lambda x:x*x, numbers))
print(result)
````


---

# Example 3: Double Every Number


````python
numbers = [10,20,30,40,50]

result = list(map(lambda x: x*2, numbers))
print(result)
````

# Example 4: Convert Strings to Uppercase


````python
names = ["gaurav","rahul","vikash","tarun","saim","dev"]

result = list(map(str.upper, names))

print(result)
````

````python
names = ['GAURAV', 'RAHUL', 'VIKASH', 'TARUN', 'SAIM', 'DEV']

result = list(map(len, names))
print(result)
````

# Example 6: Convert Strings into Integers


````python
numbers = ['6', '5', '6', '5', '4', '3']

result = list(map(int, numbers))

print(result)

result = map(str, numbers)

print(list(result))
````

# Multiple Iterables in map()

A very important concept.

`map()` can process multiple iterables simultaneously.

### Syntax

```python
map(function, iterable1, iterable2)
```

---

# Example 9: Add Two Lists


````python
list_1 = [1,2,3,4]
list_2 = [5,6,7,8]

result_addition = list(map(lambda x,y:x+y, list_1,list_2))

print(result_addition)

result_multiply = list(map(lambda x,y:x*y, list_1, list_2))

print(result_multiply)
````


---

# Important Note

If iterable lengths are different:

```python
a = [1,2,3,4]
b = [10,20]
```

````python
a = [1,2,3,4]
b = [10,20]

result = list(map(lambda x,y: x+y, a,b))

print(result)
````


Why?

Because `map()` stops at the shortest iterable.

# List Comprehension vs map()

Using map():

````python
numbers = [1,2,3,4]

result = list(map(lambda x:x*x, numbers))
````

Using List Comprehension:


````python
numbers = [1,2,3,4]

result = [x*x for x in numbers]
````



### Which is Better?

* Simple transformation → List comprehension
* Existing function available → map()
* Functional programming → map()

### Q5. Difference between map() and filter()?

| map()                   | filter()                 |
| ----------------------- | ------------------------ |
| Transforms data         | Filters data             |
| Returns modified values | Returns selected values  |
| One output per input    | May return fewer outputs |

Example:

```python
numbers = [1,2,3,4,5]
```

map:

```python
list(map(lambda x:x*2, numbers))
```

Output:

```python
[2,4,6,8,10]
```

filter:

```python
list(filter(lambda x:x%2==0, numbers))
```

Output:

```python
[2,4]
```
