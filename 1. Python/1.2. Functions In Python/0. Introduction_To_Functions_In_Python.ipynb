# Functions in Python

A **function** is a block of reusable code that performs a specific task. Functions help make programs modular, readable, and easier to maintain.

## Why Use Functions?

* Avoid code repetition
* Improve readability
* Make debugging easier
* Enable code reuse
* Break large programs into smaller parts

---

# Syntax of a Function

```python
def function_name(parameters):
    """Optional Docstring"""
    # Function body
    return value
```

### Example

````python
def greet():
    print("Hello, Welcome to the Functions in python.. ")

greet()
````

---

# Components of a Function

## 1. Function Definition

A function is defined using the `def` keyword.


## 2. Function Call

A function executes only when it is called.

````python
def greet(): ## <--- this is known as function definition using "def" keyword...
    print("Hello")

greet()   ## <--- this is known as the function call...
````


## 3. Parameters

Parameters are variables passed into a function.

````python
def greet(name): 
    print(f"Hello, {name}!")

name = input("Please Enter Your Name: ")
greet(name)   # <--- Here We Passed a Variable (name) Into a Function [greet(name)]
````


---

# Types of Functions

## 1. Built-in Functions

Functions already provided by Python.

Examples:

```python
print()
len()
type()
max()
min()
sum()
```

Example:

````python
numbers = [10, 20, 30]

print(len(numbers))
print(max(numbers))
````

---

## 2. User-defined Functions

Functions created by the programmer.

````python
def add():
    print(10 + 20)

add()
````

---

# Function Arguments

Arguments are values passed during function calls.

## 1. Positional Arguments


````python
def add(a, b):
    print(a + b)

add(10, 20)
````

---

## 2. Keyword Arguments

````python
def student(name, age):
    print(name, age)

student(age=20, name="Aman")
````

---

## 3. Default Arguments

````python
def greet(name="Guest"):
    print("Hello", name)

greet()
greet("Aman")
````


---

## 4. Variable-Length Arguments

### *args

Allows multiple positional arguments.

````python
def add(*numbers):
    print(sum(numbers))

add(10, 20, 30, 40)
````

---

### **kwargs

Allows multiple keyword arguments.

````python
def student(**details):
    print(details)

student(name="Aman", age=20)
````

---

# Return Statement

The `return` statement sends a result back to the caller.

````python
def add(a, b):
    return a + b

result = add(10, 20)

print(result)
````


---

# Function Without Return

````python
def add(a, b):
    print(a + b)

result = add(10, 20)

print(result)
````



Explanation:

* `print()` displays output.
* Function returns `None` by default.

# Local and Global Variables

## Local Variable

Accessible only inside the function.

````python
def demo():
    x = 10
    print(x)

demo()
````

---

## Global Variable

Accessible throughout the program.

````python
x = 100

def demo():
    print(x)

demo()
````

---

# Recursive Function

A function calling itself.

### Factorial Example

````python
def factorial(n):
    if n == 1:
        return 1
    return n * factorial(n - 1)

print(factorial(5))
````

---

# Lambda Function

Anonymous one-line function.

### Syntax

```python
lambda arguments : expression
```

Example:

````python
square = lambda x: x * x

print(square(5))
````

---

# Nested Functions

Function inside another function.

````python
def outer():
    
    def inner():
        print("Inner Function")
    
    inner()

outer()
````


---

# Docstrings

Used to describe a function.

````python
def add(a, b):
    """
    Returns the sum of two numbers.
    """
    return a + b

print(add.__doc__)
````



---

# Scope of Variables

| Scope     | Accessible           |
| --------- | -------------------- |
| Local     | Inside function only |
| Global    | Entire program       |
| Enclosing | Outer function       |
| Built-in  | Python predefined    |

This is known as the **LEGB Rule**:

* L → Local
* E → Enclosing
* G → Global
* B → Built-in

# Practical Examples

### Example 1: Even or Odd

````python
def Even_and_Odd(num):
    for i in range(num):
        if i % 2 == 0:
            print(f"Even: {i}")
        else:
            print(f"Odd: {i}")

num = int(input("Enter a number: "))
Even_and_Odd(num)
````


---

## Example 2: Largest Number among three numbers

````python
def largest(num1,num2, num3):
    return num1 if num1>num2 and num1 > num3 else(num2 if num2>num3 else num3)

num1 = int(input("Enter the first number: "))
num2 = int(input("Enter the second number: "))
num3 = int(input("Enter the second number: "))
result = largest(num1, num2, num3)
print(f"Largest Number is: {result}")
````
