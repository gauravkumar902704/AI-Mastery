# 2. PostgreSQL Data Types

Data type decide karta hai ki **column kis type ki value store karega**.

## A. Numeric Data Types

| Data Type          |  Storage | Kis type ki value?                   | Example      |
| ------------------ | -------: | ------------------------------------ | ------------ |
| `SMALLINT`         |  2 bytes | Small integers                       | `100`, `-50` |
| `INTEGER` / `INT`  |  4 bytes | Normal integers                      | `25000`      |
| `BIGINT`           |  8 bytes | Very large integers                  | `9000000000` |
| `NUMERIC(p,s)`     | Variable | Exact decimal values                 | `99999.99`   |
| `DECIMAL(p,s)`     | Variable | `NUMERIC` ka equivalent              | `4500.50`    |
| `REAL`             |  4 bytes | Approximate decimal                  | `12.345`     |
| `DOUBLE PRECISION` |  8 bytes | Higher precision approximate decimal | `12345.6789` |

### Integer ranges

| Type       | Approximate range           |
| ---------- | --------------------------- |
| `SMALLINT` | -32,768 to 32,767           |
| `INTEGER`  | -2.1 billion to 2.1 billion |
| `BIGINT`   | Very large integer range    |

### NUMERIC(p,s)

```sql
NUMERIC(10,2)
```

* `10` = total maximum digits
* `2` = decimal ke baad digits

Example:

```text
12345678.90
```

Total = 10 digits.

### Use cases

```text
Age          → INTEGER
Quantity      → INTEGER
Population    → BIGINT
Product price → NUMERIC(10,2)
Salary        → NUMERIC(12,2)
```

**Financial values ke liye `NUMERIC/DECIMAL` prefer karo**, approximate floating-point types nahi.

# 3. Character / String Data Types

| Data Type    | Meaning                                      | Example          | Use                |
| ------------ | -------------------------------------------- | ---------------- | ------------------ |
| `CHAR(n)`    | Fixed-length string                          | `'ABC'`          | Fixed-size values  |
| `VARCHAR(n)` | Variable-length string with limit            | `'Gaurav Kumar'` | Names, emails etc. |
| `TEXT`       | Variable-length text without specified limit | Long description | Large/general text |

### Examples

```sql
name VARCHAR(100)
email VARCHAR(150)
description TEXT
country_code CHAR(2)
```

### Difference

| `CHAR`           | `VARCHAR`                | `TEXT`              |
| ---------------- | ------------------------ | ------------------- |
| Fixed length     | Variable length          | Variable length     |
| Length specified | Maximum length specified | No declared maximum |
| Specialized use  | Common                   | Common              |

For most normal application text:

```sql
VARCHAR(100)
```

or

```sql
TEXT
```

is sufficient.

# 5. Date & Time Data Types

| Data Type     | Stores                            | Example                     |
| ------------- | --------------------------------- | --------------------------- |
| `DATE`        | Date only                         | `2026-09-21`                |
| `TIME`        | Time only                         | `14:30:00`                  |
| `TIMESTAMP`   | Date + time                       | `2026-09-21 14:30:00`       |

# 11. Constraints

Constraints database mein **rules enforce** karte hain.

| Constraint    | Purpose                      |
| ------------- | ---------------------------- |
| `NOT NULL`    | Value compulsory             |
| `UNIQUE`      | Duplicate values prevent     |
| `PRIMARY KEY` | Row uniquely identify        |
| `FOREIGN KEY` | Tables ke beech relationship |
| `CHECK`       | Condition enforce            |
| `DEFAULT`     | Value automatically provide  |

---

## Constraint Syntax

```sql
CREATE TABLE table_name (
    column datatype CONSTRAINT
);
```

Example:

```sql
CREATE TABLE students (
    student_id INTEGER PRIMARY KEY,
    name VARCHAR(100) NOT NULL,
    email VARCHAR(150) UNIQUE,
    age INTEGER CHECK (age >= 18),
    city VARCHAR(50) DEFAULT 'Delhi'
);
```

---

# 12. NOT NULL

Value missing nahi ho sakti.

```sql
name VARCHAR(100) NOT NULL
```

Allowed:

```text
Gaurav
```

Not allowed:

```text
NULL
```

Use when a field is mandatory.

# 13. UNIQUE

Duplicate values prevent karta hai.

```sql
email VARCHAR(150) UNIQUE
```

Allowed:

```text
a@gmail.com
b@gmail.com
```

Not allowed:

```text
a@gmail.com
a@gmail.com
```

**Note:** PostgreSQL's `UNIQUE` constraint normally allows multiple `NULL` values because `NULL` values are not considered equal in the usual uniqueness comparison.

---

# 14. PRIMARY KEY

A primary key uniquely identifies every row.

```sql
student_id INTEGER PRIMARY KEY
```

Primary key:

* Unique hona chahiye
* `NULL` nahi ho sakta
* Table mein normally one primary-key constraint hota hai
* One or multiple columns ka combination bhi primary key ho sakta hai

Example:

```sql
CREATE TABLE students (
    student_id INTEGER PRIMARY KEY,
    name VARCHAR(100)
);
```

---

# 15. FOREIGN KEY

Foreign key ek table ko doosre table se connect karta hai.

### Parent table

```sql
CREATE TABLE customers (
    customer_id INTEGER PRIMARY KEY,
    name VARCHAR(100)
);
```

### Child table

```sql
CREATE TABLE orders (
    order_id INTEGER PRIMARY KEY,
    customer_id INTEGER,
    FOREIGN KEY (customer_id)
        REFERENCES customers(customer_id)
);
```

Relationship:

```text
customers
   │
   │ customer_id
   ↓
orders
```

Agar customer exist nahi karta, to corresponding foreign-key value normally insert nahi kar sakte.

# 20. Creating a Table

General syntax:

```sql
CREATE TABLE table_name (
    column_name datatype constraint,
    column_name datatype constraint,
    ...
);
```

```sql
CREATE TABLE products (
    product_id INTEGER
        GENERATED ALWAYS AS IDENTITY
        PRIMARY KEY,

    product_name VARCHAR(100) NOT NULL,

    category VARCHAR(50) NOT NULL,

    price NUMERIC(10,2)
        CHECK (price >= 0),

    stock INTEGER
        CHECK (stock >= 0),

    is_active BOOLEAN DEFAULT TRUE
);
```

# 26. WHERE vs HAVING

| `WHERE`                                               | `HAVING`                                    |
| ----------------------------------------------------- | ------------------------------------------- |
| Individual rows filter karta hai                      | Groups filter karta hai                     |
| `GROUP BY` se pehle logically apply hota hai          | `GROUP BY` ke baad logically apply hota hai |
| Normal conditions                                     | Aggregate/group conditions commonly         |
| `COUNT()` etc. par generally directly based nahi hota | Aggregate conditions ke liye commonly used  |

### Example

```sql
SELECT department, AVG(salary)
FROM employees
WHERE salary > 30000
GROUP BY department
HAVING AVG(salary) > 50000;
```

Logical idea:

```text
1. salary > 30000 wale rows
        ↓
2. department-wise group
        ↓
3. average salary calculate
        ↓
4. average > 50000 wale groups
```

# 27. CASE

SQL mein conditional logic ke liye.

### Syntax

```sql
CASE
    WHEN condition THEN result
    WHEN condition THEN result
    ELSE result
END
```

Example:

```sql
SELECT
    name,
    salary,
    CASE
        WHEN salary >= 80000 THEN 'High'
        WHEN salary >= 50000 THEN 'Medium'
        ELSE 'Low'
    END AS salary_level
FROM employees;
```

```text
name   | salary | salary_level
-------+--------+-------------
Rahul  | 90000  | High
Aman   | 60000  | Medium
Ravi   | 30000  | Low
```

# 4. INNER JOIN

# 6. LEFT JOIN

**Left table ki saari rows** return karta hai.

Matching right-table rows milti hain; otherwise `NULL`.

# 7. RIGHT JOIN

Right table ki saari rows return karta hai.

# 14. JOIN + GROUP BY

# 15. LEFT JOIN + COUNT

**All customers + number of orders**

```sql
SELECT
    c.customer_id,
    c.name,
    COUNT(o.order_id) AS order_count
FROM customers c
LEFT JOIN orders o
    ON c.customer_id = o.customer_id
GROUP BY c.customer_id, c.name;
```

Why `LEFT JOIN`?

Because we also want customers who have **zero orders**.

Important:

```sql
COUNT(o.order_id)
```

is generally preferable here to:

```sql
COUNT(*)
```

because `COUNT(o.order_id)` doesn't count the NULL-extended row for customers with no orders.

# 16. Finding Customers With No Orders

# 17. JOIN Condition vs WHERE Condition

Important difference.

### Condition in ON

```sql
SELECT *
FROM customers c
LEFT JOIN orders o
    ON c.customer_id = o.customer_id
    AND o.amount > 50000;
```

Yahan customers remain karenge; sirf matching orders ₹50,000+ honge.

### Condition in WHERE

```sql
SELECT *
FROM customers c
LEFT JOIN orders o
    ON c.customer_id = o.customer_id
WHERE o.amount > 50000;
```

Yahan `WHERE` NULL orders ko remove kar dega, so behavior can become similar to an inner join for that condition.

**Interview point:** `LEFT JOIN` ke saath `ON` vs `WHERE` placement result ko significantly change kar sakta hai.

# 1. Window Functions

### What is a Window Function?

Window function kisi group of rows par calculation karta hai **without reducing the number of rows**.

Normal `GROUP BY`:

```sql
SELECT department, AVG(salary)
FROM employees
GROUP BY department;
```

Output mein har department ki **ek row** milegi.

Window function:

```sql
SELECT
    employee_name,
    department,
    salary,
    AVG(salary) OVER (PARTITION BY department) AS dept_avg_salary
FROM employees;
```

Output mein **har employee ki row bhi rahegi** aur uske department ka average bhi milega.

### Basic Syntax

```sql
function_name() OVER (
    PARTITION BY column
    ORDER BY column
)
```

| Part           | Meaning                                        |
| -------------- | ---------------------------------------------- |
| `OVER()`       | Window function define karta hai               |
| `PARTITION BY` | Data ko groups mein divide karta hai           |
| `ORDER BY`     | Window ke andar rows ka order define karta hai |

---

## 2. ROW_NUMBER()

Har row ko unique sequential number deta hai.

```sql
SELECT
    employee_name,
    department,
    salary,
    ROW_NUMBER() OVER (
        ORDER BY salary DESC
    ) AS row_num
FROM employees;
```

Example:

| employee | salary | row_num |
| -------- | -----: | ------: |
| A        | 100000 |       1 |
| B        |  90000 |       2 |
| C        |  90000 |       3 |
| D        |  80000 |       4 |

Equal salary hone par bhi number different hoga.

### Department-wise numbering

```sql
SELECT
    employee_name,
    department,
    salary,
    ROW_NUMBER() OVER (
        PARTITION BY department
        ORDER BY salary DESC
    ) AS row_num
FROM employees;
```

---

# 3. RANK() vs DENSE_RANK()

Ye interview mein bahut common hai.

### RANK()

```sql
SELECT
    employee_name,
    salary,
    RANK() OVER (
        ORDER BY salary DESC
    ) AS rank
FROM employees;
```

Agar salaries hain:

```text
100000
90000
90000
80000
```

Result:

```text
1
2
2
4
```

### DENSE_RANK()

```sql
SELECT
    employee_name,
    salary,
    DENSE_RANK() OVER (
        ORDER BY salary DESC
    ) AS rank
FROM employees;
```

Result:

```text
1
2
2
3
```

### Difference

| Function       | Equal values    | Next rank |
| -------------- | --------------- | --------- |
| `ROW_NUMBER()` | Separate number | 4         |
| `RANK()`       | Same rank       | Gap       |
| `DENSE_RANK()` | Same rank       | No gap    |

---

# 4. Top N Records

Suppose humein har department ke **top 2 highest-paid employees** chahiye.

```sql
WITH ranked_employees AS (
    SELECT
        employee_name,
        department,
        salary,
        ROW_NUMBER() OVER (
            PARTITION BY department
            ORDER BY salary DESC
        ) AS rn
    FROM employees
)
SELECT *
FROM ranked_employees
WHERE rn <= 2;
```

### Important pattern

```text
Window Function
       ↓
     CTE
       ↓
    Filter
```

Window function ke result ko generally directly `WHERE` mein use nahi kar sakte, isliye CTE/subquery useful hota hai.

# 5. LAG() and LEAD()

Ye functions previous/next row ki value access karne ke liye use hote hain.

### LAG()

Previous row:

```sql
SELECT
    sale_date,
    sales,
    LAG(sales) OVER (
        ORDER BY sale_date
    ) AS previous_sales
FROM daily_sales;
```

Example:

| date  | sales | previous_sales |
| ----- | ----: | -------------: |
| Jan 1 | 10000 |           NULL |
| Jan 2 | 12000 |          10000 |
| Jan 3 | 15000 |          12000 |

### Sales difference

```sql
SELECT
    sale_date,
    sales,
    sales - LAG(sales) OVER (
        ORDER BY sale_date
    ) AS sales_difference
FROM daily_sales;
```

---

### LEAD()

Next row ki value:

```sql
SELECT
    sale_date,
    sales,
    LEAD(sales) OVER (
        ORDER BY sale_date
    ) AS next_sales
FROM daily_sales;
```

### Real-world uses

`LAG()`:

* Previous month's revenue
* Previous day's stock price
* Previous order
* Previous salary

`LEAD()`:

* Next transaction
* Next login
* Next purchase
* Next scheduled event

# 6. Running Total

Suppose daily sales:

```text
Jan 1 → 1000
Jan 2 → 2000
Jan 3 → 1500
```

Running total:

```text
1000
3000
4500
```

Query:

```sql
SELECT
    sale_date,
    sales,
    SUM(sales) OVER (
        ORDER BY sale_date
    ) AS running_total
FROM daily_sales;
```

# 7. Moving Average

For example, 3-day moving average:

```sql
SELECT
    sale_date,
    sales,
    AVG(sales) OVER (
        ORDER BY sale_date
        ROWS BETWEEN 2 PRECEDING AND CURRENT ROW
    ) AS moving_avg
FROM daily_sales;
```

# 8. Aggregate Function vs Window Function

| Aggregate                 | Window               |
| ------------------------- | -------------------- |
| `GROUP BY` commonly used  | `OVER()` required    |
| Rows reduce ho jaati hain | Rows remain          |
| One result per group      | Result for every row |
| `SUM()`                   | `SUM() OVER()`       |
| `AVG()`                   | `AVG() OVER()`       |
| `COUNT()`                 | `COUNT() OVER()`     |

Example:

### Aggregate

```sql
SELECT department, AVG(salary)
FROM employees
GROUP BY department;
```

### Window

```sql
SELECT
    employee_name,
    department,
    AVG(salary) OVER (
        PARTITION BY department
    )
FROM employees;
```

**Interview point:** Window functions calculate across related rows while preserving individual rows.

# 12. Recursive CTE

Hierarchical data ke liye useful hai.

Example:

```text
CEO
 ├── Manager A
 │    ├── Employee 1
 │    └── Employee 2
 └── Manager B
      └── Employee 3
```

Basic structure:

```sql
WITH RECURSIVE employee_tree AS (

    -- Starting point
    SELECT
        employee_id,
        employee_name,
        manager_id
    FROM employees
    WHERE manager_id IS NULL

    UNION ALL

    -- Recursive part
    SELECT
        e.employee_id,
        e.employee_name,
        e.manager_id
    FROM employees e
    JOIN employee_tree et
        ON e.manager_id = et.employee_id
)

SELECT *
FROM employee_tree;
```

Common uses:

* Employee hierarchy
* Category/subcategory
* Folder structure
* Organization trees
* Parent-child relationships

# 22. Transactions

Transaction = multiple database operations ko **ek logical unit** ke form mein execute karna.

Example: Bank transfer.

₹10,000:

```text
Account A → -₹10,000
Account B → +₹10,000
```

Dono operations successful hone chahiye.

Agar first operation successful ho gaya aur second fail ho gaya, database inconsistent ho sakta hai.

Isliye transaction use karte hain.

---

# 23. BEGIN, COMMIT, ROLLBACK

### Start transaction

```sql
BEGIN;
```

Operations:

```sql
UPDATE accounts
SET balance = balance - 10000
WHERE account_id = 1;

UPDATE accounts
SET balance = balance + 10000
WHERE account_id = 2;
```

Everything successful:

```sql
COMMIT;
```

Agar problem:

```sql
ROLLBACK;
```

---

# 24. SAVEPOINT

Transaction ke andar intermediate checkpoint bana sakte ho.

```sql
BEGIN;

UPDATE accounts
SET balance = balance - 10000
WHERE account_id = 1;

SAVEPOINT transfer_step;

UPDATE accounts
SET balance = balance + 10000
WHERE account_id = 2;
```

Agar second operation ko undo karna ho:

```sql
ROLLBACK TO SAVEPOINT transfer_step;
```

Transaction completely cancel nahi hoti.

Finally:

```sql
COMMIT;
```

# 25. ACID Properties

Transactions ka most important interview topic.

| Property        | Meaning                                                            |
| --------------- | ------------------------------------------------------------------ |
| **Atomicity**   | Transaction ke all operations successful ya none                   |
| **Consistency** | Database valid state mein rahe                                     |
| **Isolation**   | Concurrent transactions ek dusre ko improperly interfere na karein |
| **Durability**  | `COMMIT` ke baad changes persist rahen                             |

### Easy trick:

**A C I D**

> **A**ll or Nothing
> **C**orrect State
> **I**ndependent Transactions
> **D**ata Persists

# 6. Trigger Function

Pehle trigger function:

```sql
CREATE FUNCTION log_employee_update()
RETURNS TRIGGER
LANGUAGE plpgsql
AS $$
BEGIN

    INSERT INTO employee_audit (
        employee_id,
        action_time
    )
    VALUES (
        NEW.employee_id,
        CURRENT_TIMESTAMP
    );

    RETURN NEW;

END;
$$;
```

Then trigger:

```sql
CREATE TRIGGER employee_update_trigger
AFTER UPDATE ON employees
FOR EACH ROW
EXECUTE FUNCTION log_employee_update();
```

# 7. NEW and OLD

```sql
NEW.salary
```

# 26. Real-World Interview Queries

### Q1. Each customer's total spending

```sql
SELECT
    c.customer_id,
    c.name,
    SUM(oi.quantity * oi.price) AS total_spending
FROM customers c
JOIN orders o
    ON c.customer_id = o.customer_id
JOIN order_items oi
    ON o.order_id = oi.order_id
GROUP BY c.customer_id, c.name
ORDER BY total_spending DESC;
```

---

### Q2. Top 3 customers by spending

```sql
SELECT
    c.customer_id,
    c.name,
    SUM(oi.quantity * oi.price) AS total_spending
FROM customers c
JOIN orders o
    ON c.customer_id = o.customer_id
JOIN order_items oi
    ON o.order_id = oi.order_id
GROUP BY c.customer_id, c.name
ORDER BY total_spending DESC
LIMIT 3;
```

### Q3. Products never ordered

```sql
SELECT
    p.product_id,
    p.product_name
FROM products p
LEFT JOIN order_items oi
    ON p.product_id = oi.product_id
WHERE oi.product_id IS NULL;
```

### Q6. Customers with more than 5 orders

```sql
SELECT
    customer_id,
    COUNT(*) AS order_count
FROM orders
GROUP BY customer_id
HAVING COUNT(*) > 5;
```

### Q7. Second-highest product price

```sql
SELECT MAX(price)
FROM products
WHERE price < (
    SELECT MAX(price)
    FROM products
);
```

### Q8. Top 2 products in every category

```sql
WITH ranked_products AS (
    SELECT
        product_id,
        product_name,
        category,
        price,
        ROW_NUMBER() OVER (
            PARTITION BY category
            ORDER BY price DESC
        ) AS rn
    FROM products
)
SELECT *
FROM ranked_products
WHERE rn <= 2;
```
