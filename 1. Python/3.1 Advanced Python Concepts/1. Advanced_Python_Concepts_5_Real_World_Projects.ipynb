# Advanced Python Concepts — 5 Real-World Projects

## Practical implementation workbook

This workbook is designed to strengthen Python language features that are often used in professional codebases but are not fully practised through basic programs.

The five projects progressively apply:

- `*args` and `**kwargs`
- Decorators and decorator factories
- Closures and higher-order functions
- Iterators and generators
- Generator expressions and lazy evaluation
- Context managers (`with`, `__enter__`, `__exit__`, `contextlib`)
- Comprehensions and unpacking
- `lambda`, `map`, `filter`, `sorted` and `zip`
- `any`, `all`, `enumerate` and `itertools`
- Exception handling and custom exceptions
- Type hints, dataclasses and enums
- File handling, JSON, CSV and `pathlib`
- Logging, timing, validation and reusable utilities

These projects focus on practical use, not using a feature just to show that it exists.

## How to use this workbook

1. Read the full project brief before writing code.
2. Identify where each advanced Python feature genuinely improves the design.
3. Write a small design or flow before implementation.
4. Build a working basic version, then add the advanced features.
5. Test normal inputs, invalid inputs, empty data and large data.
6. Keep functions focused and avoid unnecessary cleverness.
7. Add type hints and docstrings to important public functions.
8. Do not copy a complete solution. Implement and explain each design decision yourself.

# Project 1 — Logging, Validation & Monitoring Toolkit

<badge>Intermediate</badge>

## 1. Problem statement

Create a reusable Python toolkit that developers can add to their own applications to monitor function execution. It should log function calls, measure execution time, validate inputs and capture failures without changing the original business logic of the functions being monitored.

For example, a developer should be able to apply a decorator to a function that calculates an invoice total, and the toolkit should record its execution time and outcome automatically.

## 2. Core requirements

### Execution logging
- Record the function name, start time, end time and execution outcome.
- Capture positional and keyword arguments in a safe, readable form.
- Do not log sensitive values such as passwords, API keys or tokens.
- Record exceptions and re-raise them; do not silently hide failures.
- Allow the logging destination to be configured.

### Execution timing
- Create a decorator that measures function execution duration.
- Support functions that return values.
- Ensure the decorator returns the original function's result unchanged.
- Use a monotonic clock for duration measurement.

### Input validation
- Create reusable validation decorators.
- Validate required keyword arguments and permitted value ranges.
- Allow validation rules to be configured when applying a decorator.
- Raise clear custom exceptions when validation fails.

### Function compatibility
- Decorated functions must retain useful metadata such as their original name and docstring.
- Support functions with different signatures.
- Support a function receiving both positional and keyword arguments.

## 3. Suggested design

| Component | Responsibility |
|---|---|
| `ExecutionLogger` | Writes structured execution events |
| `ExecutionTimer` | Stores or reports duration |
| `ValidationError` | Represents invalid input |
| `log_execution` | Decorator for function-call logging |
| `measure_time` | Decorator for execution timing |
| `validate_arguments` | Decorator factory for configurable validation |
| `MonitoringReport` | Summarises function execution statistics |

You may use functions instead of classes where a class would add no value.

## 4. Concepts you must demonstrate

- **`*args` and `**kwargs`:** wrappers must forward any positional and keyword arguments to the original function.
- **Decorators:** implement logging and timing without rewriting business functions.
- **Decorator factory:** allow validation rules or logging settings to be passed to a decorator.
- **Closures:** retain decorator configuration between decoration and function execution.
- **Higher-order functions:** accept and return function objects.
- **`functools.wraps`:** preserve function metadata.
- **Exception handling:** record an error and re-raise the original exception.

## 5. Functional requirements and rules

1. A wrapper must pass all received arguments to the wrapped function correctly.
2. A wrapper must return the original return value.
3. A wrapper must not swallow exceptions.
4. Timing must be recorded for both successful and failed executions.
5. Decorators must work when stacked together.
6. Validation must run before the business function executes.
7. Sensitive argument values must be masked or excluded from logs.
8. Logging failures should not silently change the business result; define and document your chosen failure policy.
9. The toolkit must work with functions that have no arguments, positional arguments, keyword arguments and mixed arguments.

## 6. Required Python features

- `*args`, `**kwargs` and argument unpacking.
- Nested functions and closures.
- Decorators and decorator factories.
- `functools.wraps`.
- `time.perf_counter()` or `time.monotonic()`.
- `logging`.
- Custom exceptions.
- Dictionaries for configuration and statistics.
- Type hints for public interfaces.

## 7. Acceptance tests

- Decorate a function with no parameters and verify its result is unchanged.
- Decorate a function with positional and keyword arguments.
- Verify that the original function name and docstring remain available.
- Stack timing and logging decorators and verify both execute.
- Cause the wrapped function to raise an exception; verify it is logged and re-raised.
- Configure validation to reject a negative amount.
- Verify that a sensitive keyword such as `password` is not written in plain text to logs.
- Verify that timing is recorded when a function fails.

## 8. Suggested milestones

1. Build a basic wrapper using `*args` and `**kwargs`.
2. Add `functools.wraps`.
3. Implement execution timing.
4. Implement logging and exception recording.
5. Create a configurable validation decorator.
6. Test decorator stacking and edge cases.

## 9. Expected deliverables

- Reusable Python package.
- At least three decorators: logging, timing and validation.
- Example module demonstrating usage.
- Automated tests for argument forwarding, metadata, exceptions and decorator stacking.
- README explaining decorator behaviour and configuration.

## 10. Challenge extensions

- Add a retry decorator with configurable retry count.
- Add a decorator that caches results for selected functions.
- Add a decorator that restricts access based on a user role.
- Add a decorator that records call counts and average duration.
- Support asynchronous functions using `async def` wrappers.

**Your task:** Build decorators that can wrap different functions safely. The wrapper must preserve the original function's behaviour and interface as much as practical.

# Project 2 — Flexible Report & Data Processing Engine

<badge>Intermediate</badge>

## 1. Problem statement

Build a data-processing utility that reads large sales or transaction datasets and generates summaries without loading the entire dataset into memory.

The system should accept different data sources, transform records, filter invalid entries and produce reports. The same processing pipeline should work with different iterable inputs.

## 2. Core requirements

### Data input
- Read records from CSV files.
- Accept lists, tuples or other iterable sources.
- Validate required fields such as transaction ID, date, category and amount.
- Handle missing or malformed records according to a documented policy.
- Keep the input layer separate from report calculations.

### Processing pipeline
- Filter records by date range, category or minimum amount.
- Transform raw records into a consistent representation.
- Calculate total sales, average transaction value and category totals.
- Find the highest-value transactions.
- Group records by month or category.

### Reporting
- Generate a summary report.
- Export filtered records to CSV or JSON.
- Support configurable report columns.
- Process records lazily where possible.

## 3. Suggested design

| Component | Responsibility |
|---|---|
| `Transaction` | Represents a validated transaction |
| `CSVRecordReader` | Iterates over CSV records |
| `RecordValidator` | Checks required fields and values |
| `TransactionPipeline` | Coordinates filtering and transformation |
| `SalesAggregator` | Calculates totals and grouped metrics |
| `ReportExporter` | Writes reports to CSV or JSON |

## 4. Concepts you must demonstrate

- **Iterables and iterators:** distinguish a collection from an object that produces values one at a time.
- **Custom iterator:** implement `__iter__()` and `__next__()` for a paginated or batched record source.
- **Generators:** use `yield` to produce validated records lazily.
- **Generator expressions:** create lazy filters or transformations.
- **`yield from`:** delegate iteration to another iterable where useful.
- **Comprehensions:** build compact dictionaries and sets for summaries.
- **Functional tools:** use `filter`, `map`, `sorted`, `zip`, `enumerate`, `any` and `all` where they make the code clearer.
- **`*args` and `**kwargs`:** allow configurable filter or report options without hard-coding every combination.

## 5. Functional requirements and rules

1. The processing pipeline must accept any suitable iterable, not only a list.
2. Large input files must be processed incrementally.
3. A record with an invalid amount must not corrupt the totals.
4. Filtering must not mutate the original input records.
5. Empty input must produce a valid empty report.
6. Sorting should be performed only when requested, because it may require storing records.
7. Report columns must be configurable.
8. The pipeline should allow multiple filters to be combined.
9. Document whether invalid rows are skipped, collected as errors or cause the process to stop.

## 6. Required Python features

- Generator functions and generator expressions.
- Custom iterator protocol.
- `itertools` such as `islice`, `chain`, `groupby` or `tee` where appropriate.
- `collections.defaultdict` or `Counter`.
- List, dictionary and set comprehensions.
- `csv`, `json` and `pathlib`.
- `Decimal` for financial values.
- Type hints and dataclasses.
- `*args` and `**kwargs` for flexible configuration.

## 7. Acceptance tests

- Process a list of transaction dictionaries.
- Process a CSV file with at least several thousand rows.
- Verify that the generator yields one validated transaction at a time.
- Filter transactions by date and category.
- Calculate total and average transaction value correctly.
- Handle an empty input source.
- Handle malformed rows without producing incorrect totals.
- Export a report with only the requested columns.
- Verify that a pipeline can be run on a generator source without requiring it to be converted into a list first.

## 8. Suggested milestones

1. Define a transaction representation and validation rules.
2. Implement CSV reading as an iterator.
3. Convert raw records using a generator.
4. Add reusable filter functions.
5. Implement aggregation and report generation.
6. Add export formats and large-input tests.

## 9. Expected deliverables

- Data-processing package.
- CSV sample dataset with test records.
- Lazy record reader and transformation pipeline.
- Sales summary report in JSON or CSV.
- Automated tests for iterator behaviour and aggregation.
- README with examples of passing different iterable sources.

## 10. Challenge extensions

- Add streaming input from a simulated event source.
- Add batch processing with configurable batch size.
- Add a pipeline that can be paused and resumed.
- Add duplicate transaction detection.
- Compare memory usage of eager and lazy processing.

**Your task:** Build a pipeline that can process a large source incrementally. Avoid converting every generator into a list just for convenience.

# Project 3 — File Processing, Compression & Backup Utility

<badge>Intermediate–Advanced</badge>

## 1. Problem statement

Create a command-line utility that scans a directory, finds files matching user-defined rules, processes them and creates a backup. It should handle file resources safely, report progress and recover cleanly from errors.

The project should demonstrate why context managers are useful and how generators make file-processing workflows memory-efficient.

## 2. Core requirements

### Directory scanning
- Accept a root directory and optional file extensions.
- Scan files recursively.
- Ignore configured folders such as `.git`, `venv` and `__pycache__`.
- Provide file name, extension, size and modified time.
- Handle missing directories and permission errors.

### File processing
- Read large text or CSV files line by line.
- Calculate file size and optionally a checksum.
- Copy selected files to a backup directory.
- Avoid overwriting an existing backup accidentally.
- Support a dry-run mode that reports planned actions without modifying files.

### Backup reporting
- Produce a manifest containing source path, backup path, size, checksum and status.
- Export the manifest as JSON or CSV.
- Report successful, skipped and failed operations separately.

## 3. Suggested design

| Component | Responsibility |
|---|---|
| `DirectoryScanner` | Discovers files lazily |
| `FileMetadata` | Stores file details |
| `ManagedFile` | Provides controlled file access |
| `BackupManager` | Coordinates backup operations |
| `ChecksumCalculator` | Calculates file checksums incrementally |
| `BackupManifest` | Stores operation results |
| `BackupReport` | Exports a summary |

## 4. Concepts you must demonstrate

- **Context manager protocol:** implement a class with `__enter__()` and `__exit__()`.
- **`contextlib.contextmanager`:** implement a second context manager using a generator and `yield`.
- **Generators:** scan directories and read files incrementally.
- **Iterators:** expose a stream of discovered file records.
- **Closures:** create configurable file filters if useful.
- **`*args` and `**kwargs`:** support flexible scan or backup options.
- **Exception handling:** handle file-not-found, permission and I/O errors.
- **`pathlib`:** use path objects instead of manual path-string concatenation.

## 5. Functional requirements and rules

1. Every opened file must be closed, including when an exception occurs.
2. Directory scanning must not load all file metadata into memory unnecessarily.
3. A dry run must not create, overwrite or delete files.
4. Existing backups must be handled according to an explicit overwrite policy.
5. A failed file copy must be recorded as failed.
6. A checksum must be calculated incrementally for large files.
7. The utility must not follow symbolic links unless the user explicitly enables that behaviour.
8. Source files must never be deleted as part of the backup operation.
9. Use temporary output files and a safe replacement strategy where appropriate.
10. Clearly report partial failures instead of claiming the entire backup succeeded.

## 6. Required Python features

- `pathlib.Path`.
- `with open(...)`.
- Custom context manager class.
- `contextlib.contextmanager`.
- Generator functions and `yield`.
- `hashlib` for checksums.
- `shutil` for file copying.
- `json` and `csv` for manifests.
- `logging` and custom exceptions.
- `argparse` for command-line arguments.

## 7. Acceptance tests

- Scan a directory containing nested folders.
- Filter only `.csv` and `.txt` files.
- Confirm ignored folders are not scanned.
- Process a large file in chunks.
- Raise an exception inside a managed file block and verify the file is still closed.
- Run a dry run and verify that no backup files are created.
- Back up a file and verify its checksum matches the source.
- Attempt to back up to an existing destination and verify the overwrite policy.
- Simulate a missing source file and confirm the failure appears in the manifest.
- Run the utility against an empty directory.

## 8. Suggested milestones

1. Build directory scanning with `pathlib`.
2. Convert scanning and reading into generators.
3. Implement a custom file context manager.
4. Implement a generator-based context manager.
5. Add backup and checksum functionality.
6. Add dry-run mode, manifest export and tests.

## 9. Expected deliverables

- Command-line backup utility.
- Separate scanning, file-access, backup and reporting modules.
- JSON/CSV manifest.
- Tests proving resources close after success and failure.
- README with command examples and overwrite policy.

## 10. Challenge extensions

- Add ZIP archive creation.
- Add backup retention rules.
- Add incremental backup based on checksum or modification time.
- Add progress reporting.
- Add a restore command with path validation.

**Your task:** Demonstrate safe resource handling. Your context manager must release resources even if the code inside the `with` block fails.

# Project 4 — API Request Simulator & Retry Framework

<badge>Advanced</badge>

## 1. Problem statement

Build a reusable request-execution framework that simulates calling external APIs. It should support configurable request options, retry temporary failures, enforce rate limits, measure execution time and log request outcomes.

No real API key or external service is required. Use a simulated request function so you can test success, failure and timeout scenarios safely.

## 2. Core requirements

### Flexible request function
- Accept an endpoint, HTTP-like method, headers, query parameters and payload.
- Accept optional configuration values such as timeout and retry count.
- Support additional keyword options without changing the function signature every time.
- Validate required fields and reject unsupported methods.

### Retry mechanism
- Retry temporary failures up to a configurable maximum.
- Do not retry permanent validation errors.
- Support a delay between retries.
- Record attempt count and final outcome.
- Allow retry rules to be configured per function.

### Request monitoring
- Record duration, status, attempt count and error details.
- Log request metadata while masking credentials.
- Provide a decorator that applies request timing and logging.
- Allow decorators to be stacked without losing function metadata.

### Rate limiting
- Simulate a limit on the number of requests in a time window.
- Reject or delay requests when the configured limit is reached.
- Keep rate-limit logic separate from request execution.

## 3. Suggested design

| Component | Responsibility |
|---|---|
| `RequestConfig` | Stores request settings |
| `RequestResult` | Represents simulated response details |
| `RequestError` | Base request exception |
| `TransientRequestError` | Error that may be retried |
| `PermanentRequestError` | Error that must not be retried |
| `retry` | Configurable retry decorator |
| `request_logger` | Logs request execution |
| `RateLimiter` | Enforces a request policy |
| `APIClient` | Coordinates simulated requests |

## 4. Concepts you must demonstrate

- **`*args` and `**kwargs`:** wrappers and request functions must forward flexible arguments.
- **Decorator factory:** configure maximum retries, delay and retryable exception types.
- **Closures:** retain retry configuration between calls.
- **Decorator stacking:** combine retry, timing and logging decorators.
- **Higher-order functions:** build reusable wrappers around callables.
- **`functools.wraps`:** preserve the original callable metadata.
- **Keyword-only arguments:** make important configuration explicit where appropriate.
- **Argument unpacking:** pass configuration dictionaries using `**config`.
- **Custom exceptions:** distinguish retryable and non-retryable failures.

## 5. Functional requirements and rules

1. Retry count must be zero or greater.
2. Only configured exception types should trigger a retry.
3. Validation errors must not be retried.
4. The original function's successful return value must be returned unchanged.
5. If all attempts fail, the final failure must be raised or represented consistently.
6. Do not retry non-idempotent operations automatically unless the request policy explicitly allows it.
7. Credentials must be masked in logs.
8. A rate-limit rejection must not be counted as a successful request.
9. Retry delay should be configurable and tests should use a short or mocked delay.
10. Decorators must work with functions that accept different signatures.

## 6. Required Python features

- Decorators and decorator factories.
- `*args`, `**kwargs`, keyword-only parameters.
- Closures and `functools.wraps`.
- Custom exception hierarchy.
- `time.perf_counter()` and `time.sleep()` (or an injectable delay function).
- `logging`.
- `dataclasses` and `Enum`.
- Dictionaries for headers and configuration.
- Type hints and callable types.

## 7. Acceptance tests

- Call a simulated endpoint successfully.
- Verify a transient failure is retried and then succeeds.
- Verify a permanent error is not retried.
- Verify the configured maximum attempt count is respected.
- Verify the final exception is propagated after all retries fail.
- Pass extra keyword options and confirm they reach the wrapped function.
- Stack retry, timing and logging decorators.
- Verify function metadata remains intact.
- Confirm credentials are masked in logs.
- Trigger a rate limit and verify the request is rejected or delayed according to policy.

## 8. Suggested milestones

1. Implement a basic simulated API client.
2. Add flexible keyword configuration.
3. Create a retry decorator factory.
4. Add timing and request-logging decorators.
5. Implement rate limiting.
6. Test decorator composition and failure scenarios.

## 9. Expected deliverables

- Reusable request framework.
- Simulated API client with configurable outcomes.
- Retry, timing and logging decorators.
- Custom exception hierarchy.
- Tests for retry rules, argument forwarding and rate limits.
- README with examples and a clear statement that the client is simulated.

## 10. Challenge extensions

- Add exponential backoff.
- Add a circuit breaker decorator.
- Add request caching.
- Support asynchronous request functions.
- Add a per-endpoint rate-limit policy.
- Add structured JSON logs.

# Project 5 — Mini Task Scheduler & Event Processing Framework

<badge>Advanced</badge>

## 1. Problem statement

Build a small framework that registers tasks, schedules them for execution and processes generated events. Tasks may produce many results, so the system should support lazy processing rather than holding every result in memory.

The framework should support reusable task wrappers for logging, timing and validation, plus controlled resource management for task execution.

## 2. Core requirements

### Task registration
- Register a task with a unique name, callable, description and configuration.
- Allow tasks to receive positional and keyword arguments.
- Validate required task configuration before execution.
- Prevent accidental duplicate registration unless replacement is explicitly enabled.

### Execution
- Execute a task immediately or at a configured time.
- Record start time, finish time, status and result summary.
- Handle task failures without stopping unrelated tasks.
- Allow a task to produce a generator of results.
- Support cancellation of tasks that have not started.

### Event processing
- Tasks can yield events one at a time.
- Process events through a pipeline of handlers.
- Filter events by type or severity.
- Count processed, skipped and failed events.
- Allow the pipeline to process a large event stream lazily.

### Execution context
- Provide a context manager that marks task execution as started and finished.
- Ensure cleanup and final status updates happen even when a task fails.
- Make task configuration available to handlers without relying on global variables.

## 3. Suggested design

| Component | Responsibility |
|---|---|
| `Task` | Task metadata and execution callable |
| `TaskRegistry` | Registers and retrieves tasks |
| `TaskRunner` | Executes tasks and records outcomes |
| `TaskExecutionContext` | Manages task lifecycle |
| `Event` | Represents one event |
| `EventSource` | Produces events lazily |
| `EventPipeline` | Applies handlers and filters |
| `TaskResult` | Stores execution outcome |
| `TaskScheduler` | Selects tasks that are ready to run |

## 4. Concepts you must demonstrate

- **`*args` and `**kwargs`:** allow tasks to receive flexible arguments.
- **Decorators:** wrap task execution with timing, logging or validation.
- **Closures:** create configurable task policies or event filters.
- **Generators:** produce task results and event streams lazily.
- **Custom iterators:** implement a source that returns events in batches.
- **Context managers:** manage execution lifecycle and cleanup.
- **Comprehensions:** summarise task outcomes and event categories.
- **`itertools`:** batch, chain or group event streams where useful.
- **Higher-order functions:** accept handler functions and compose pipelines.
- **`dataclasses` and `Enum`:** represent tasks, events and lifecycle states.

## 5. Functional requirements and rules

1. Task names must be unique by default.
2. A task must be validated before it is scheduled.
3. A failed task must not be marked completed.
4. One task failure must not prevent independent tasks from running.
5. Event processing must preserve input order unless reordering is explicitly requested.
6. A generator must not be consumed twice accidentally; document whether a source is single-use.
7. A task context must be closed even if execution raises an exception.
8. Event handlers must have a consistent callable contract.
9. Task and event state transitions must be validated.
10. Do not use `eval` or execute arbitrary strings as Python code.

## 6. Required Python features

- Functions as values and callable type hints.
- `*args`, `**kwargs` and argument unpacking.
- Decorators and decorator factories.
- Generator functions and iterator protocol.
- Context managers and `contextlib`.
- `dataclasses` and `Enum`.
- `datetime` for scheduled execution metadata.
- `collections` and `itertools`.
- `logging` and custom exceptions.
- JSON persistence for task metadata and execution summaries.

## 7. Acceptance tests

- Register two tasks with different function signatures.
- Execute a task using both positional and keyword arguments.
- Reject duplicate task registration.
- Run a task that yields multiple events and verify lazy event processing.
- Filter events by type and severity.
- Cause one task to fail and verify an independent task still executes.
- Verify task execution context cleanup after success and failure.
- Attempt an invalid task-state transition and reject it.
- Process an event generator without converting the entire stream into a list.
- Export task execution summaries to JSON.

## 8. Suggested milestones

1. Build task model and registry.
2. Implement flexible task invocation.
3. Add decorators for task monitoring.
4. Implement task lifecycle context manager.
5. Build generator-based event source.
6. Add event handlers and lazy pipeline.
7. Add scheduling metadata, persistence and tests.

## 9. Expected deliverables

- Working task and event framework.
- Clear module structure for tasks, scheduling, events, handlers and storage.
- At least three example tasks with different signatures.
- Generator-based event source.
- Custom context manager for task lifecycle.
- Automated tests for state transitions, failures and lazy iteration.
- README with architecture and execution examples.

## 10. Challenge extensions

- Add priority queues.
- Add recurring task schedules.
- Add task dependencies.
- Add bounded event queues.
- Add an asynchronous task runner.
- Add a plugin registry for custom event handlers.
- Add a small command-line dashboard showing task status.

**Your task:** Keep the scheduler, task execution and event processing responsibilities separate. The framework should be extensible without turning into one giant class.

2. Complete Project 2 before building the event pipeline in Project 5.

## Completion standard

For every project, you should be able to:

- Explain why each advanced feature is used.
- Demonstrate at least five edge cases.
- Write tests for both success and failure paths.
- Refactor duplicated code into reusable functions or components.
- Explain the limitations of your design.
- Add one new feature without breaking existing behaviour.

**Final goal:** Do not just memorise syntax. Be able to recognise a real problem and choose the right Python feature to solve it cleanly.
