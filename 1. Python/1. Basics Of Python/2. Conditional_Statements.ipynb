# Conditional Statements in Python

Conditional statements are used to make decisions in a program based on conditions.

---

## Types of Conditional Statements

| Statement      | Description                                     |
| -------------- | ----------------------------------------------- |
| `if`           | Executes block if condition is true             |
| `if-else`      | Executes one block if true and another if false |
| `if-elif-else` | Checks multiple conditions                      |
| Nested `if`    | `if` statement inside another `if`              |

---

## 1. if Statement

### Syntax:
```python id="4p3dnz"
if condition:
    statement
```
### Example: 


````python
x = 10
if x > 5:
    print("Greater")
    
````

---

## 2. if-else Statement

### Example:

````python
x = 3
if x > 5:
    print("Greater")
else:
    print("Smaller")
````


---

## 3. if-elif-else Statement

### Example:

````python
marks = 75

if marks >= 90:
    print("A")
elif marks >= 60:
    print("B")
else:
    print("C")
````



---

## 4. Nested if Statement

### Example:


````python

x = 10

if x > 5:
    if x < 20:
        print("Valid")
````



---

## Advantages of Conditional Statements

* Helps in decision making
* Controls program flow
* Executes code based on conditions


# Conditional Statements in Python (Tabular Form)

| Statement Type | Description                                  | Syntax                              | Example            | Output                                           |
| -------------- | -------------------------------------------- | ----------------------------------- | ------------------ | ------------------------------------------------ |
| `if`           | Executes code if condition is true           | `if condition:`                     | `if x > 5:`        | Runs if condition is true                        |
| `if-else`      | Executes one block if true, another if false | `if condition: else:`               | `if x > 5: else:`  | One block executes                               |
| `if-elif-else` | Checks multiple conditions                   | `if...elif...else`                  | `if marks>=90:`    | Matching block executes                          |
| Nested `if`    | `if` inside another `if`                     | `if condition:` inside another `if` | `if x>5: if x<20:` | Inner block executes if both conditions are true |

# Comparison of Conditional Statements

| Statement      | Number of Conditions Checked  | Use                |
| -------------- | ----------------------------- | ------------------ |
| `if`           | One                           | Simple checking    |
| `if-else`      | One                           | Two-way decision   |
| `if-elif-else` | Multiple                      | Multiple decisions |
| Nested `if`    | Multiple dependent conditions | Complex checking   |
