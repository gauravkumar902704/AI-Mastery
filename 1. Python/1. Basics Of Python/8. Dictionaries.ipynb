# Dictionary in Python

## 1. Introduction to Dictionary

A **Dictionary** is a built-in Python data structure used to store data in **key-value pairs**.

* Mutable (can be modified)
* Ordered (Python 3.7+)
* Keys must be unique
* Values can be duplicated
* Fast data retrieval using keys

### Syntax

```python
dictionary_name = {
    key1: value1,
    key2: value2,
    key3: value3
}
```

### Example

````python
student = {
    'Name' : "Gaurav Kumar",
    'Roll_Number' : 2303410130012,
    'Age' : 21,
    'Course' : "B.Tech",
    'Branch' : "Information Technology"
}

print(type(student))
print(student)
````

# 8. Dictionary Methods

---

## keys()

Returns all keys.

````python
print(student)
print(student.keys())
````


---

## values()

Returns all values.

````python
print(student)
print(student.values())
````

---

## items()

Returns key-value pairs.

````python
print(student.items())
````

---

## update()

Updates dictionary.


````python
student3 = {
    'Name' : "Aman",
    'Age' : '19',
    'Branch' : 'C.S.E'
}

print(student3)

student3.update({
    'Name' : "Gaurav Kumar",
    'Branch' : 'I.T.',
    'Course' : "AKTU LUCKNOW",
    'Roll_Number' : 2303410130012
})

print(student3)
````

---

## copy()

Creates shallow copy.


````python
new_student = student3.copy()
print(new_student)
````

---

## setdefault()

Returns value if key exists, otherwise inserts key.

````python
student = {
    "Name" : "Amit"
}

student.setdefault("Age", 21)
print(student)
````

---

## fromkeys()

Creates dictionary from keys.

````python
keys = ["Name","Age","Branch"]

student = dict.fromkeys(keys, "NA")

print(student)
````

# 9. Looping Through Dictionary

## Loop Through Keys

````python
student = {
    "Name" : "Gaurav Kumar",
    "Age"  : 21,
    "Branch" : "I.T."
}

for key in student:
    print(key)
````

---

## Loop Through Values

````python
student = {
    "Name" : "Gaurav Kumar",
    "Age" : 21,
    "Branch" : "I.T."
}

for value in student.values():
    print(value)
````


---

## Loop Through Key-Value Pairs

````python
for key,value in student.items():
    print(key, ":", value)
````

# 11. Nested Dictionary

Dictionary inside another dictionary.

````python
students = {
    "student1" : {
        "Name" : "Gaurav Kumar",
        "Age"  : 21,
        "Branch" : "I.T."
    },
    "student2" : {
        "Name" : "Tarun Kumar",
        "Age" : 22,
        "Branch" : "C.S.E."
    }
}

print(students["student1"]["Name"])
print(students["student2"]["Name"])
print(students["student2"]["Age"])
````

````python
students = {
    "Student1" : {
        "Name" : "Gaurav Kumar",
        "Age"  : 21,
        "Branch" : "I.T."
    },
    "student2" : {
        "Name" : "Tarun",
        "Age"  : 22,
        "Branch" : "C.S.E."
    },
    "Student3" : {
        "Name" : "Amit",
        "Age"  : 19,
        "Branch" : "A.I./M.L."
    }
}

for key,value in students.items():
    print()
    print(key, ":", value)
    for key,value in value.items():
        print(f"{key} : {value}")
````

# 12. Dictionary Comprehension

### Syntax
```python
{key:value for item in iterable}
```

Example:


````python
# To Calculate Squares..

square = {x:x*x for x in range(1,11)}
print(square)
````

````python
# To Convert a list to a dictionary..

names = ["Alice","Bob","Charlie"]
name_lengths = {name : len(name) for name in names}

print(name_lengths)
print(type(name_lengths))
````

````python
# Dictionary Comperehension With Condition..

even_squares = {x:x*x for x in range(1,11) if x % 2 == 0}

print(even_squares)
````

# 13. Built-in Functions

### len()
### max()
### min()
### Sorted()


````python
students = {
    "student1" : {
        "Name" : "Gaurav Kumar",
        "Roll_Number" : 2303410130012,
        "Age" : 21,
        "Branch" : "I.T.",
        "Grade" : 7
    },
    "student2" : {
        "Name" : " Tarun Kumar",
        "Roll_Number" : 2303410100047,
        "Age" : 20,
        "Branch" : 'C.S.E.',
        "Grade" : 8
    },
    "student3" : {
        "Name" : "Saim",
        "Roll_Number" : 2303410130021,
        "Age" : 22,
        "Branch" : "I.T.",
        "Grade" : 7
    },
    "student4" : {
        "Name" : "Dev Karan",
        "Roll_Number" : 2303410130010,
        "Age" : 20,
        "Branch" : "I.T.",
        "Grade" : 7.5
    }
}

## printing Records.
print(students) 

print(f"Total students: {len(students)}")

highest = max(students, key=lambda x: students[x]["Grade"])
print(f"Highest Grade Student : {students[highest]["Name"]}")

lowest = min(students, key=lambda x:students[x]['Grade'])
print(f"Lowest Grade Student : {students[lowest]["Name"]}")

print("\nStudent Sorted by Grade: ")

for key,value in sorted(students.items(), key=lambda x:x[1]['Grade']):
    print(f"{value["Name"]} --> {value["Grade"]}")


````

````python
marks = {
    1: 50,
    2: 80,
    3: 60
}

print(max(marks))
print(min(marks))
print(sorted(marks))
````

# 17. Most Important Viva Questions

### Q1. What is a Dictionary?

A dictionary is a mutable collection of key-value pairs.

### Q2. Can Dictionary Have Duplicate Keys?

No.

```python
d = {
    "name": "Gaurav",
    "name": "Rahul"
}

print(d)
```

Output:

```python
{'name': 'Rahul'}
```

Last value overwrites the previous one.

---

### Q3. Can Dictionary Have Duplicate Values?

Yes.

```python
d = {
    "a": 10,
    "b": 10
}
```

---

### Q4. Difference Between get() and []?

| get()                       | []              |
| --------------------------- | --------------- |
| Returns None if key missing | Raises KeyError |
| Safer                       | Less safe       |

---

### Q5. What is Dictionary Comprehension?

A compact way to create dictionaries.

```python
squares = {x:x*x for x in range(5)}
```
