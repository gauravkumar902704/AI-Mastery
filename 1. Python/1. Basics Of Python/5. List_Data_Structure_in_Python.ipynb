# Python List and List Comprehension:

# 1. List in Python

A **List** is an ordered, mutable (changeable) collection of elements. It can store different types of data such as integers, strings, floats, and even other lists.

### Features of List

* Ordered (elements have index positions)
* Mutable (can be modified)
* Allows duplicate values
* Can store heterogeneous data types
* Written inside square brackets `[]`

### Syntax

````python
list_name = [element1, element2, element3]
````

### Example

````python
fruits = ["Apple", "Banana", "Mango"]
print(fruits)
````

**Output**

```python
['Apple', 'Banana', 'Mango']
```

# 10. List Comprehension

## Definition

**List Comprehension** is a concise and efficient way to create lists in a single line using loops and optional conditions.

### Syntax

```python
new_list = [expression for item in iterable]
```

### General Form

```python
new_list = [expression for item in iterable if condition]
```

---

# 11. Why Use List Comprehension?

### Traditional Method

````python
squares = []

for i in range(1, 6):
    squares.append(i*i)

print(squares)
````

Output:

```python
[1, 4, 9, 16, 25]
```

### Using List Comprehension

````python
squares = [i*i for i in range(1,6)]

print(squares)
````


Output:

```python
[1, 4, 9, 16, 25]
```

**Advantage:** Shorter, cleaner, faster.

# 13. List Comprehension with if-else

### Syntax
            [expression_if_true if condition else expression_if_false for item in iterable]

### Example:

````python
result = [f"Even: {x} "  if x%2==0 else f"Odd: {x}" for x in range(1,11)]

print(result)
````


Output:

```python
['Odd','Even','Odd','Even','Odd','Even','Odd','Even','Odd','Even']
```

# 14. Nested List Comprehension

### Example

````python
matrix = [[i*j for j in range(1,4)]
          for i in range(1,4)]

for row in matrix:
    print(row)
````



Output:

```python
[[1, 2, 3],
 [2, 4, 6],
 [3, 6, 9]]
```

### 21. Count how many times a particular element occurs in a list.

````python
list = [2,1,5,3,2,4,2,1,4,2,18,9,7,1,3,5,9,7,1,8]
#+++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++
# Method - I

counting = [list.count(num) for num in list]
print(counting)

#++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++
# Method - II 
perticular_element = 2
count = 0
for num in list:
    if num == perticular_element:
        count += 1

print(f"Frequency of {perticular_element} is: {count}")
````


### 22. Find the second largest number in a list.

````python
list = [10,65,14,75,31,92,3,60]

largest = list[0]
second_largest = 0

for num in list:
    if num > largest:
        second_largest = largest
        largest = num
    elif num > second_largest:
        second_largest = num        

print(f"Largest: {largest}")
print(f"Second Largest: {second_largest}")

````


### 23. Find the second smallest number in a list.

````python
#list = [64,32,8,525,45,24,70,95,45,52]
list = [5,10,15,20]

smallest = float('inf')
second_smallest = float('inf')

for num in list:
    if num < smallest:
        second_smallest = smallest
        smallest = num

    elif num > smallest and num < second_smallest:
        second_smallest = num

if second_smallest == float('inf'):
    print("No second smallest element exists.")

else:
    print(f"Second Smallest: {second_smallest}")
    print(f"Smallest: {smallest}")

````


### 24. Remove duplicate elements from a list.
