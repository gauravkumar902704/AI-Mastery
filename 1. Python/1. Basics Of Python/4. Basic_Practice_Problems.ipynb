7. Swap two variables without using third variable.

````python
def print_(a,b):
    print(f"First Number Before Swapping {a} and Second Number is {b}")
    print(' ')
def print_f(num1, num2):
    print(f"First Number After Swapping {num1} and Second Number is {num2}")
    print(' ')

def swap1(a,b):
    a = a + b
    b = a - b
    a = a - b
    print_f(a,b)

def swap2(a,b):
    a = a * b
    b = a / b
    a = a / b
    print_f(a,b)

def swap3(a,b):
    a = a ^ b
    b = a ^ b
    a = a ^ b
    print_f(a,b)

def swap4(a,b):
    a, b = b, a
    print_f(a,b)

num1 = int(input("Enter the first Number: ")) 
num2 = int(input("Enter the Second Number: "))

print_(num1,num2)

if num1 == 0 or num2 == 0:

    print("BY USING BITWISE XOR( MOST EFFICIENT AND SAFE): ")
    swap3(num1, num2)

    print("BY USING PYTHON SHOORTCUT(RECOMMENDED IN PYTHON): ")
    swap4(num1, num2)
else:

    print("BY USING ADDITION AND SUBTRACTION: ")
    swap1(num1, num2)

    print("BY USING MULTIPLICATION AND DIVISION: ")
    swap2(num1, num2)


````



# Method 1: Using Integer Division (Most Common)

### Formula

[
\text{Remainder} = a - \left(\frac{a}{b}\right) \times b
]

Here `/` means **integer division** (fractional part discarded).

### Example 1

Find remainder of **17 ÷ 5**

Step 1:

```
17 / 5 = 3
```

(integer division)

Step 2:

```
3 × 5 = 15
```

Step 3:

```
17 - 15 = 2
```

**Answer = 2**

### Python

```python
a = 17
b = 5

remainder = a - (a // b) * b

print(remainder)
```

Output

```
2
```

---

# Method 2: Using Repeated Subtraction

Subtract the divisor repeatedly until the remaining value becomes smaller than the divisor.

### Example

```
17 - 5 = 12
12 - 5 = 7
7 - 5 = 2
```

Now

```
2 < 5
```

So,

**Remainder = 2**

### Python

```python
a = 17
b = 5

while a >= b:
    a -= b

print(a)
```

Output

```
2
```

---

# Method 3: Using Recursion

```python
def remainder(a, b):
    if a < b:
        return a
    return remainder(a - b, b)

print(remainder(17, 5))
```

Output

```
2
```

# Method 4: Using Bitwise Operations (Only for Powers of 2)

If the divisor is a power of 2 (2, 4, 8, 16, ...), then:

```
remainder = number & (divisor - 1)
```

### Example

Find

```
29 % 8
```

Since

```
8 = 2³
```

```
29 & 7
```

Binary

```
29 = 11101
 7 = 00111
--------------
     00101
```

```
00101 = 5
```

Answer

```
29 % 8 = 5
```

Python

```python
print(29 & 7)
```

Output

```
5
```

# Method 5: Using a Loop Counter

```python
a = 17
b = 5
temp = a

while temp >= b:
    temp = temp - b

print("Remainder =", temp)
```

Output

```
Remainder = 2
```

# Time Complexity Comparison

| Method               | Time Complexity | Space Complexity | Best Use                           |
| -------------------- | --------------- | ---------------- | ---------------------------------- |
| `a - (a // b) * b`   | **O(1)**        | O(1)             | Fastest and most practical         |
| Repeated subtraction | O(a / b)        | O(1)             | Good for understanding the concept |
| Recursion            | O(a / b)        | O(a / b)         | Educational purposes               |
| Bitwise (`&`)        | **O(1)**        | O(1)             | Only when divisor is a power of 2  |

## Best Logical Approach

For most programming interviews and real-world code, the best approach without using `%` is:

```python
a = 17
b = 5

remainder = a - (a // b) * b

print(remainder)
```

This works because:

* `a // b` gives the integer quotient.
* `quotient × divisor` gives the largest multiple of the divisor that does not exceed `a`.
* Subtracting that multiple from `a` leaves the remainder.
