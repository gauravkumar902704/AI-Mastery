6. Swap two variables using third variable.

```python
num1 = int(input("Enter first number:"))
num2 = int(input("Enter second number:"))
temp = 0
print(f"Before swapping: num1 = {num1}, num2 = {num2}")

temp = num1
num1 = num2
num2 = temp
print(f"After swapping: num1 = {num1}, num2 = {num2}")
```

7. Swap two variables without using third variable.

```python
def print_(a,b):
    print(f"First Number Before Swapping {a} and Second Number is {b}")
    print(' ')
def print_f(num1, num2):
    print(f"First Number After Swapping {num1} and Second Number is {num2}")
    print(' ')

def swap1(a,b):
    a = a + b
    b = a - b
    a = a - b
    print_f(a,b)

def swap2(a,b):
    a = a * b
    b = a / b
    a = a / b
    print_f(a,b)

def swap3(a,b):
    a = a ^ b
    b = a ^ b
    a = a ^ b
    print_f(a,b)

def swap4(a,b):
    a, b = b, a
    print_f(a,b)

num1 = int(input("Enter the first Number: ")) 
num2 = int(input("Enter the Second Number: "))

print_(num1,num2)

if num1 == 0 or num2 == 0:

    print("BY USING BITWISE XOR( MOST EFFICIENT AND SAFE): ")
    swap3(num1, num2)

    print("BY USING PYTHON SHOORTCUT(RECOMMENDED IN PYTHON): ")
    swap4(num1, num2)
else:

    print("BY USING ADDITION AND SUBTRACTION: ")
    swap1(num1, num2)

    print("BY USING MULTIPLICATION AND DIVISION: ")
    swap2(num1, num2)


```

8. Find square and cube of a number.

```python
def Square(a):
    s = a ** 2
    print(f"Square of {a} is: {s}")

def Cube(a):
    C = a ** 3
    print(f"Cube of {a} is: {C}")

num1 = int(input("Enter a Number: "))
C = int(input("If you want to calculate Square than press '1' else press '0' :  "))

if C == 1:
    Square(num1)
else:
    Cube(num1)
```
