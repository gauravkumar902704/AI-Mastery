# Loops in Python

Loops are used to execute a block of code repeatedly until a condition becomes false.

---

# Types of Loops in Python

| Loop Type    | Description                          | Syntax                      | Example                                 |
| ------------ | ------------------------------------ | --------------------------- | --------------------------------------- |
| `for` loop   | Repeats code for a sequence          | `for variable in sequence:` | `for i in range(5):`                    |
| `while` loop | Repeats code while condition is true | `while condition:`          | `while x < 5:`                          |
| Nested loop  | Loop inside another loop             | Loop inside loop            | `for i in range(2): for j in range(2):` |

---

---
# for Loop

| Feature           | Description                |
| ----------------- | -------------------------- |
| Used for          | Iterating over sequence    |
| Sequence Examples | List, tuple, string, range |
| Executes          | Fixed number of times      |

### Example:

````python
for i in range(5):
    print(i) # This will print the numbers 0 to 4, each on a new line.
````

## Basic Syntax

```python
range(start, stop, step)
```

Parameters:

* **start** (optional): First number in the sequence. Default is `0`.
* **stop** (required): Sequence stops *before* this number.
* **step** (optional): Difference between consecutive numbers. Default is `1`.

# Example of break

````python
for i in range(5):
    if i == 3:
        break
    print(i)
````

---

# Example of continue

````python
for i in range(5):
    if i == 3:
        continue
    print(i) # This will print the numbers 0, 1, 2, and 4, each on a new line.
````

## 'Pass' Statement:
The `pass` statement in Python is a **placeholder** that does nothing when executed.

It's useful when Python requires a statement syntactically, but you haven't written the actual code yet.

## Basic Example

```python
if True:
    pass

print("Program continues")
```

Output:

```text
Program continues
```

The `pass` statement is ignored, and execution continues normally.

---

## Why Use `pass`?

### 1. Empty Function

Without `pass`, Python would raise an error because a function body cannot be empty.

```python
def my_function():
    pass
```

You can implement the function later.

---

### 2. Empty Loop

```python
for i in range(5):
    pass

print("Loop finished")
```

Output:

```text
Loop finished
```

The loop runs 5 times but performs no action.

---

### 3. Empty Class

```python
class Student:
    pass
```

Creates a class with no attributes or methods yet.

---

### 4. Placeholder in Conditional Logic

```python
age = 18

if age >= 18:
    pass      # Code will be added later
else:
    print("Minor")
```

Currently nothing happens when `age >= 18`.

---

## `pass` vs `continue`

These are often confused:

### `pass`

Does nothing.

```python
for i in range(3):
    if i == 1:
        pass
    print(i)
```

Output:

```text
0
1
2
```

---

### `continue`

Skips the rest of the current iteration.

```python
for i in range(3):
    if i == 1:
        continue
    print(i)
```

Output:

```text
0
2
```

---

## `pass` vs `break`

```python
for i in range(5):
    if i == 2:
        break
    print(i)
```

Output:

```text
0
1
```

`break` exits the loop completely, while `pass` simply does nothing.

## Summary

| Statement  | Purpose                                       |
| ---------- | --------------------------------------------- |
| `pass`     | Do nothing (placeholder)                      |
| `continue` | Skip current loop iteration                   |
| `break`    | Exit the loop                                 |
| `return`   | Exit a function and optionally return a value |

A common real-world use is when you're designing a program structure first:

```python
def login():
    pass

def logout():
    pass

def register():
    pass
```

This lets you run the program without implementing the functions immediately.

## Nested Loop:- 
A **nested loop** is a loop inside another loop.

The **inner loop** runs completely for each iteration of the **outer loop**.

## Basic Syntax

````python
for i in range(3):      # Outer loop
    for j in range(2):  # Inner loop
        print(i, j)
````


Output:

```text
0 0
0 1
1 0
1 1
2 0
2 1
```

### How it works

* When `i = 0`, `j` runs from `0` to `1`
* When `i = 1`, `j` runs from `0` to `1`
* When `i = 2`, `j` runs from `0` to `1`

---

## Example: Multiplication Table

````python
for i in range(1, 4):
    for j in range(1, 11):
        print(i * j, end="\t")
    print()
````

Output:

```text
1   2   3   4   5
2   4   6   8   10
3   6   9   12  15
```

The `print()` after the inner loop moves to the next line.

---
## Example: Print a Square Pattern

````python
for i in range(4):
    for j in range(4):
        print("*", end=" ")
    print()
````

Output:

```text
* * * *
* * * *
* * * *
* * * *
```

---

## Example: Right Triangle Pattern

````python
for i in range(1, 6):
    for j in range(i):
        print("*", end=" ")
    print()
````

Output:

```text
*
* *
* * *
* * * *
* * * * *
```

## Using Nested Loops with Lists

````python
colors = ["Red", "Blue"]
fruits = ["Apple", "Banana"]

for color in colors:
    for fruit in fruits:
        print(color, fruit)
````


Output:

```text
Red Apple
Red Banana
Blue Apple
Blue Banana
```

## Key Point

If:


````python
for i in range(3):
    for j in range(4):
        pass
````



* Outer loop runs **3** times.
* Inner loop runs **4** times for each outer iteration.
* Total executions of the inner block = **3 × 4 = 12**.

Nested loops are commonly used for:

* Pattern printing
* Working with tables/matrices
* Comparing items from two lists
* Generating combinations of values
