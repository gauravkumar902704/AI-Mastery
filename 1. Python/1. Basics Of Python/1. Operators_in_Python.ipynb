## Operators in Python

Operators are symbols used to perform operations on variables and values.

---

## Types of Operators in Python

### 1. Arithmetic Operators

Used for mathematical operations.

| Operator | Meaning        |
| -------- | -------------- |
| `+`      | Addition       |
| `-`      | Subtraction    |
| `*`      | Multiplication |
| `/`      | Division       |
| `%`      | Modulus        |
| `//`     | Floor Division |
| `**`     | Exponent       |

### Example:

```python
x = 10
y = 3
print(f"Addition = {x + y}")
print(f"Subtraction = {x - y}")
print(f"Multiplication = {x * y}")
print(f"Division = {x / y}")
print(f"Modulus = {x % y}")
print(f"Floor Division = {x // y}")
print(f"Exponent/Power = {x ** y}")
```


---

### 2. Comparison Operators

Used to compare values.

| Operator | Meaning                  |
| -------- | ------------------------ |
| `==`     | Equal to                 |
| `!=`     | Not equal to             |
| `>`      | Greater than             |
| `<`      | Less than                |
| `>=`     | Greater than or equal to |
| `<=`     | Less than or equal to    |

### Example:

```python
x,y = 7,8
firstname = "Gaurav"
lastname = "Kumar"
print(f"{x} and {y} is equal = {x == y}")
print(f"{x} and {y} is not equal = {x != y}")
print(f"{x} is greater than {y} = {x > y} ")
print(f"{x} is less than {y} = {x < y}")
print(f"{firstname} is greater than or equal to {lastname} = {x >= y}")
print(f"{firstname} is less than or equal to {lastname} = {x <= y}")
```

---

### 3. Logical Operators

Used to combine conditions.

| Operator | Meaning                          |
| -------- | -------------------------------- |
| `and`    | True if both conditions are true |
| `or`     | True if one condition is true    |
| `not`    | Reverses the condition           |

### Example:

```python
a = 12 
b = 0
c = " "
d = "Gaurav"

print(a and b)
print(a or b)
print(a and c)
print(c and a)
print(a or c)
print(a and d)
print(a or d)
print(c and d)
print(c or d)
```


---

### 4. Assignment Operators

Used to assign values.

| Operator | Example  |
| -------- | -------- |
| `=`      | `x = 5`  |
| `+=`     | `x += 2` |
| `-=`     | `x -= 2` |

```python
x = 10
x += 5
print(x)
```


---

### 5. Bitwise Operators

Perform operations on binary values.



| Operator | Name        | Description                     |                                    |
| -------- | ----------- | ------------------------------- | ---------------------------------- |
| `&`      | Bitwise AND | Returns 1 if both bits are 1    |                                    |
| `        | `           | Bitwise OR                      | Returns 1 if at least one bit is 1 |
| `^`      | Bitwise XOR | Returns 1 if bits are different |                                    |
| `~`      | Bitwise NOT | Reverses the bits               |                                    |
| `<<`     | Left Shift  | Shifts bits to left             |                                    |
| `>>`     | Right Shift | Shifts bits to right            |                                    |


```python
x = 10
y = 5

print(x & y)
print(x | y)
print(x ^ y)
print(~x)
```


---

### 6. Membership Operators

Check membership in a sequence.

| Operator | Meaning     |
| -------- | ----------- |
| `in`     | Present     |
| `not in` | Not Present |

### Example:

```python
print("a" in "apple")
```


---

### 7. Identity Operators

Compare memory locations.

| Operator | Meaning          |
| -------- | ---------------- |
| `is`     | Same object      |
| `is not` | Different object |


```python
x = [1, 2]
y = x
print(x is y)
```

# Bitwise Operators Quick Recap

| Operator  | Description                     | Condition for Result `1`      | Example                 | Result   |     |
| --------- | ------------------------------- | ----------------------------- | ----------------------- | -------- | --- |
| `&` (AND) | Returns 1 if both bits are 1    | Both bits must be `1`         | `5 & 3`                 | `1`      |     |
| `         | ` (OR)                          | Returns 1 if any one bit is 1 | At least one bit is `1` | `5 \| 3` | `7` |
| `^` (XOR) | Returns 1 if bits are different | Bits must be different        | `5 ^ 3`                 | `6`      |     |
| `~` (NOT) | Reverses all bits               | Converts `1→0` and `0→1`      | `~5`                    | `-6`     |     |
| `<<`      | Left shift bits                 | Shifts bits left              | `5 << 1`                | `10`     |     |
| `>>`      | Right shift bits                | Shifts bits right             | `5 >> 1`                | `2`      |     |
