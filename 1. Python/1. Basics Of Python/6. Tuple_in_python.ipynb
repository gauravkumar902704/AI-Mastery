# Tuple in Python

## 1. Introduction to Tuple

A **Tuple** is a built-in Python data structure used to store multiple items in a single variable.

* Ordered collection of elements.
* Immutable (cannot be changed after creation).
* Allows duplicate values.
* Can store different data types.
* Faster than lists for fixed data.

### Syntax

```python
tuple_name = (item1, item2, item3, ...)
```

### Example


````python
student = ("Gaurav", 20, "IT")
print(student)
print(type(student))
````


**Output:**

```python
('Gaurav', 20, 'IT')
```

---

# 2. Characteristics of Tuple

| Feature           | Description                           |
| ----------------- | ------------------------------------- |
| Ordered           | Elements maintain insertion order     |
| Immutable         | Cannot modify elements                |
| Allows Duplicates | Same values can appear multiple times |
| Indexed           | Access elements using index           |
| Heterogeneous     | Different data types allowed          |
| Faster            | Faster than lists for fixed data      |

# 3. Creating Tuples

## Empty Tuple

````python
tuple1 = tuple()

print(type(tuple1))
print(tuple1)
````

Output:

```python
()
```

---

## Tuple with One Element

A comma is mandatory.

````python
single = (10,)
print(type(single))
````


Output:

```python
<class 'tuple'>
```

Without comma:


````python
single = (10)

print(type(single))
````

Output:

```python
<class 'int'>
```

---



## Multiple Elements


````python
numbers = (1, 2, 3, 4, 5)
print(numbers)
````


Output:

```python
(1, 2, 3, 4, 5)
```

---

# 4. Tuple Packing

Packing means storing multiple values into a tuple.

````python
student = ("Gaurav", 20, "IT")
print(student)
````


Output:

```python
('Gaurav', 20, 'IT')
```

Python automatically packs values into a tuple.


````python
student = "Gaurav", 20, "IT"
print(student)
````


Output:

```python
('Gaurav', 20, 'IT')
```

---

# 5. Tuple Unpacking

Extracting tuple values into variables.


````python
student = ("Gaurav", 20, "IT")

name, age, branch = student

print(name)
print(age)
print(branch)
````

Output:

```python
Gaurav
20
IT
```

# 6. Accessing Tuple Elements

## Positive Indexing


````python
fruits = ("Apple", "Banana", "Mango")

print(fruits[0])
print(fruits[1])
print(fruits[2])
````

Output:

```python
Apple
Banana
Mango
```



---

## Negative Indexing

````python
fruits = ("Apple", "Banana", "Mango")

print(fruits[-1])
print(fruits[-2])
````

Output:

```python
Mango
Banana
```


---

# 7. Tuple Slicing

Syntax:

```python
tuple[start:end:step]
```

Example:

````python
numbers = (10,20,30,40,50,60)

print(numbers[1:4])
print(numbers[:3])
print(numbers[3:])
print(numbers[::2])
````

Output:

```python
(20, 30, 40)
(10, 20, 30)
(40, 50, 60)
(10, 30, 50)
```

---

Output:

# 9. Tuple Operations

````python
t1 = (1,2,3)
t2 = (4,5,6)

result = t1 + t2

print(result)
````

# 13. Tuple Methods

Tuple has only two methods.

| Method  | Description        |
| ------- | ------------------ |
| count() | Counts occurrences |
| index() | Returns index      |

---

## count()


````python
numbers = (1,2,2,3,2)

print(numbers.count(2))
````

Output:

```python
3
```

---

## index()


````python
numbers = (10,20,30,40)

print(numbers.index(30))
````

Output:

```python
2
```

# 14. Nested Tuple

Tuple inside another tuple.


````python
student = (
    ("Gaurav", 20),
    ("Rahul", 21),
    ("Amit", 22)
)

print(student[0])
print(student[1][0])
````



Output:

```python
('Gaurav', 20)
Rahul
```

# 15. Tuple vs List

| Feature      | List | Tuple  |
| ------------ | ---- | ------ |
| Syntax       | []   | ()     |
| Mutable      | Yes  | No     |
| Faster       | No   | Yes    |
| Methods      | Many | Only 2 |
| Memory Usage | More | Less   |
| Security     | Less | More   |

---

# 16. Advantages of Tuple

1. Faster execution.
2. Less memory usage.
3. Data safety due to immutability.
4. Useful for fixed data.
5. Can be used as dictionary keys (if elements are immutable).

## List vs Tuple (Very Important)

| Feature           | List         | Tuple      |
| ----------------- | ------------ | ---------- |
| Syntax            | `[]`         | `()`       |
| Mutable           | Yes          | No         |
| Ordered           | Yes          | Yes        |
| Duplicate Allowed | Yes          | Yes        |
| Memory Usage      | More         | Less       |
| Speed             | Slower       | Faster     |
| Methods           | Many         | Few        |
| Use Case          | Dynamic Data | Fixed Data |
