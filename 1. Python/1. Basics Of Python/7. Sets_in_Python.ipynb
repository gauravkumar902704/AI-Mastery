# Sets in Python

## 1. Introduction to Sets

A **Set** is a built-in Python data structure used to store multiple unique items in a single variable.

### Characteristics of Sets

* Unordered collection of elements.
* Mutable (elements can be added or removed).
* Does not allow duplicate values.
* Elements must be immutable (e.g., numbers, strings, tuples).
* Faster lookup compared to lists.

### Syntax

```python
set_name = {item1, item2, item3}
```

### Example

````python
fruits = {"Apple", "Banana", "Mango"}
print(fruits)
````

# 12. Symmetric Difference
In Python, the **symmetric difference** of two sets means:

> Elements that are in **either set**, but **not in both**.

Mathematically:

[
A \triangle B = (A - B) U (B - A)
]

### Example

```python
A = {1, 2, 3, 4}
B = {3, 4, 5, 6}

result = A.symmetric_difference(B)

print(result)
```

Output:

```python
{1, 2, 5, 6}
```

Explanation:

* Common elements (`3`, `4`) are removed.
* Elements unique to either set remain (`1`, `2`, `5`, `6`).

---

````python
A = {5,9,3,4,7,4,2,1}
B = {3,7,6,1,4}
result = A.symmetric_difference(B)
print(result)
````


### Using the `^` Operator

Python provides a shortcut operator:

````python
C = {3,4,6,9,7}

print(f"  A:  {A}")
print(f"  B:  {B}")
print(f"  C:  {C}")
print(f"A ^ B ^ C: {A ^ B ^ C}")
````

Reasoning:

* `A ^ B` → `{1, 2, 4}`
* `{1, 2, 4} ^ C` → `{1, 2, 5}`

---

### Updating a Set In-Place

````python
print(f"  A:  {A}")
print(f"  B:  {B}")
print(f"  C:  {C}")
result = A.symmetric_difference_update(B)
print(f"Result: {result}")
print(f"A : {A}")
````


`A` itself is modified.

---

### Quick Visualization

```text
A = {1, 2, 3, 4}
B = {3, 4, 5, 6}

Only in A:  {1, 2}
Only in B:  {5, 6}
In both:    {3, 4}

Symmetric Difference = {1, 2, 5, 6}
```

Think of it as "**keep the non-common elements**" or "**XOR for sets**."

### 15.2. Remove duplicate words from a sentence in ordered.

````python
sentence = " # # # 15 . Remove duplicate words from a sentence and print them in ordered preservence format and text should be plain text , unlike sets printing ( 'an' , 'are' , 'and' ) ."
words = sentence.split()

result = []
seen = set()

for word in words:
    if word not in seen:
        result.append(word)
        seen.add(word)

output = " ".join(result)
print(output)
````
