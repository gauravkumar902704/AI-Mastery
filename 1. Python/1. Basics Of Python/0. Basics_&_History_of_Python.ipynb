## What is Python?

Python is a high-level, interpreted, and general-purpose programming language. It is easy to learn because of its simple syntax and is widely used in web development, artificial intelligence, data science, automation, and software development.

### Features of Python

* Easy to read and write
* Interpreted language
* Object-oriented
* Platform independent
* Open source
* Large library support

---

## Origin and History of Python

Python was created by Guido van Rossum in the late 1980s at CWI (Centrum Wiskunde & Informatica) in the Netherlands.

* Python development started in **1989**.
* The first official version, **Python 1.0**, was released in **1991**.
* The name “Python” was inspired by the comedy show **Monty Python's Flying Circus**.
* Python 2.0 was released in 2000 with new features.
* Python 3.0 was released in 2008 with major improvements and is the modern version used today.

Python has become one of the most popular programming languages because of its simplicity and powerful capabilities.


## Syntax in Python

Syntax refers to the set of rules used to write Python programs correctly. It defines the proper structure of statements and code.

### Example:

````python
print("Hello, World!")
print("Welcome to Python programming.")
````


If syntax rules are not followed, Python gives a **Syntax Error**.

## Semantics in Python

Semantics refers to the meaning or behavior of a statement in a program. It explains what the code actually does.

### Example:

````python
x = 10
y = 20

result = x + y
print("The sum of", x, "and", y, "is:", result)
````


### Output:

```python
30
```

Here, the semantics of `+` operator is addition of two numbers.

---




## Difference Between Syntax and Semantics

| Syntax                                  | Semantics                                 |
| --------------------------------------- | ----------------------------------------- |
| Rules of writing code                   | Meaning of code                           |
| Checks structure                        | Checks logic/behavior                     |
| Syntax errors occur if rules are broken | Semantic errors occur if meaning is wrong |


## Comments in Python

Comments are non-executable lines in a program used to explain the code and improve readability. Python ignores comments during execution.

---

## Types of Comments in Python

### 1. Single-Line Comment

A single-line comment starts with `#`.

### Example:

````python
# This is a comment in Python. Comments are ignored by the interpreter and are used to explain code.
# You can also use comments to temporarily disable code.
# The following line of code is commented out and will not be executed.
# print("This line of code is commented out and will not run.")
````


---

### 2. Multi-Line Comment

Multi-line comments are written using triple quotes (`''' '''` or `""" """`).

### Example:


````python
'''
this is a multi-line comment in Python.
It can span multiple lines and is often used for longer explanations or documentation. 
Multi-line comments are enclosed in triple quotes (''' or """). 
This allows you to write comments that are more detailed and easier to read.
You can use multi-line comments to provide context, explain complex logic, or document your code for future reference.
Multi-line comments are especially useful when you want to include examples, instructions, or any information that requires more than one line.
Remember that comments are an essential part of writing clean and maintainable code, as they help others (or yourself in the future) understand the purpose and functionality of your code.

"""

````

## Data Types in Python

Data types define the type of value a variable can store. Python automatically identifies the data type of a variable.

---

## Types of Data Types in Python

### 1. Numeric Data Type

Used to store numbers.

* `int` → Integer numbers
* `float` → Decimal numbers
* `complex` → Complex numbers

### Example:

````python
x = 10
y = 3.5
z = 2 + 3j
print(type(x))  # Output: <class 'int'>
print(type(y))  # Output: <class 'float'>
print(type(z))  # Output: <class 'complex'>
````

---

### 2. String Data Type

Used to store text data.

### Example:

````python
Name = "Gaurav"
print(Name)
print(type(Name))
````


---

### 3. List Data Type

An ordered and mutable collection.

### Example:

````python
colors = ["red", "blue", "green"]

print(colors[0])  # Output: red
print(colors[1])  # Output: blue
print(colors[2])  # Output: green
````


---

### 4. Tuple Data Type

An ordered and immutable collection.

### Example:

````python
numbers = (1, 2, 3)

print(numbers[0])  # Output: 1
print(numbers[1])  # Output: 2
print(numbers[2])  # Output: 3
````

---

### 5. Set Data Type

An unordered collection of unique elements.

### Example:

````python
s = {1, 2, 3, 4, 5}


print(type(s))  # Output: <class 'set'>
print(s)  # Output: {1, 2, 3, 4, 5}
````


---

### 6. Dictionary Data Type

Stores data in key-value pairs.

### Example:

````python
students = {
    "Alice": 85, 
    "Bob": 90, 
    "Charlie": 78
}

                        # OR 

student = {"name": "Alice", "age": 20, "grade": "A"}
print(student["name"])  # Output: Alice
print(student["age"])   # Output: 20
print(student["grade"]) # Output: A
````

---

### 7. Boolean Data Type

Stores `True` or `False`.

### Example:

````python
x = True
y = False

print(type(x))  # Output: <class 'bool'>
print(type(y))  # Output: <class 'bool'>
print(x)  # Output: True
print(y)  # Output: False
````


### Output:

```python id="if2vst"

```


````python
text = "Hello, World!"

print(text[0])  # Output: H
print(text[7])  # Output: W
print(text[-1])  # Output: !

print(' ')

print(text[0:5])  # Output: Hello
print(text[7:12])  # Output: World

print(' ')

for i in range(-1,-14,-1):
    print(text[i])  # Output: !, d, l, r, o, W,  , o, l, l, e, H
````

## Type Conversion in Python

Type conversion means converting one data type into another data type.

---

## Types of Type Conversion

### 1. Implicit Type Conversion

Python automatically converts one data type into another.

### Example:

````python
x = 5
y = 2.5
print(x + y) # Output: 7.5
````

---

### 2. Explicit Type Conversion

The programmer manually converts the data type using functions like `int()`, `float()`, `str()`.

### Example:


````python
x = "10"
y = int(x)
print(y) # Output: 10
````



---

## Common Type Conversion Functions

| Function  | Description         |
| --------- | ------------------- |
| `int()`   | Converts to integer |
| `float()` | Converts to float   |
| `str()`   | Converts to string  |
| `list()`  | Converts to list    |
| `tuple()` | Converts to tuple   |
