# Exception Handling in Python

**Exception handling** is a mechanism used to detect and handle runtime errors so that the program does not terminate unexpectedly.

## What is an Exception?

An **exception** is an error that occurs during the execution of a program.

### Example (Without Exception Handling)

````python
a = 10
b = 0

print(a / b)
````

**Output**

```text
ZeroDivisionError: division by zero
```

The program stops immediately because the exception is not handled.

---


# Why Use Exception Handling?

* Prevents the program from crashing.
* Handles errors gracefully.
* Improves program reliability.
* Allows the program to continue execution after an error.

---

# `try` and `except`

The basic syntax is:

```python
try:
    # Code that may cause an exception
except ExceptionType:
    # Code to handle the exception
```
### Example

````python
try:
    num = int(input("Enter a number: "))
    print(100 / num)
except ZeroDivisionError:
    print("You cannot divide by zero.")
````

---

# Handling Multiple Exceptions

You can handle different exceptions separately.


````python
try:
    num = int(input("Enter a number: "))
    result = 100 / num
    print(result)

except ZeroDivisionError:
    print("Division by zero is not allowed.")

except ValueError:
    print("Please enter a valid integer.")
````

---

# Handling Unknown Exceptions

````python
try:
    num = int(input("Enter a Number: "))
    result = 100/num
except Exception as e:
    print(e)
else: 
    print(f"Result: {result}")
````


---

# Using a Generic `except`

A generic `except` catches any exception.


````python
try:
    x = int(input("Enter a number: "))
    print(10 / x)

except Exception as e:
    print("An error occurred:", e)
````


Here, `e` contains the error message.

---


# `else` Block

The `else` block executes **only if no exception occurs**.


````python
try:
    num = int(input("Enter a number: "))
    print(100 / num)

except ZeroDivisionError:
    print("Cannot divide by zero.")

else:
    print("Operation completed successfully.")
````

---

# `finally` Block

The `finally` block always executes, whether an exception occurs or not. It is commonly used to release resources such as closing files or database connections.


````python
try:
    f = open("sample.txt", "r")
    print(f.read())

except FileNotFoundError:
    print("File not found.")

finally:
    print("Program finished.")
````


---

# Complete Example


````python
try:
    num = int(input("Enter a number: "))
    result = 100 / num

except ValueError:
    print("Invalid input.")

except ZeroDivisionError:
    print("Cannot divide by zero.")

else:
    print("Result =", result)

finally:
    print("Execution completed.")
````

---

# Raising Exceptions (`raise`)

You can create your own exceptions using the `raise` keyword.


````python
age = int(input("Enter your age: "))

if age < 18:
    raise ValueError("Age must be at least 18.")

print("You are eligible.")
````

---

# User-Defined Exception

You can define your own exception class by inheriting from `Exception`.


````python
class NegativeNumberError(Exception):
    pass

num = int(input("Enter a number: "))

if num < 0:
    raise NegativeNumberError("Negative numbers are not allowed.")

print("You entered:", num)
````

| ------------------- | ------------------------------------------------- |
| `ZeroDivisionError` | Division by zero                                  |
| `ValueError`        | Invalid value (e.g., converting `"abc"` to `int`) |
| `TypeError`         | Operation on incompatible data types              |
| `IndexError`        | List index out of range                           |
| `KeyError`          | Dictionary key not found                          |
| `NameError`         | Variable not defined                              |
| `FileNotFoundError` | File does not exist                               |
| `AttributeError`    | Object has no requested attribute                 |
| `ImportError`       | Module cannot be imported                         |

# Flow of Exception Handling

```text
           Start
              │
              ▼
          try block
              │
      ┌───────┴────────┐
      │                │
 No Exception     Exception Occurs
      │                │
      ▼                ▼
 else block      matching except block
      │                │
      └───────┬────────┘
              ▼
       finally block
              │
              ▼
             End
```

---

# Summary

| Keyword   | Purpose                                            |
| --------- | -------------------------------------------------- |
| `try`     | Encloses code that may raise an exception          |
| `except`  | Handles the exception                              |
| `else`    | Executes if no exception occurs                    |
| `finally` | Executes regardless of whether an exception occurs |
| `raise`   | Explicitly raises an exception                     |

## Best Practices

* Catch **specific exceptions** (e.g., `ValueError`, `FileNotFoundError`) instead of using a generic `except` whenever possible.
* Use `finally` to clean up resources such as files, network connections, or database connections.
* Use `raise` when you want to signal an error condition from your own code.
* Keep `try` blocks as small as practical so they only cover code that might raise the expected exception.
