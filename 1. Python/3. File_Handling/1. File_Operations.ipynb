# File Handling in Python

File handling is used to **create, read, write, append, and manage files** in Python.

## Opening a File

Syntax: 
```python
file = open("filename.txt", "mode")
```

### Common Modes

| Mode   | Description                                                     |
| ------ | --------------------------------------------------------------- |
| `"r"`  | Read (default)                                                  |
| `"w"`  | Write (creates file if not exists, overwrites existing content) |
| `"a"`  | Append (adds data to end of file)                               |
| `"x"`  | Create a new file                                               |
| `"r+"` | Read and write                                                  |
| `"b"`  | Binary mode (e.g., images, PDFs)                                |

---

## Reading a File

### Read Entire File

````python
file = open("sample.txt", "r")
content = file.read()
print(content)
file.close()
````

### Read One Line

````python
file = open("sample.txt", "r")
print(file.readline())
file.close()
````

### Read All Lines

````python
with open('sample.txt', mode='r') as file:
    All_lines = file.readlines()
    print(All_lines)
    
````


---

## Writing to a File


````python
with open('sample1.txt', mode='w',newline='') as file:
    writer = file.write('Hello, Gaurav\n')
    writer = file.write("What's up..\n")
    writer = file.write("I hope you're doing well...")

    
````

````python
with open('sample1.txt', mode='r') as file:

# By using For Loop..

    for line in file:
        print(line.strip())   # strip function removes the newline charector

    
````

````python

# with read() function..
with open('sample1.txt',mode='r') as file:
    content = file.read()
    print(content)
    
````

````python
# with readlines() function
with open('sample1.txt',mode='r') as file:
    content = file.readlines()
    print(content)
    
````

````python
# with readline() funtion.
file = open('sample.txt','r')
content = file.readline()
print(content)
file.close()
````


---

## Appending Data


````python
file = open('sample.txt','+a')
file.write("Welcome to the File Handling..\n")
file.close()

with open('sample.txt', mode='r') as file:
    for line in file:
        print(line.strip())
    
    
````

---

## Using `with` Statement (Recommended)

The file is automatically closed after use.


````python
with open("sample.txt", "r") as file:
    content = file.read()
    print(content)
````

---

## Creating a New File


````python
with open("newfile.txt", "x") as file:
    file.write("New file created")
````


---

## Checking File Existence


````python
import os

if os.path.exists("sample.txt"):
    print("File exists")
else:
    print("File not found")
````

### Writing multiple lines at onces


````python
lines = ['\nFirst line\n','Second line\n','Third line\n','Fourth line\n']
try:
    with open('newfile.txt', mode='a') as file:
        file.writelines(lines)

except FileNotFoundError:
    print("Such File does not exist")


with open('newfile.txt', mode='r') as file:
    content = file.read()
    print(content)
````


---

## Complete Example


````python
# Writing data
with open("student.txt", "w") as file:
    file.write("Name: John\n")
    file.write("Age: 20\n")

# Reading data
with open("student.txt", "r") as file:
    print(file.read())
````

---

## Deleting a File


````python
import os

os.remove("student.txt")
````


## Best Practices

* Use `with open()` instead of `open()` and `close()`.
* Handle exceptions using `try-except`.
* Use appropriate file modes.
* Close files properly to avoid memory leaks.

Example with exception handling:


````python
try:
    with open("sample.txt", "r") as file:
        print(file.read())
except FileNotFoundError:
    print("File does not exist")
````

This covers the basics of file handling in Python: **open, read, write, append, create, and delete files**.


### working with Binary Files

### Writing In Binary File

````python


data = b'\x00\x01\x02\x03\x05'

with open('sample.bin',mode='wb') as file:
    file.write(data)

````

### Reading in Binary File

````python
with open('sample.bin',mode='rb') as file:
    content = file.read()
    print(content)
````
