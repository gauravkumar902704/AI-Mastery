Git ka normal workflow:

```bash
git add .
git commit -m "Meaningful message"
git push origin main
```

### Commit Message Likhne ka Rule

| Kaam                       | Commit Message               |
| -------------------------- | ---------------------------- |
| Nayi file add ki           | `Add Python notes`           |
| Existing file update ki    | `Update DSA notes`           |
| Bug fix kiya               | `Fix login issue`            |
| New feature add ki         | `Add calculator project`     |
| Documentation likhi        | `Add project documentation`  |
| Folder structure change ki | `Refactor project structure` |

### Kaise Check Karein Ki GitHub Se Connected Hai?

```bash
git remote -v
```

Agar output me GitHub URL aa jaye to repository connect ho chuki hai.

### First Time Setup

Agar `origin` add nahi hai:

```bash
git remote add origin https://github.com/<username>/AI-Mastery.git
git push -u origin main
```

### Baad Me Har Roz

Sirf ye 3 commands:

```bash
git add .
git commit -m "Update notes"
git push
```

`origin main` baar-baar likhne ki zarurat nahi hoti jab ek baar `git push -u origin main` chal chuka ho.

Abhi tumhare screenshot ke hisab se next step ye check karna hai:

```bash
git remote -v
```

Aur output bhej do. Agar `origin` configured nahi hai to usse setup karwa deta hoon. 🚀
